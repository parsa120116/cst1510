## Warm-up 1 · `01_index_error.py`

```python
labels = ["R-001", "R-002", "R-003"]
print(labels[3])
```
```
IndexError: list index out of range
```

**What goes wrong:** list positions (**indexes**) start at **0**. Three items have indexes 0, 1, 2. Index 3 doesn't exist.

**The fix:** to get the last item, use index 2, or better, `-1` (which always means "the last one").

````python
labels = ["R-001", "R-002", "R-003"]

print(labels[-1])
````

> 💡 **Explanation**
>
> Rule: the last valid index is always `len(list) - 1`. `-1` is safer because it works whatever the length.

```python
record = {"label": "R-001", "value": 87.0, "limit": 100.0}

```
```
KeyError: 'status'
```

**What goes wrong:** a dictionary finds values by **key** (name). There is no key `"status"` in this dict.

**Two fixes:** ask for a key that exists, or use `.get()` which returns a default instead of crashing.

````python
record = {"label": "R-001", "value": 87.0, "limit": 100.0}

print(record["label"])
print(record.get("status", "no status yet"))
````

> 💡 **Explanation**
>
> `record["key"]` → crashes if the key is missing (good when it **must** exist). `record.get("key", default)` → returns the default (good when missing is normal).

# Drills D1–D6 · lists, tuples, strings

# Topic 1 · Lists

A **list** is an ordered row of values in `[ ]`, separated by commas. Each item has a position number, starting at **0**.

## D1 · Create a list and index into it

````python
labels = ["R-001", "R-002", "R-003"]

print(labels[0])        # first item
print(labels[-1])       # last item (negative index counts from the end)
print(len(labels))      # how many items
````

> 💡 **Explanation**
>
> ```
> labels:   "R-001"  "R-002"  "R-003"
> index:       0        1        2
> negative:   -3       -2       -1
> ```
> `len()` counts items: 3. The last positive index is always `len - 1`.

## D2 · Slicing

````python
values = [5, 10, 15, 20, 25, 30]

print(values[1:5])   # the middle four
print(values[:2])    # the first two
print(values[3:])    # from position 3 to the end
````

> 💡 **Explanation**
>
> `list[start:stop]` = from `start` up to **but not including** `stop` (same rule as `range`).
>
> | Slice | Positions taken | Result |
> |---|---|---|
> | `[1:5]` | 1, 2, 3, 4 | `[10, 15, 20, 25]` |
> | `[:2]` | 0, 1 (empty start = from the beginning) | `[5, 10]` |
> | `[3:]` | 3 to the end (empty stop = to the end) | `[20, 25, 30]` |
>
> Quick check: `stop - start` = number of items (5 − 1 = 4).

## D3 · Change a list, then loop over it

````python
values = [10, 20, 30]

values[0] = 99          # replace the item at position 0
print(values)

values.append(40)       # add to the end
print(values)

values.remove(20)       # remove the first item equal to 20
print(values)

for value in values:    # visit each item in turn
    print(value)
````

> 💡 **Explanation**
>
> Lists are **mutable** (changeable). `values[0] = 99` replaces by **position**; `.remove(20)` deletes by **value** (the first 20 it finds; `ValueError` if there's none). `for value in values:` gives you each item directly, no index needed.

## D4 · Tuples cannot be changed

````python
point = (3, 7)
print(point[0])

# point[0] = 5
# TypeError: 'tuple' object does not support item assignment
# A tuple is a list that cannot be changed after it is created (immutable),
# so you cannot replace one of its items.
````

> 💡 **Explanation**
>
> A **tuple** uses `( )` instead of `[ ]`. You can read it (`point[0]`) but never change it. Use tuples for data that must stay fixed, like a coordinate `(x, y)`. If you remove the `#` from the `point[0] = 5` line and run it, you'll see the error.

## D5 · Strings index and slice the same way

````python
record_id = "R-001"

print(record_id[0])     # first character
print(record_id[-1])    # last character
print(record_id[2:])    # from position 2 to the end
````

> 💡 **Explanation**
>
> A string is like a tuple of characters: `R`(0) `-`(1) `0`(2) `0`(3) `1`(4). So `[2:]` → `"001"`. Like tuples, strings are **immutable**: `record_id[0] = "X"` is a `TypeError`.

## D6 · List comprehensions

````python
values = [40, 55, 70, 25]

doubled = [v * 2 for v in values]
print(doubled)

big = [v for v in values if v >= 50]
print(big)
````

> 💡 **Explanation**
>
> A **list comprehension** builds a new list in one line. Read it as: **[what to keep  for each item  (if condition)]**.
>
> `[v * 2 for v in values]` = "for each `v` in values, keep `v * 2`". The long version:
> ```python
> doubled = []
> for v in values:
>     doubled.append(v * 2)
> ```
> `[v for v in values if v >= 50]` keeps `v` **only** when the condition is True → `[55, 70]`.

# Drills D7–D11 · dictionaries and sets

# Topic 2 · Dictionaries and sets

A **dictionary** (`dict`) stores **key: value** pairs in `{ }`. You find a value by its **key** (name), not by position.

## D7 · Create, read, add, update

````python
record = {"label": "R-001", "value": 87.0, "limit": 100.0}

print(record["label"])      # read by key

record["status"] = "OK"     # new key -> added
record["value"] = 90.0      # existing key -> updated

print(record)
````

> 💡 **Explanation**
>
> The same syntax `record[key] = value` **adds** if the key is new and **updates** if it exists. New keys appear at the end when you print. Each key can appear only once.

## D8 · Missing keys and `.get()`

````python
print(record.get("status"))
print(record.get("owner"))
print(record.get("owner", "n/a"))
````

> 💡 **Explanation**
>
> `.get(key)` returns `None` when the key is missing (no crash); `.get(key, default)` returns your default. `record["owner"]` would be a `KeyError` (Warm-up 2).

## D9 · Loop over a dict

````python
# a) keys only
for key in record:
    print(key)

# b) values only
for value in record.values():
    print(value)

# c) both
for key, value in record.items():
    print(f"{key}: {value}")
````

> 💡 **Explanation**
>
> Looping over a dict directly gives the **keys**. `.values()` gives the values. `.items()` gives **pairs** `(key, value)`, and `key, value` unpacks each pair into two names (the same unpacking as Week 3's `return difference, percent`).

## D10 · Build a dict from two lists with `zip()`

````python
hostnames = ["srv-01", "srv-02", "srv-03"]
used = [87, 45, 120]

usage = dict(zip(hostnames, used))
print(usage)
print(usage["srv-03"])
````

> 💡 **Explanation**
>
> `zip()` pairs the two lists item by item: `("srv-01", 87)`, `("srv-02", 45)`, `("srv-03", 120)`. `dict()` turns each pair into `key: value`. Careful: if the lists have different lengths, the extras are silently dropped.

## D11 · Sets: unique values and quick checks

````python
ips = ["10.0.0.5", "10.0.0.9", "10.0.0.5", "10.0.0.7"]

unique_ips = set(ips)
print(len(unique_ips))
print("10.0.0.9" in unique_ips)
print("10.0.0.1" in unique_ips)
````

> 💡 **Explanation**
>
> A **set** keeps each value **once** and has no order. `set(ips)` removes the duplicate `10.0.0.5` → 3 unique IPs. `in` asks "is it inside?" → `True`/`False` (works on lists too, but sets are much faster for big data). Real use: *how many different IP addresses tried to log in?*

# Topic 3 · Files

## D12 · Write a file, then read it back

````python
with open("notes.txt", "w") as f:
    f.write("R-001\n")
    f.write("R-002\n")

with open("notes.txt") as f:
    print(f.read())
````

> 💡 **Explanation**
>
> - `open("notes.txt", "w")`: open for **writing** (creates the file, or **empties** it if it exists). Modes: `"r"` read (default), `"w"` write, `"a"` append (add at the end).
> - `with ... as f:` opens the file, gives it the name `f`, and **closes it automatically** when the indented block ends.
> - `f.write()` does **not** add a new line, so you add `\n` yourself.
> - The second block opens it for reading; `f.read()` returns the whole content as one string.

## D13 · Read line by line and split

````python
with open("records_sample.txt") as f:
    for line in f:
        label, value, limit = line.strip().split(",")
        value = float(value)
        limit = float(limit)
        print(label, value, limit)
````

> 💡 **Explanation**
>
> Trace one line, `"R-002,120,100\n"`:
>
> | Step | Result |
> |---|---|
> | `for line in f:` gives | `"R-002,120,100\n"` |
> | `.strip()` removes spaces/`\n` at both ends | `"R-002,120,100"` |
> | `.split(",")` cuts at each comma | `["R-002", "120", "100"]` |
> | unpack into 3 names | `label="R-002"`, `value="120"`, `limit="100"` |
> | `float()` | `120.0`, `100.0` |
>
> Without `.strip()`, the last field would be `"100\n"` (`float` copes with that, but text comparisons wouldn't). Without `float()`, the numbers would still be **text**.

# Topic 4 · Handling errors

## D14 · Catch a `ValueError`

````python
text = "abc"

try:
    number = float(text)
    print(number)
except ValueError:
    print(f"Not a number: {text}")
````

````python
# Your turn: try it yourself here
text = "42"

try:
    number = float(text)
    print(number)
except ValueError:
    print(f"Not a number: {text}")
````

> 💡 **Explanation**
>
> **How `try`/`except` runs:** Python runs the `try` block. If an error of the named type happens, it **jumps** to the `except` block instead of crashing, then continues after it. If no error happens, the `except` block is skipped. With `"abc"` → `Not a number: abc`; with `"42"` → `42.0`.

## D15 · Catch a `FileNotFoundError`

````python
try:
    with open("missing.txt") as f:
        print(f.read())
except FileNotFoundError:
    print("missing.txt not found")
````

> 💡 **Explanation**
>
> Always **name** the error you expect. A bare `except:` would also hide real bugs (like a typo in a variable name), and you'd never know.

## D16 · A list of dicts

````python
records = [
    {"label": "R-001", "value": 87.0, "limit": 100.0},
    {"label": "R-002", "value": 120.0, "limit": 100.0},
    {"label": "R-003", "value": 45.0, "limit": 100.0},
]

for record in records:
    if record["value"] > record["limit"]:
        status = "OVER LIMIT"
    else:
        status = "OK"
    print(f"{record['label']}: {status}")
````

> 💡 **Explanation**
>
> **A list of dicts = a table.** The list holds the rows in order; each dict is one row; the keys are the column names. `for record in records:` gives one dict per round, then `record["value"]` reads a field from it.
>
> Inside the f-string I wrote `record['label']` with **single** quotes, because the f-string itself uses double quotes: mixing them up would end the string early.

## C1 · Load records safely from a file

````python
def load_records(path):
    """Read label,value,limit lines from a file and return a list of dicts."""
    records = []
    try:
        with open(path) as f:
            for line in f:
                label, value, limit = line.strip().split(",")
                records.append({"label": label, "value": float(value), "limit": float(limit)})
    except FileNotFoundError:
        print(f"Could not find {path}, starting with no records.")
    return records


records = load_records("records_sample.txt")
for record in records:
    status = "OVER LIMIT" if record["value"] > record["limit"] else "OK"
    print(f"{record['label']}: {status}")

print(load_records("nope.txt"))
````

> 💡 **Explanation**
>
> **Line by line**
> - `records = []`: start with an empty list; add one dict per line.
> - The whole `with` block is inside `try`, so a missing file jumps to `except`.
> - `records.append({...})`: build a dict from the split pieces (numbers converted) and add it.
> - `return records` is **after** the `try`/`except`, so it runs in both cases: the full list, or `[]` if the file was missing.
> - Test 2 prints the clear message, then `[]`.
>
> The function only **loads**; it doesn't decide any status. That keeps it reusable: the mini-project uses this same function.

## C2 · Write the results to a file

````python
with open("results.txt", "w") as f:
    for record in records:
        status = "OVER LIMIT" if record["value"] > record["limit"] else "OK"
        f.write(f"{record['label']},{status}\n")

with open("results.txt") as f:
    print(f.read())
````

> 💡 **Explanation**
>
> One `with` for writing, a loop inside it writing one line per record (`label,status` + `\n`), then a second `with` to read it back and check. The format `label,status` is the same comma-separated shape as the input file, so another program could load it later (that's how CSV files work).

# Mini-project · Record Check, version 4: many records

Same IT lane, but now the program handles **many records at once**: first from a list in the code, then from **your own data file**, then writing results to a new file.

The template asks to bring back two functions from Week 3, slightly changed: `check_record(value, limit)` now returns **three** things (difference, percent, status), and `print_report(...)` does all the printing.

## Threshold · a list of dicts in the code + the two functions

````python
def check_record(value, limit):
    """Return the space left, the percentage used and the status."""
    difference = limit - value
    percent = value / limit * 100
    if percent >= 100:
        status = "OVER LIMIT"
    elif percent >= 90:
        status = "WARNING"
    else:
        status = "OK"
    return difference, percent, status


def print_report(label, value, limit, difference, percent, status):
    """Print the bordered report for one record."""
    print()
    print("=" * 34)
    print(f"  RECORD CHECK  -  {label}")
    print("=" * 34)
    print(f"  Used        : {value:>10.2f}")
    print(f"  Total       : {limit:>10.2f}")
    print(f"  Free        : {difference:>10.2f}")
    print(f"  Percent     : {percent:>10.2f} %")
    print(f"  Status      : {status:>10}")
    print("=" * 34)


records = [
    {"label": "srv-01", "value": 87.0, "limit": 120.0},
    {"label": "srv-02", "value": 125.0, "limit": 120.0},
    {"label": "srv-03", "value": 110.0, "limit": 120.0},
]

for record in records:
    difference, percent, status = check_record(record["value"], record["limit"])
    print_report(record["label"], record["value"], record["limit"], difference, percent, status)
````

> 💡 **Explanation**
>
> **What's new compared to Week 3**
> - No more `input()`: the data is a **list of dicts** (Topic 5), at least 3 records.
> - `check_record` returns 3 values; unpack them into 3 names.
> - One `for` loop runs check + report for every record. Adding a 4th server = adding one line to the list; the rest of the code doesn't change.
> - The status logic still lives in **one** place (`check_record`).

## Excellent · results file + OVER LIMIT count + `try`/`except` around the load

This is the complete program, as it would be in `template.py`:

````python
"""
RECORD CHECK  -  my version
Name  : Parsa
Lane  : IT
Run it:   python template.py
Data  :   servers.txt  (one record per line: hostname,GB used,GB total)
"""

# ================================================================ FUNCTIONS
def check_record(value, limit):
    """Return the space left, the percentage used and the status."""
    difference = limit - value
    percent = value / limit * 100
    if percent >= 100:
        status = "OVER LIMIT"
    elif percent >= 90:
        status = "WARNING"
    else:
        status = "OK"
    return difference, percent, status


def print_report(label, value, limit, difference, percent, status):
    """Print the bordered report for one record."""
    print()
    print("=" * 34)
    print(f"  RECORD CHECK  -  {label}")
    print("=" * 34)
    print(f"  Used        : {value:>10.2f}")
    print(f"  Total       : {limit:>10.2f}")
    print(f"  Free        : {difference:>10.2f}")
    print(f"  Percent     : {percent:>10.2f} %")
    print(f"  Status      : {status:>10}")
    print("=" * 34)


def load_records(path):
    """Read label,value,limit lines from a file and return a list of dicts."""
    records = []
    with open(path) as f:
        for line in f:
            line = line.strip()
            if line == "":
                continue                      # skip empty lines
            label, value, limit = line.split(",")
            records.append({"label": label, "value": float(value), "limit": float(limit)})
    return records


# ===================================================================== STORE
try:
    records = load_records("servers.txt")
except FileNotFoundError:
    print("servers.txt not found - starting with no records.")
    records = []

# ========================================================== PROCESS + OUTPUT
over_limit_count = 0

with open("results.txt", "w") as results_file:
    for record in records:
        difference, percent, status = check_record(record["value"], record["limit"])
        print_report(record["label"], record["value"], record["limit"],
                     difference, percent, status)
        results_file.write(f"{record['label']},{status}\n")
        if status == "OVER LIMIT":
            over_limit_count += 1

print(f"\n{over_limit_count} of {len(records)} record(s) OVER LIMIT. Results saved to results.txt")
````

> 💡 **Explanation**
>
> **What's new**
> - `try: records = load_records(...)` / `except FileNotFoundError:` → a missing file prints a clear message and continues with `[]` (Topic 4). The `try` is **around the call**, as the brief says, so `load_records` stays simple.
> - `with open("results.txt", "w") as results_file:` wraps the loop, so each record's `label,status` line is written while it's processed (C2).
> - `over_limit_count` starts before the loop, goes up when the returned status is `OVER LIMIT`, and is printed **once** after the loop.
> - I added a small safety: `if line == "": continue` skips empty lines (an empty last line in a data file is very common and would crash the `split`).
>
> **Check by hand** (the brief asks): srv-01 87/120 = 72.50% OK · srv-02 125/120 = 104.17% OVER LIMIT · srv-03 110/120 = 91.67% WARNING · srv-04 40/120 = 33.33% OK → **1 of 4 OVER LIMIT**.
