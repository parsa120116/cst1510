## Banker Roulette
Store a list of names in `friends` and pick one at random, both with `random.choice(friends)` and with `random.randint()` as an index.

```python
import random

friends = ["Alice", "Bob", "Charlie", "Diana", "Emma"]

# Way 1: let random pick an item directly
print(random.choice(friends))

# Way 2: pick a random index, then use it
index = random.randint(0, len(friends) - 1)
print(friends[index])
```

> 💡 **Explanation**
>
> - Way 1: `random.choice(list)` directly returns a random item. This is the simplest way.
> - Way 2: the indexes go from 0 to `len - 1` (here 0 to 4). `randint(0, len(friends) - 1)` makes a valid index, and `friends[index]` takes that item.
>
> **Trap:** `randint(0, len(friends))` is wrong, because `randint` includes the last number too. So sometimes it gives 5 and you get an `IndexError` (this is warm-up 1 from this same week!).

## Password Generator
Ask how many letters, symbols and numbers. Keep allowed letters, symbols and numbers in lists. Easy: build the password in order with `random.choice()` in loops. Hard: collect the characters in a list, `random.shuffle()` it, then join.

```python
import random

letters = list("abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ")
numbers = list("0123456789")
symbols = list("!#$%&()*+")

print("Welcome to the PyPassword Generator!")
n_letters = int(input("How many letters would you like in your password? "))
n_symbols = int(input("How many symbols would you like? "))
n_numbers = int(input("How many numbers would you like? "))

# Easy version: in order
easy = ""
for _ in range(n_letters):
    easy += random.choice(letters)
for _ in range(n_symbols):
    easy += random.choice(symbols)
for _ in range(n_numbers):
    easy += random.choice(numbers)
print(f"Easy password: {easy}")

# Hard version: shuffled
chars = []
for _ in range(n_letters):
    chars.append(random.choice(letters))
for _ in range(n_symbols):
    chars.append(random.choice(symbols))
for _ in range(n_numbers):
    chars.append(random.choice(numbers))
random.shuffle(chars)
hard = "".join(chars)
print(f"Hard password: {hard}")
```

> 💡 **Explanation**
>
> **Line by line**
> - `list("abc...")` turns a string into a list of characters: `['a', 'b', 'c', ...]`. This way you don't have to type all 52 letters one by one with commas.
> - `for _ in range(n_letters):` means "repeat n times". `_` is the name for a variable we don't need (only the number of repeats matters).
> - Easy version: `easy += random.choice(letters)` adds one character to the end of the string each time. Its problem is that the pattern is easy to guess (letters first, numbers last).
> - Hard version: we collect the characters in a **list**, because you can't shuffle a string but you can shuffle a list. `random.shuffle(chars)` shuffles the list **in place** and returns nothing. Then `"".join(chars)` joins the items of the list with no spaces and makes a string.
>
> **Trap:** `chars = random.shuffle(chars)` is wrong, because `shuffle` returns `None` (the `print` vs `return` lesson from Week 3).

## Blind Auction
Ask each bidder for name and bid, store them as `{name: bid}`, ask if there's another bidder, then loop over `.items()` to find the highest and print only the winner.

```python
bids = {}

more_bidders = True
while more_bidders:
    name = input("What is your name? ")
    bid = float(input("What is your bid? $"))
    bids[name] = bid
    answer = input("Are there any other bidders? Type 'yes' or 'no': ").lower()
    if answer == "no":
        more_bidders = False

highest_bid = 0
winner = ""
for bidder, amount in bids.items():
    if amount > highest_bid:
        highest_bid = amount
        winner = bidder

print(f"The winner is {winner} with a bid of ${highest_bid:.2f}")
```

> 💡 **Explanation**
>
> **Two parts:**
> 1. **Collecting:** `bids[name] = bid` makes the name the key and the bid the value (Week 4, D7). The loop keeps going until someone says `no`.
> 2. **Finding the biggest value:** the classic "biggest so far" pattern. We start with `highest_bid = 0`. For each bid, if it is bigger than the biggest so far, we replace both the amount and its owner. At the end of the loop, `highest_bid` is the biggest number.
>
> `for bidder, amount in bids.items():` unpacks each key–value pair (D9).
>
> **A shorter way:** `winner = max(bids, key=bids.get)`. But the loop version shows what happens behind the scenes, and this pattern comes up a lot in exams and interviews.
>
> **Note:** if two people make the same bid, the first one wins, because `>` (not `>=`) only replaces with a bigger number.

## Coffee Machine
Resources dict `{"water": 300, "milk": 200, "coffee": 100}`; a `MENU` dict of dicts with ingredients and cost. Ask for a drink, check every needed resource with a function (print what's missing), deduct ingredients, take payment, confirm. Also support `report` and `off`.

```python
MENU = {
    "espresso": {"ingredients": {"water": 50, "coffee": 18}, "cost": 1.5},
    "latte": {"ingredients": {"water": 200, "milk": 150, "coffee": 24}, "cost": 2.5},
    "cappuccino": {"ingredients": {"water": 250, "milk": 100, "coffee": 24}, "cost": 3.0},
}

resources = {"water": 300, "milk": 200, "coffee": 100}
money = 0.0


def report(resources, money):
    """Print the machine's resources and money."""
    print(f"Water: {resources['water']}ml")
    print(f"Milk: {resources['milk']}ml")
    print(f"Coffee: {resources['coffee']}g")
    print(f"Money: ${money:.2f}")


def missing_ingredients(ingredients, resources):
    """Return a list of the ingredients there is not enough of."""
    missing = []
    for item, amount in ingredients.items():
        if resources.get(item, 0) < amount:
            missing.append(item)
    return missing


def make_coffee(drink_name, ingredients, resources):
    """Take the drink's ingredients out of the resources."""
    for item, amount in ingredients.items():
        resources[item] -= amount
    print(f"Here is your {drink_name} ☕️. Enjoy!")


machine_on = True
while machine_on:
    choice = input("What would you like? (espresso/latte/cappuccino): ").lower()
    if choice == "off":
        machine_on = False
    elif choice == "report":
        report(resources, money)
    elif choice in MENU:
        drink = MENU[choice]
        missing = missing_ingredients(drink["ingredients"], resources)
        if missing:
            print(f"Sorry, there is not enough {', '.join(missing)}.")
        else:
            paid = float(input(f"That's ${drink['cost']:.2f}. How much do you insert? $"))
            if paid < drink["cost"]:
                print("Sorry, that's not enough money. Money refunded.")
            else:
                change = paid - drink["cost"]
                money += drink["cost"]
                if change > 0:
                    print(f"Here is ${change:.2f} in change.")
                make_coffee(choice, drink["ingredients"], resources)
    else:
        print("Sorry, we don't have that.")
```

> 💡 **Explanation**
>
> **This is the biggest extra project so far. Read it piece by piece:**
>
> **1. The data:** `MENU["latte"]["ingredients"]["milk"]` → `150`. That means keys one after another: first the drink, then the section, then the ingredient. It's like opening a box that has another box inside it.
>
> **2. The functions, each with one job:**
> - `report`: only prints.
> - `missing_ingredients`: only **checks**, and returns a list of the ingredients that are short. `resources.get(item, 0)` counts an ingredient as zero if the machine doesn't have it at all, so it doesn't crash.
> - `make_coffee`: takes the ingredients out. Because a dictionary is **mutable**, a change we make to `resources` inside the function stays outside too. It wasn't like this with numbers (Week 3, D10)!
>
> **3. The main loop:** three special commands (`off`, `report`, a drink name) with `if/elif/else`.
> - `if missing:` an empty list counts as `False` and a non-empty list counts as `True`. So this means "if anything is short".
> - `', '.join(missing)` turns the list into text: `['milk', 'coffee']` → `milk, coffee`.
> - Money: if they pay too little, we give it back; if they pay too much, we give back the change and add the price to `money`.
>
> **Try it:** buy `latte` twice (each one uses 200 ml of water and 150 ml of milk). The second time it says `Sorry, there is not enough water, milk.`, because only 300 − 200 = 100 ml of water and 200 − 150 = 50 ml of milk are left. Then type `report`, and then `off`.
