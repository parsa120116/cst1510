```

```

````python
import pandas as pd

df = pd.DataFrame({"label": ["R-001", "R-002"], "value": [87, 120], "limit": [100, 100]})

print(df["value"])
````

> 💡 **Explanation**
>
> **How do we read the error?**
> - A pandas traceback is very long and full of the names of pandas' own internal files (like `pandas/core/indexes/base.py`). **Ignore those.** Two places matter:
>   1. **The last line:** `KeyError: 'Value'`. It means "I couldn't find a key called `'Value'`". In a DataFrame, the "key" is the **column name**.
>   2. **The line from your own file:** `line 8` → `print(df["Value"])`.
> - So the question is: do we have a column called `Value`? No. Our column is `value` with a **small v**.
>
> **Fixed:** just `"Value"` → `"value"`. Column names, like variable names, are **case-sensitive**. To Python, `Value` and `value` are two completely different names.
>
> **Real output:**
> ```
> 0     87
> 1    120
> Name: value, dtype: int64
> ```
> - The left column (`0`, `1`) is the **index**, which means the row numbers.
> - `Name: value` is the name of this column, and `dtype: int64` means all the values are whole numbers (integers).
>
> **Remember:** in pandas, `KeyError` almost always means "you didn't write the column name correctly". First look at the real names with `print(df.columns)`. You saw the same error with the Week 4 dictionaries too (`my_dict["wrong_key"]`), because a DataFrame also works with column names, like a dictionary works with keys.

## Warm-up 3 · `03_type_error.py`

```python
import pandas as pd

df = pd.DataFrame({"value": ["87", "120", "45"], "limit": [100, 100, 100]})

df["percent"] = (df["value"] / df["limit"]) * 100
```
```
Traceback (most recent call last):
  File "03_type_error.py", line 8, in <module>
    df["percent"] = (df["value"] / df["limit"]) * 100
  ...
TypeError: unsupported operand type(s) for /: 'str' and 'int'
```

````python
import pandas as pd

df = pd.DataFrame({"value": ["87", "120", "45"], "limit": [100, 100, 100]})
print(df.dtypes)

df["value"] = df["value"].astype(int)     # turn the text "87" into the number 87
df["percent"] = (df["value"] / df["limit"]) * 100

print(df)
````

> 💡 **Explanation**
>
> **How do we read the error?**
> - Last line: `TypeError: unsupported operand type(s) for /: 'str' and 'int'`. It means "the `/` operator is not defined between a `str` (text) and an `int` (number)".
> - This is exactly the Week 1 error: `"87" / 100` doesn't work in plain Python either. You can't divide text.
> - The cause: in the dictionary we wrote `["87", "120", "45"]`, **with quotes**. So pandas kept them as text.
>
> **How can we tell if a column is text or numbers?** With `df.dtypes` (the lab's table says this too). I put it at the start of the program so you can see it:
> ```
> value      str
> limit    int64
> dtype: object
> ```
> - `value` → `str` (on pandas 2 you see `object` instead). Both mean **text**.
> - `limit` → `int64`, which means whole numbers.
> - The `dtype: object` on the last line belongs to this report itself, not to the columns. Don't worry about it.
>
> **The fix, line by line**
> - `df["value"].astype(int)`: turns the whole column into integers in one go. It's the same job as `int("87")` from Week 1, but on **all** the rows together, with no loop. `astype` means "as this type".
> - `df["value"] = ...`: we put the result **back into the same column**. If you only write `df["value"].astype(int)` and don't store it anywhere, `df` doesn't change (like `text.upper()` in Week 4, which didn't change the string itself).
> - Now the division works.
>
> **Real output (after the dtypes table):**
> ```
>    value  limit  percent
> 0     87    100     87.0
> 1    120    100    120.0
> 2     45    100     45.0
> ```
>
> **A second way (simpler in this file):** remove the quotes: `"value": [87, 120, 45]`. But in the real world, data usually comes from a file and you can't fix it by hand, which is why I showed `astype`.
>
> **Trap:** if one of the texts is not a number (for example `"n/a"`), `astype(int)` itself gives a `ValueError`. That means the data is dirty and must be cleaned first.

## D1 · A Series

*Topic 1 · Series and DataFrame*

1. Run `import pandas as pd`.
2. Create `values = pd.Series([87, 120, 45])` and print it.
3. Print `values.max()`.

**Expected output:**
```
0     87
1    120
2     45
dtype: int64
120
```

````python
import pandas as pd

values = pd.Series([87, 120, 45])
print(values)

print(values.max())
````

> 💡 **Explanation**
>
> **Line by line**
> - `import pandas as pd`: imports the pandas library and gives it a short nickname (alias): `pd`. From now on we write `pd.Series` instead of `pandas.Series`. This `as pd` is a convention that **everyone** follows, so you should always write it too.
> - `pd.Series([87, 120, 45])`: takes a normal Python list and makes a **Series**. A Series is one **column** of data, where each value has a label next to it called the **index**.
> - `print(values)` shows three things:
>   - The left column (`0`, `1`, `2`) is the index. Because we didn't give an index ourselves, pandas numbered from 0 (like list indexes).
>   - The right column is the values themselves.
>   - `dtype: int64` is the data type of this column: a 64-bit integer. "64-bit" only means there is enough room for very big numbers; for us it's just an `int`.
> - `values.max()`: returns the biggest value → `120`. This is a **method**, so it needs brackets.
>
> **Why does this matter? (no loop)** In Week 2, to find the biggest number we wrote this:
> ```python
> biggest = numbers[0]
> for n in numbers:
>     if n > biggest:
>         biggest = n
> ```
> Here it's just `values.max()`. Similar methods: `.min()`, `.mean()` (average), `.sum()` (total). All of them work on the whole column at once.
>
> **Series vs list:** a list only has values. A Series has values + an index + a data type (dtype), and you can do maths on it (for example, `values * 2` doubles every value; with a list, `[87, 120, 45] * 2` **repeats** the list!).
>
> **Trap:** `pd.series` with a small s → `AttributeError`. The name is `Series` with a capital S (in the same way, `DataFrame` has a capital D and F).

## D2 · A DataFrame from a dict

1. Create a DataFrame from a dict of lists, with three columns:
   `label` (`"R-001"`, `"R-002"`, `"R-003"`), `value` (`87`, `120`, `45`) and `limit` (`100`, `100`, `100`).
2. Print it, then print its `.shape`.

**Expected output:** a table with 3 rows and 3 columns, then `(3, 3)`

````python
import pandas as pd

data = {
    "label": ["R-001", "R-002", "R-003"],
    "value": [87, 120, 45],
    "limit": [100, 100, 100],
}

df = pd.DataFrame(data)
print(df)
print(df.shape)
````

> 💡 **Explanation**
>
> **Line by line**
> - `data = {...}`: a normal dictionary from Week 4. The **keys** (`"label"`, `"value"`, `"limit"`) become the **column names**, and each **list** becomes the content of that column, from top to bottom.
> - `pd.DataFrame(data)`: builds a table from the dictionary. The name `df` (short for DataFrame) is also a common convention.
> - The first row of the table = the first item of each list: `R-001`, `87`, `100`. In other words, the lists are read "vertically".
> - `df.shape`: the size of the table, as a **tuple**: `(number of rows, number of columns)`. The index is **not** counted as a column.
>
> **Real output:**
> ```
>    label  value  limit
> 0  R-001     87    100
> 1  R-002    120    100
> 2  R-003     45    100
> (3, 3)
> ```
>
> **Important point: `.shape` has no brackets.** `shape` is an **attribute**, not a method. That means it's a value you read from the object (like `self.label` in the Week 5 classes), not an action you call. `.max()` was a method and did a job, so it had brackets. If you write `df.shape()` you get `TypeError: 'tuple' object is not callable`, which means "you can't call a tuple like a function".
>
> **Remember (Warm-up 2):** all the lists must be the same length, or you get `ValueError: All arrays must be of the same length`.
>
> **Link to Week 4:** in Week 4 we made each record a dictionary and had a **list of dictionaries** (each dictionary was one row). Here it's the other way round: **one dictionary of lists** (each list is one column).

## D4 · Add a column, without a loop

Add a `percent` column to `df`: `(value / limit) * 100`. Print `df`.

**Expected output:** the `percent` column holds `87.0`, `120.0`, `45.0`, `93.0`, `101.0`, `60.0`.

````python
import pandas as pd

df = pd.read_csv("records_sample.csv")

df["percent"] = (df["value"] / df["limit"]) * 100
print(df)
````

> 💡 **Explanation**
>
> **This is the most important idea of the whole week: an operation on a whole column, in one go.**
>
> **Line by line**
> - `df["value"]`: takes the `value` column as a Series (`87, 120, 45, ...`).
> - `df["value"] / df["limit"]`: divides the two Series **row by row**: row 0 by row 0, row 1 by row 1, and so on. The result is a new Series: `0.87, 1.20, 0.45, ...`.
> - `* 100`: multiplies every item by 100 → `87.0, 120.0, ...`.
> - `df["percent"] = ...`: because there is **no** column called `percent`, pandas creates a new column with this name and puts the result in it (like adding a new key to a dictionary: `d["new_key"] = value`). If the column already existed, it would replace it.
>
> **Real output:**
> ```
>    label  value  limit  percent
> 0  R-001     87    100     87.0
> 1  R-002    120    100    120.0
> 2  R-003     45    100     45.0
> 3  R-004     93    100     93.0
> 4  R-005    101    100    101.0
> 5  R-006     60    100     60.0
> ```
>
> **Why `87.0` and not `87`?** Because in Python `/` **always** returns a float (a decimal number); even `200 / 100` gives `2.0`. pandas has the same rule, so the `percent` column has the type `float64`.
>
> **What does "without a loop" mean?** In Week 4 it was like this:
> ```python
> for record in records:
>     record["percent"] = (record["value"] / record["limit"]) * 100
> ```
> Here we don't write a loop ourselves; pandas does the calculation on all the rows by itself (and much faster, with code written in the C language). This is called **vectorised**: one operation on a whole column. With 6 rows you won't feel any difference, but with a million rows a Python loop is much slower.
>
> **Hand check:** `R-005`: `101 / 100 * 100 = 101.0` ✓
>
> **Trap:** `df.percent = ...` (with a dot) does **not** create a new column; it only gives a warning. To create a column, always use square brackets: `df["percent"]`.

## D5 · `iloc` and `loc`

*Topic 2 · Selecting and filtering*

1. Print the `label` of the first row, using `df.iloc[0]["label"]`.
2. Print the `value` in the row with index `2`, using `df.loc[2, "value"]`.

**Expected output:** `R-001`, then `45`

````python
import pandas as pd

df = pd.read_csv("records_sample.csv")

print(df.iloc[0]["label"])     # by position: the first row
print(df.loc[2, "value"])      # by label: the row whose index is 2
````

> 💡 **Explanation**
>
> **Two ways to reach a row:**
> | | Means | Example |
> |---|---|---|
> | `iloc` | **i**nteger **loc**ation: by **position**, like a list index | `df.iloc[0]` = the **first** row of the table, whatever its index is |
> | `loc` | by **label**, which means the index value itself | `df.loc[2]` = the row that has `2` written next to it |
>
> **Line by line**
> - `df.iloc[0]`: takes the first row. The result is a **Series** whose index is the column names:
>   ```
>   label    R-001
>   value       87
>   limit      100
>   ```
> - `["label"]`: from that Series, takes the value next to `label` → `R-001`. So it's two steps: first the row, then the column.
> - `df.loc[2, "value"]`: says in one go "the row with label `2`, the column `value`". The order is always **row first, then column**, with a comma between them. → `45` (row `R-003`).
>
> **Real output:**
> ```
> R-001
> 45
> ```
>
> **"But these two are the same!" Not always.** Right now the index values are `0` to `5` and in order, so position and label give the same result. But after sorting or filtering (D7, D9), the index values **move with the rows**. For example, after the sort in D9, the first row of the table is `R-002`, with index `1`. Then:
> - `sorted_df.iloc[0]` → the **first** row = `R-002`
> - `sorted_df.loc[0]` → the row whose label is `0` = `R-001` (which is now in the middle of the table!)
>
> **Traps**
> - `df.iloc(0)` with round brackets → error. `iloc` and `loc` work with **square brackets** `[]`.
> - `df.iloc[0, "label"]` → error, because `iloc` only accepts **numbers** (the column must also be a number: `df.iloc[0, 0]`). Column names are for `loc`.
> - `df.loc[6, "value"]` → `KeyError: 6`, because there is no row with the label `6` (the last one is `5`).

## D7 · Filter rows with a condition

1. Keep only the rows where `percent` is 100 or more, using a condition inside `df[...]`. Print them.
2. Print how many rows that is, with `len()`.

**Expected output:** the rows for `R-002` and `R-005`, then `2`

````python
import pandas as pd

df = pd.read_csv("records_sample.csv")
df["percent"] = (df["value"] / df["limit"]) * 100

over_limit = df[df["percent"] >= 100]
print(over_limit)
print(len(over_limit))
````

> 💡 **Explanation**
>
> **This is the second very important idea of the week: the boolean filter (or boolean mask).** It has two steps; let's look at them one at a time.
>
> **Step 1: a condition on the whole column.** `df["percent"] >= 100` doesn't compare one number; it compares the **whole column**, and gives a `True` or `False` for each row. If you print it:
> ```
> 0    False
> 1     True
> 2    False
> 3    False
> 4     True
> 5    False
> Name: percent, dtype: bool
> ```
> This Series of True/False values is called a **mask**: like a stencil that only lets things through its holes (the `True` values).
>
> **Step 2: give the mask to df.** `df[mask]` means "keep only the rows that have `True` next to them". So `df[df["percent"] >= 100]` = "the rows whose percent is 100 or more".
>
> - `over_limit = ...`: the result is a **new** DataFrame. `df` itself is untouched and still has 6 rows.
> - `len(over_limit)`: on a DataFrame, `len` gives the number of **rows** (like `len` on a list).
>
> **Real output:**
> ```
>    label  value  limit  percent
> 1  R-002    120    100    120.0
> 4  R-005    101    100    101.0
> 2
> ```
> - **The index values stayed `1` and `4`**; they were not renumbered (they did not become `0` and `1`). Each row takes its original label with it. This is exactly where `loc` and `iloc` become different (D5): `over_limit.iloc[0]` is the row `R-002`, but `over_limit.loc[0]` gives an error.
>
> **Why not a loop?** The Week 4 version:
> ```python
> over_limit = []
> for record in records:
>     if record["percent"] >= 100:
>         over_limit.append(record)
> ```
> Here all of those 4 lines became one expression inside the brackets.
>
> **Traps**
> - `df["percent" >= 100]` (bracket closed in the wrong place) → `TypeError`, because it's comparing the text `"percent"` with the number 100. The condition must be on the **column**: `df["percent"] >= 100`.
> - `>` is different from `>=`: "100 or more" means `>=`. If a record were exactly 100, `>` would miss it.

## D8 · Combine conditions with `&` and `|`

1. Print the rows where `percent` is 90 or more **and** less than 100.
2. Print the rows where `percent` is less than 50 **or** more than 110.

*Put brackets round each condition, and use `&` and `|`, not `and` / `or`.*

**Expected output:** first only `R-004`, then `R-002` and `R-003`

````python
import pandas as pd

df = pd.read_csv("records_sample.csv")
df["percent"] = (df["value"] / df["limit"]) * 100

warning_rows = df[(df["percent"] >= 90) & (df["percent"] < 100)]
print(warning_rows)

far_rows = df[(df["percent"] < 50) | (df["percent"] > 110)]
print(far_rows)
````

> 💡 **Explanation**
>
> **Line by line**
> - `(df["percent"] >= 90)`: the first mask (D7): True/False for each row.
> - `(df["percent"] < 100)`: the second mask.
> - `&` ("and"): combines the two masks **row by row**. A row is `True` only when it is `True` in **both** masks.
> - `|` ("or", a vertical line): a row is `True` when **at least one** of the two masks is `True`.
>
> **Hand-check table for the first condition** (`>= 90` & `< 100`):
> | label | percent | `>= 90` | `< 100` | `&` |
> |---|---|---|---|---|
> | R-001 | 87 | False | True | False |
> | R-002 | 120 | True | False | False |
> | R-003 | 45 | False | True | False |
> | R-004 | 93 | True | True | **True** |
> | R-005 | 101 | True | False | False |
> | R-006 | 60 | False | True | False |
>
> **Real output:**
> ```
>    label  value  limit  percent
> 3  R-004     93    100     93.0
>    label  value  limit  percent
> 1  R-002    120    100    120.0
> 2  R-003     45    100     45.0
> ```
>
> **The big question: why don't `and` / `or` work?**
> Python's `and` and `or` want **one single** True or False on each side. When you give them a Series with 6 True/False values, Python asks pandas: "Is this whole column True or False?". pandas can't answer (do you mean "all of them are True"? or "at least one is"?), so it gives an error:
> ```
> ValueError: The truth value of a Series is ambiguous. Use a.empty, a.bool(), a.item(), a.any() or a.all().
> ```
> **Ambiguous** means "unclear, it could mean two things". This is the `ValueError` the course README talks about. `&` and `|`, on the other hand, work **item by item** and give a new mask, which is exactly what we want.
>
> **Why brackets round each condition?**
> In Python, `&` and `|` run **before** `>=` and `<` (they have higher priority, or precedence, just like `*` is done before `+`). So without brackets:
> ```python
> df["percent"] >= 90 & df["percent"] < 100
> ```
> Python first works out `90 & df["percent"]` (!), which makes no sense and gives an error (on pandas 3: `TypeError: Cannot perform 'rand_' with a dtyped [float64] array and scalar of type [bool]`; on other versions the text may be different). The brackets say "do the comparisons first, then combine the results with `&`".
>
> **Learn the rule like this:**
> ```
> df[ (condition 1) & (condition 2) ]     → and
> df[ (condition 1) | (condition 2) ]     → or
> ```
>
> **Link to Record Check:** the first condition is exactly the definition of the `WARNING` status (90 or more, but below 100). That's why I named the variable `warning_rows`.

## D10 · `groupby()`, `.agg()` and `value_counts()`

1. Add a column `over` that is `True` when `percent` is 100 or more: `df["over"] = df["percent"] >= 100`
2. Group by `over` and print the mean, max and min of `percent` for each group, using `.agg(["mean", "max", "min"])`.
3. Print `df["over"].value_counts()`.

**Expected output:**
```
         mean    max    min
over
False   71.25   93.0   45.0
True   110.50  120.0  101.0
```
then `False` 4 and `True` 2.

````python
import pandas as pd

df = pd.read_csv("records_sample.csv")
df["percent"] = (df["value"] / df["limit"]) * 100

df["over"] = df["percent"] >= 100

summary = df.groupby("over")["percent"].agg(["mean", "max", "min"])
print(summary)

print(df["over"].value_counts())
````

> 💡 **Explanation**
>
> **Line by line**
> - `df["over"] = df["percent"] >= 100`: the same mask as D7, but this time, instead of filtering, we store it **as a column**. So each row has a `True` or `False` (column type: `bool`).
> - `df.groupby("over")`: **groups** the rows by the value of `over`: one group for `False` (4 rows) and one group for `True` (2 rows). On its own it **prints nothing**; it only makes a "grouped" object that waits for you to say what to do with each group.
> - `["percent"]`: from each group, we take only the `percent` column.
> - `.agg([...])`: `agg` is short for **aggregate**, which means "summarise many values into one value". We give it a list of function names, and it makes one column for each: `mean` (average), `max`, `min`.
> - `df["over"].value_counts()`: counts how many times each value appears, sorted **from most to least**.
>
> **The big idea of groupby: "split, calculate, put back together"** (in English, split-apply-combine):
> ```
> False → 87, 45, 93, 60    → mean 71.25, max 93, min 45
> True  → 120, 101          → mean 110.5, max 120, min 101
> ```
>
> **Real output:**
> ```
>          mean    max    min
> over
> False   71.25   93.0   45.0
> True   110.50  120.0  101.0
> over
> False    4
> True     2
> Name: count, dtype: int64
> ```
> - `over` above the index means that this table's index came from the `over` column.
> - `Name: count` is the same on pandas 2 and 3 (very old versions showed `Name: over`).
>
> **Hand check of the mean:** `False`: `(87 + 45 + 93 + 60) / 4 = 285 / 4 = 71.25` ✓. `True`: `(120 + 101) / 2 = 110.5` ✓
>
> **Compared with Week 4:** to count things, you needed a counter dictionary and a loop (`counts[key] = counts.get(key, 0) + 1`). Now it's `value_counts()`.
>
> **Traps**
> - `df.groupby("over").agg(...)` without `["percent"]` → it tries to take the `mean` of **all** the columns, including `label`, which is text, and gives an error (on pandas 3: `TypeError: dtype 'str' does not support operation 'mean'`). Always say which column to summarise.
> - If you only want one statistic, you don't need `agg`: `df.groupby("over")["percent"].mean()`. (You'll see this in C2.)

## D11 · A lambda

*Topic 4 · Apply, lambda and finishing up*

1. Write `double = lambda x: x * 2` and print `double(21)`.
2. Underneath, write the same thing as a normal `def` function called `double_def`, and print `double_def(21)`.

**Expected output:** `42`, then `42`

````python
double = lambda x: x * 2
print(double(21))


def double_def(x):
    """Return x multiplied by 2."""
    return x * 2


print(double_def(21))
````

> 💡 **Explanation**
>
> **What is a lambda?** A **small function with no name** (an anonymous function) that is written on one line. These two do exactly the same job:
> ```
> lambda x: x * 2
>    │    │   └── what it returns (automatic return)
>    │    └────── the parameter(s)
>    └─────────── the keyword
> def double_def(x):
>     return x * 2
> ```
>
> **Line by line**
> - `double = lambda x: x * 2`: makes a function that takes `x` and returns `x * 2`, and puts that function in the variable `double`.
> - `double(21)`: we call it just like any other function → `42`.
> - `def double_def(x):` the same function, written the Week 3 way, with a docstring.
>
> **Real output:**
> ```
> 42
> 42
> ```
>
> **The rules of lambda:**
> - **Only one expression.** You can't write several lines, a loop, or a multi-line `if`/`else`.
> - You **don't write** the word `return`; the result of that one expression is returned automatically. (`lambda x: return x * 2` → `SyntaxError`.)
> - It has no name (that's why it's "anonymous"). Here we forced it into `double` just so we could test it.
>
> **So when lambda, and when `def`?** (one of the revision questions)
> - **lambda:** when you need a very short job **once**, **right there**, usually as the input to another function (like `.apply()` in D12). Writing it there is easier to read than defining a separate function.
> - **`def`:** when the logic is more than one expression, or you want to use it in several places, or it needs a docstring and tests, or giving it a name makes the code easier to read. The lab's rule says the same.
> - A subtle point: `double = lambda ...` (a lambda stored in a variable) is not recommended by Python's style guide (PEP 8); if you need a name, write a `def`. We only did it here for learning.

## D12 · `.apply()` with a lambda

1. Add a `status` column to `df` using `df["percent"].apply(lambda p: ...)`:
   `"OVER LIMIT"` if 100 or more, `"WARNING"` if 90 or more, otherwise `"OK"`.
2. Print just the `label` and `status` columns.
3. Print `df["status"].value_counts()`.

**Expected output:** `R-001` OK, `R-002` OVER LIMIT, `R-003` OK, `R-004` WARNING, `R-005` OVER LIMIT,
`R-006` OK; then `OK` 3, `OVER LIMIT` 2, `WARNING` 1.

````python
import pandas as pd

df = pd.read_csv("records_sample.csv")
df["percent"] = (df["value"] / df["limit"]) * 100

df["status"] = df["percent"].apply(
    lambda p: "OVER LIMIT" if p >= 100 else "WARNING" if p >= 90 else "OK"
)

print(df[["label", "status"]])
print(df["status"].value_counts())
````

> 💡 **Explanation**
>
> **What does `.apply()` do?** It takes a function and calls it on **each** value in the column, one by one, then puts all the answers into a new Series. Here: for `87.0` it calls the function → `"OK"`, for `120.0` → `"OVER LIMIT"`, and so on.
>
> **Line by line**
> - `df["percent"].apply(...)`: `apply` is called on the `percent` column, so each time `p` is one of the percents.
> - `lambda p: "OVER LIMIT" if p >= 100 else "WARNING" if p >= 90 else "OK"`: this is a chained **conditional expression**: the same `if`/`elif`/`else` from Week 2, but on one line:
>   ```python
>   if p >= 100:
>       return "OVER LIMIT"
>   elif p >= 90:
>       return "WARNING"
>   else:
>       return "OK"
>   ```
>   The one-line form is: `A if condition else B`. When B is itself another `if ... else`, you get a chain. It is read from left to right.
> - **Order matters:** check `>= 100` first. If you check `>= 90` first, `120` is also bigger than 90 and gets `"WARNING"`, and nobody would ever be `OVER LIMIT` (the same `elif` lesson from Week 2).
> - Putting the lambda on its own line inside the brackets of `apply(...)` is only for readability (the course code in DEMO is written the same way). Inside brackets, Python lets you break a line.
> - `df["status"] = ...`: the result becomes a new column (like D4).
> - `df[["label", "status"]]`: two columns, with a list inside the brackets (D6).
> - `value_counts()`: counts each status (D10).
>
> **Real output:**
> ```
>    label      status
> 0  R-001          OK
> 1  R-002  OVER LIMIT
> 2  R-003          OK
> 3  R-004     WARNING
> 4  R-005  OVER LIMIT
> 5  R-006          OK
> status
> OK            3
> OVER LIMIT    2
> WARNING       1
> Name: count, dtype: int64
> ```
>
> **"Didn't we say no loops?"** To be honest, behind the scenes `apply` really is a loop and passes each value to the function separately (that's why it's slower than vectorised operations like D4). But **you** don't write a loop, you don't need a counter, an empty list or `append`, and the result is a column straight away. For logic like this (several conditions), `apply` is the natural, readable pandas way.
>
> **The `def` version** (if the logic were longer, the lab's rule says do this):
> ```python
> def status_of(p):
>     """Return OVER LIMIT, WARNING or OK for a percent."""
>     if p >= 100:
>         return "OVER LIMIT"
>     elif p >= 90:
>         return "WARNING"
>     return "OK"
>
> df["status"] = df["percent"].apply(status_of)
> ```
> Notice: `apply(status_of)` has **no brackets** after the function name. We give the function itself so that pandas can call it; `status_of()` would mean we call it ourselves right now (and because we haven't given it a `p`, we'd get a `TypeError`).

> - pandas always puts the last line (`Name: percent, ...`) under a Series; it was left out of the lab's "expected output", but this is the answer.
> - The groups are sorted **alphabetically** by status name (`OK`, `OVER LIMIT`, `WARNING`), because `groupby` sorts the keys itself. (`value_counts`, on the other hand, sorted by **count**.)
>
> **Hand check:**
> - `OK`: R-001, R-003, R-006 → `(87 + 45 + 60) / 3 = 192 / 3 = 64.0` ✓
> - `OVER LIMIT`: R-002, R-005 → `(120 + 101) / 2 = 110.5` ✓
> - `WARNING`: only R-004 → `93.0` ✓
>
> **Why is `round(1)` needed here?** With this data, by chance, all the averages are neat. But if an average came out as, say, `64.333333`, without rounding you would see six decimal places. The question asks for one, so we keep it so that the answer is right for any data.
>
> **Trap:** put `.round(1)` **after** `.mean()`, because we want the **final answer** to be rounded. `df.groupby("status")["percent"].round(1)` doesn't work at all: `AttributeError: 'SeriesGroupBy' object has no attribute 'round'`. In other words, until you say what to do with each group (`mean`), you don't have any numbers yet to round.

## Typical · add `status` with `.apply()` and a lambda, show the rows that are not OK

Add a `status` column using `.apply()` with a lambda (`"OVER LIMIT"` for 100%+, `"WARNING"` for 90%+, otherwise `"OK"`). Print only the rows that are not `"OK"`, using a boolean condition in `[]`, not a loop.

````python
"""
RECORD CHECK  -  my version
===========================

Name  :  Parsa
Lane  :  IT
Date  :  November 2026

Run it:   python template.py

Reads servers.csv (hostname, used_gb, total_gb), works out how full each
server's disk is, gives it a status and shows the servers that need attention.
"""

import pandas as pd

# ==================================================================== LOAD
# 1. Read the data file: one row per server.

df = pd.read_csv("servers.csv")


# ================================================================= PROCESS
# 2. percent = how full each disk is: (used_gb / total_gb) * 100
#    status  = OVER LIMIT (100%+), WARNING (90%+), otherwise OK

df["percent"] = (df["used_gb"] / df["total_gb"]) * 100
df["status"] = df["percent"].apply(
    lambda p: "OVER LIMIT" if p >= 100 else "WARNING" if p >= 90 else "OK"
)


# ================================================================== OUTPUT
# 3. Threshold: print the whole DataFrame.
#    Typical  : print only the servers that are not OK (boolean filter, no loop).

print("All servers:")
print(df)

not_ok = df[df["status"] != "OK"]
print("\nServers that need attention:")
print(not_ok)


# ==========================================================================
# 4. Before you finish:
#    [x] Checked one row by hand: srv-03 -> 460 / 500 * 100 = 92.0 -> WARNING
#    [x] One condition only, so no & / | needed
#    [x] Column names match the CSV header exactly: hostname, used_gb, total_gb
````

## Excellent · count per status, save a CSV and a bar chart

Also group by `status` and print a count per group, save the full DataFrame to a new CSV with `.to_csv()`, and save one bar chart of the counts per status as a PNG.

````python
"""
RECORD CHECK  -  my version
===========================

Name  :  Parsa
Lane  :  IT
Date  :  November 2026

Run it:   python template.py

Reads servers.csv (hostname, used_gb, total_gb), works out how full each
server's disk is, gives it a status, reports the servers that need attention
and a count per status, then saves a CSV report and a bar chart.
"""

import pandas as pd
import matplotlib.pyplot as plt

# ==================================================================== LOAD
# 1. Read the data file: one row per server.

df = pd.read_csv("servers.csv")


# ================================================================= PROCESS
# 2. percent = how full each disk is: (used_gb / total_gb) * 100
#    status  = OVER LIMIT (100%+), WARNING (90%+), otherwise OK

df["percent"] = (df["used_gb"] / df["total_gb"]) * 100
df["status"] = df["percent"].apply(
    lambda p: "OVER LIMIT" if p >= 100 else "WARNING" if p >= 90 else "OK"
)


# ================================================================== OUTPUT
# 3. Threshold : print the whole DataFrame.
#    Typical   : print only the servers that are not OK.
#    Excellent : count per status, save a CSV report and a bar chart.

print("All servers:")
print(df)

not_ok = df[df["status"] != "OK"]
print("\nServers that need attention:")
print(not_ok)

status_counts = df.groupby("status")["hostname"].count()
print("\nServers per status:")
print(status_counts)

df.to_csv("servers_report.csv", index=False)
print(f"\nChecked {len(df)} servers. {len(not_ok)} need attention.")
print("Saved servers_report.csv")

status_counts.plot(kind="bar")
plt.title("Servers per status")
plt.xlabel("Status")
plt.ylabel("Number of servers")
plt.tight_layout()
plt.savefig("server_status.png")
print("Saved server_status.png")
plt.show()


# ==========================================================================
# 4. Before you finish:
#    [x] Checked one row by hand: srv-05 -> 126 / 120 * 100 = 105.0 -> OVER LIMIT
#    [x] One condition only, so no & / | needed
#    [x] Column names match the CSV header exactly: hostname, used_gb, total_gb
````
