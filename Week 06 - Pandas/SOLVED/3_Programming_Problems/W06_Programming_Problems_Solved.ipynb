## Task 3: Selecting Columns and Rows
1. Create a new DataFrame called `grades_df` containing only the columns `Name`, `Course`, and `Grade`. Display the first 8 rows.
2. Use `iloc` to display rows **5 to 9** (positions).
3. Use `loc` to display the **Name** and **City** of the student at index `3`.

**Hint:** For multiple columns, use `df[['col1', 'col2']]`.

````python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

df = pd.read_csv("DATA/students.csv")

# 1. Only three columns -> a new DataFrame (note the DOUBLE brackets)
grades_df = df[["Name", "Course", "Grade"]]
print("grades_df, first 8 rows:")
print(grades_df.head(8))

# 2. iloc = by POSITION. Rows 5 to 9 -> 5:10 (the end number is NOT included)
print("\nRows at positions 5 to 9 (iloc[5:10]):")
print(df.iloc[5:10])

# 3. loc = by LABEL. Row with index label 3, columns Name and City
print("\nName and City of the student at index 3:")
print(df.loc[3, ["Name", "City"]])

name = df.loc[3, "Name"]
city = df.loc[3, "City"]
print(f"\nIndex 3 is {name}, from {city}.")
````

> 💡 **Explanation**
>
> **Line by line**
> - `df[["Name", "Course", "Grade"]]`: **two pairs of square brackets**. The outer brackets mean "select from df", and the inner brackets are a Python **list** of column names. The result is a new DataFrame.
>   - Compare: `df["Name"]` (one pair of brackets) gives a **Series**, which is a single column. `df[["Name"]]` (two pairs of brackets) gives a **DataFrame** with one column.
> - `grades_df.head(8)`: the first 8 rows.
> - `df.iloc[5:10]`: `iloc` means "integer location": it works **by position**, like list indexes. `5:10` is the same **slice** from week 4: start at 5, go up to 10, but **do not include 10 itself**. So we get rows 5, 6, 7, 8, 9 = "5 to 9" ✓.
> - `df.loc[3, ["Name", "City"]]`: `loc` means "**by label**". Before the comma = which row (label 3). After the comma = which columns. The result is a small Series.
> - `df.loc[3, "Name"]`: one row and one column = **a single value** (`"Daniel Evans"`). This way you can use it in an f-string.
>
> **The difference between `iloc` and `loc` (very important):**
>
> | | `iloc` | `loc` |
> |---|---|---|
> | What it works with | position (0, 1, 2, ...) | index label |
> | Slice `5:10` | 5 to 9 (the last one is **not** included) | `loc[5:10]` means 5 to 10 (the last one **is** included!) |
> | Columns | by number: `iloc[3, 0]` | by name: `loc[3, "Name"]` |
>
> Here the index is the simple 0, 1, 2, ..., so `iloc[3]` and `loc[3]` give the same row. But after a sort or a filter, the labels and the positions become different, and then this difference becomes very important.
>
> **Output:**

## Task 4: Simple Filtering
1. Display all students who study in **London**. How many are there?
2. Display students whose **Grade is below 65**. Print their Name and Grade.
3. Use `value_counts()` on the `City` column to see how many students are in each city.

**Expected output (partial):**
```
Students in London: 11
```

````python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

df = pd.read_csv("DATA/students.csv")

# 1. Students in London
london = df[df["City"] == "London"]
print(london)
print(f"Students in London: {len(london)}")

# 2. Grade below 65 -> only Name and Grade
low_grades = df[df["Grade"] < 65]
print(f"\nStudents with a grade below 65 ({len(low_grades)}):")
print(low_grades[["Name", "Grade"]])

# 3. How many students in each city?
print("\nStudents per city:")
print(df["City"].value_counts())
````

> 💡 **Explanation**
>
> **Line by line**
> - `df["City"] == "London"`: this is not **one** comparison, it is **25** comparisons! pandas checks every row and builds a column of `True`/`False`. This is called a **boolean mask**:
> ```
> 0     True
> 1     True
> 2    False
> 3    False
> 4     True
> ...
> ```
> - `df[mask]`: only the rows where the mask is `True` are kept. **The original index is kept**: that is why you see 0, 1, 4, 6, ... in the output, not 0, 1, 2, 3.
> - `len(london)`: the number of filtered rows = 11.
> - `df[df["Grade"] < 65]`: the same pattern, this time with `<`.
> - `low_grades[["Name", "Grade"]]`: from the filter result, only two columns (two pairs of brackets!).
> - `df["City"].value_counts()`: how many times each value appears, **sorted from most to least**. Here one line does the job of a loop and a counter dictionary from week 4.
>
> **Output (important parts):**
> ```
> Students in London: 11
>
> Students with a grade below 65 (4):
>               Name  Grade
> 18      Sam Turner     58
> 19  Tara Underwood     55
> 23      Yara Young     62
> 24      Zane Adams     45
>
> Students per city:
> City
> London        11
> Manchester     6
> Birmingham     4
> Leeds          4
> Name: count, dtype: int64
> ```
> The line `Students in London: 11` matches the expected output exactly ✓.
>
> **Hand check:** 11 + 6 + 4 + 4 = 25 = all the students ✓. The London students: Alice, Ben, Emma, Grace, Jack, Layla, Nina, Oliver, Rosa, Umar, Zane = 11 ✓.
>
> **Be careful:** "below 65" means `< 65`. Yara, with 62, is included too. (This is different from "at risk of failing" in Task 6, which is `< 60`. There, there are 3 students.)
>
> **Traps**
> - `df[df["City"] = "London"]` (one equals sign) gives a syntax error. A comparison is always `==`.
> - Capital and small letters matter: `"london"` finds no rows and returns an **empty** table, **with no error**. If your filter comes back empty, check the spelling of the value first.
> - `Name: count` under `value_counts` is just the name of the Series (from pandas 2 onwards). In older tutorials you will see `Name: City`. It makes no difference.
>
> **Try it yourself:** Try `df["City"].value_counts(normalize=True)`: instead of counts, it gives the **share** of each city (for example, London 0.44 means 44%).

## Task 5: Sort the Data
1. Sort the students by **Grade** from **highest to lowest** and display the top 5.
2. Sort by **City** (A → Z), and within each city, by **Grade** (lowest → highest). Display the first 10 rows of the result.

**Hint:** `sort_values(['col1', 'col2'], ascending=[True, False])`

````python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

df = pd.read_csv("DATA/students.csv")

# 1. Grade, highest -> lowest, top 5
by_grade = df.sort_values("Grade", ascending=False)
print("Top 5 students by Grade:")
print(by_grade.head(5))

# 2. City A -> Z, and inside each city Grade lowest -> highest
by_city_grade = df.sort_values(["City", "Grade"], ascending=[True, True])
print("\nSorted by City (A-Z), then Grade (low-high), first 10 rows:")
print(by_city_grade[["Name", "City", "Grade"]].head(10))
````

> 💡 **Explanation**
>
> **Line by line**
> - `df.sort_values("Grade", ascending=False)`: sort by the `Grade` column. `ascending` means "going up". `False` means **descending** (largest to smallest). The default is `True`.
> - `sort_values` returns a **new** table and does not change `df`. That is why we stored the result in `by_grade`.
> - `.head(5)`: the first 5 rows of the **sorted** table = the top 5 grades.
> - `df.sort_values(["City", "Grade"], ascending=[True, True])`: **two** sort keys, in order of priority:
>   - First by `City`, alphabetically (`True`).
>   - Only when two rows have **the same city**, by `Grade`, going up (`True`).
>   - The `ascending` list must be **the same length** as the list of columns. Each item is the direction for the matching column. Here both are ascending, so `ascending=True` on its own would also be enough, but I wrote the list so that you learn the general form.
> - `by_city_grade[["Name", "City", "Grade"]].head(10)`: only the three important columns, so you can easily see that the sort is correct.
>
> **Output:**

## Task 6: Grade Analysis with Multiple Conditions
1. Find all students who:
   - Study **Computer Science** or **Data Science** (`isin()`), **AND**
   - Have a **Grade above 75**, **AND**
   - Have an **Attendance above 80**

   Display their Name, Course, Grade, and Attendance.
2. How many students are at **risk of failing** (Grade < 60)? Print their names and grades.
3. Find all **scholarship students from Manchester** with a Grade above 80. Display Name, Grade, and Scholarship.

````python
import pandas as pd

df = pd.read_csv("DATA/students.csv")

# 1. (CS or DS) AND Grade > 75 AND Attendance > 80
strong_students = df[
    (df["Course"].isin(["Computer Science", "Data Science"]))
    & (df["Grade"] > 75)
    & (df["Attendance"] > 80)
]
print(f"CS/DS students with Grade > 75 and Attendance > 80: {len(strong_students)}")
print(strong_students[["Name", "Course", "Grade", "Attendance"]])

# 2. At risk of failing: Grade < 60
at_risk = df[df["Grade"] < 60]
print(f"\nStudents at risk of failing (Grade < 60): {len(at_risk)}")
print(at_risk[["Name", "Grade"]])

# 3. Scholarship students from Manchester with Grade > 80
manchester_scholars = df[
    (df["Scholarship"] == "Yes")
    & (df["City"] == "Manchester")
    & (df["Grade"] > 80)
]
print(f"\nManchester scholarship students with Grade > 80: {len(manchester_scholars)}")
print(manchester_scholars[["Name", "Grade", "Scholarship"]])
````

> 💡 **Explanation**
>
> **Line by line**
> - `df["Course"].isin(["Computer Science", "Data Science"])`: for each row it asks "is the Course value **in this list**?". It is like the normal Python `in`, but on the whole column. Without `isin` you would have to write:
>   `(df["Course"] == "Computer Science") | (df["Course"] == "Data Science")`, which is longer.
> - `&` means **AND**: a row is kept only when **all three** conditions are `True`. `|` means **OR**.
> - **Each condition goes in its own parentheses.** Why? Because in Python `&` is **stronger** than `>` (it runs first). Without parentheses, `df["Grade"] > 75 & df["Attendance"] > 80` is read like this: `df["Grade"] > (75 & df["Attendance"]) > 80`, which either gives an error or a wrong answer.
> - **Why several lines?** The whole filter is inside `df[ ... ]`. When a bracket is still open, Python lets you continue on the next line. This way each condition has its own line and is easy to read.
> - `df[df["Grade"] < 60]`: a simple filter.
> - Part 3: three conditions with `&`. `"Yes"` has a capital Y, exactly as it is in the file.
>
> **Output:**
> ```
> CS/DS students with Grade > 75 and Attendance > 80: 8
>              Name            Course  Grade  Attendance
> 0   Alice Johnson  Computer Science     95          95
> 1      Ben Carter  Computer Science     88          90
> 2    Chloe Davies  Computer Science     82          88
> 3    Daniel Evans  Computer Science     79          85
> 4     Emma Foster  Computer Science     76          82
> 14   Oliver Price      Data Science     92          94
> 15     Priya Shah      Data Science     84          88
> 16     Quinn Reed      Data Science     78          86
>
> Students at risk of failing (Grade < 60): 3
>               Name  Grade
> 18      Sam Turner     58
> 19  Tara Underwood     55
> 24      Zane Adams     45
>
> Manchester scholarship students with Grade > 80: 2
>            Name  Grade Scholarship
> 2  Chloe Davies     82         Yes
> 8    Isla Jones     91         Yes
> ```
>
> **Hand check:** Why is Frank Green (CS, grade 72) not there? Because 72 is not > 75 ✓. Emma Foster, with 76 and 82, passed both conditions **only just**, so she is in ✓. Priya Shah from Manchester has a grade of 84, but she has no scholarship, so she is not in part 3 ✓.
>
> **Traps**
> - `and` / `or` instead of `&` / `|`: the famous error `ValueError: The truth value of a Series is ambiguous. Use a.empty, a.bool(), a.item(), a.any() or a.all().` It means: "`and` only understands a single True/False, but you gave it a column of 25 values". (This is exactly the error in the second warm-up of this week's lab.)
> - If you forget the parentheses: an error (a `ValueError` or a `TypeError`, depending on the columns) or a strange result.
> - `isin("Computer Science")` (one string instead of a list) gives a `TypeError`. `isin` always needs a **list**, even for one value: `isin(["Computer Science"])`.
>
> **Try it yourself:** With `~` (which means NOT), find all the students who are **not** in CS or DS: `df[~df["Course"].isin([...])]`.

````python
import pandas as pd

df = pd.read_csv("DATA/students.csv")

# 1. Result: Pass / Fail with apply + lambda
df["Result"] = df["Grade"].apply(lambda grade: "Pass" if grade >= 60 else "Fail")

# 2. Combined_Score = 70% grade + 30% attendance, 1 decimal place
df["Combined_Score"] = (df["Grade"] * 0.7 + df["Attendance"] * 0.3).round(1)

# 3. First 10 rows of the columns we care about
print(df[["Name", "Grade", "Attendance", "Result", "Combined_Score"]].head(10))

# 4. How many passed and how many failed?
result_counts = df["Result"].value_counts()
print("\nPass/Fail counts:")
print(result_counts)
print(f"\nPassed: {result_counts['Pass']}")
print(f"Failed: {result_counts['Fail']}")
````

> - `result_counts['Pass']`: the result of `value_counts()` is a Series with the labels `Pass` and `Fail`. We use the label to get each number separately.

## Task 8: Course Statistics with Groupby
Group the data by **Course** and calculate:
- `Count` — number of students
- `Avg_Grade` — average grade (rounded to 1 decimal)
- `Max_Grade` — highest grade
- `Avg_Attendance` — average attendance (rounded to 1 decimal)

Use `groupby().agg()` with named aggregations. Sort the result by **Avg_Grade** from highest to lowest.

**Expected output (approximate):**
```
                  Count  Avg_Grade  Max_Grade  Avg_Attendance
Course
Computer Science      8       78.1         95            83.4
Mathematics           6       82.2         91            86.3
...
```

````python
import pandas as pd

df = pd.read_csv("DATA/students.csv")

course_stats = df.groupby("Course").agg(
    Count=("Name", "count"),
    Avg_Grade=("Grade", "mean"),
    Max_Grade=("Grade", "max"),
    Avg_Attendance=("Attendance", "mean"),
)

# Round the two averages to 1 decimal place
course_stats["Avg_Grade"] = course_stats["Avg_Grade"].round(1)
course_stats["Avg_Attendance"] = course_stats["Avg_Attendance"].round(1)

# Highest average grade first
course_stats = course_stats.sort_values("Avg_Grade", ascending=False)
print(course_stats)
````

> 💡 **Explanation**
>
> **The main idea: split → apply → combine.** `groupby` does three jobs:
> 1. **Split:** it splits the rows into 4 groups by `Course`.
> 2. **Apply:** it does the calculations on each group (count, mean, max).
> 3. **Combine:** it puts the results into a new table, with **one row for each course**.
>
> **Line by line**
> - `df.groupby("Course").agg(...)`: inside `agg`, each **argument** makes a new column. The form is:
>   `new_column_name=("original_column", "function")`
>   - `Count=("Name", "count")`: in each group, count the filled cells of the `Name` column. (Any column with no empty cells would work. `Name` is the most natural.)
>   - `Avg_Grade=("Grade", "mean")`: the mean `Grade` of each group.
>   - `Max_Grade=("Grade", "max")`: something else from **the same** `Grade` column. One column can be used several times.
>   - `Avg_Attendance=("Attendance", "mean")`.
> - The function names are **strings** (`"mean"`, with quotes).
> - `course_stats["Avg_Grade"].round(1)`: we round only the two average columns. `Count` and `Max_Grade` are whole numbers.
> - `sort_values("Avg_Grade", ascending=False)`: highest to lowest.
> - `Course` has become the **index** of the new table. (That is why its name is printed one line above the other column names.)
>
> **Output:**
> ```
>                   Count  Avg_Grade  Max_Grade  Avg_Attendance
> Course
> Mathematics           6       82.2         91            86.3
> Computer Science      8       78.1         95            83.4
> Data Science          6       73.5         92            78.2
> Cyber Security        5       69.0         87            75.8
> ```
>
> **⚠ Something strange in the "expected output":** There, Computer Science (78.1) comes **above** Mathematics (82.2). But the task itself says "sort by Avg_Grade from highest to lowest", so 82.2 must come **first**. The expected output was probably taken before sorting (or it is only "approximate", as it says). The **numbers** are the same as ours. Only the **order** is different. Our output is correct.
>
> **Hand check:** Computer Science: (95+88+82+79+76+72+68+65) ÷ 8 = 625 ÷ 8 = 78.125 → `78.1` ✓. Attendance: 667 ÷ 8 = 83.375 → `83.4` ✓. Mathematics: 493 ÷ 6 = 82.17 → `82.2` ✓.
>
> **A point:** Mathematics has the highest **average**, but the **highest single grade** (95) belongs to Computer Science. "Best" depends on what you measure.
>
> **Traps**
> - `df.groupby("Course").mean()` (without choosing columns): a `TypeError`, because it tries to take the mean of `Name` and `City` too. Either choose the columns, or use `mean(numeric_only=True)`.
> - In `agg`, do not forget the **parentheses** of the tuple: `Count="Name", "count"` gives a syntax error.
> - Here it makes no difference whether you round first and then sort, or sort first and then round. But always do the **calculation** with the exact numbers and round at the **end** (for display).

## Task 9: Handling Missing Data
Load **`DATA/students_missing.csv`** (this file has some missing values).

````python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

df = pd.read_csv("DATA/students_missing.csv")
print(f"Rows before cleaning: {len(df)}")

# 1. Missing values in each column
print("\nMissing values per column:")
print(df.isna().sum())

# 2. Rows with at least one missing value
rows_with_missing = df[df.isna().any(axis=1)]
print(f"\nRows with at least one missing value ({len(rows_with_missing)}):")
print(rows_with_missing)

# 3. Fix the missing values (always save the result back with =)
grade_average = round(df["Grade"].mean(), 1)
print(f"\nAverage Grade used to fill the gaps: {grade_average}")
df["Grade"] = df["Grade"].fillna(grade_average)
df["Scholarship"] = df["Scholarship"].fillna("Unknown")
df["Course"] = df["Course"].fillna("Unknown")
df = df.dropna(subset=["Attendance"])
df = df.dropna(subset=["Age"])

# NaN had turned Age and Attendance into decimals (18.0); make them whole numbers again
df["Age"] = df["Age"].astype(int)
df["Attendance"] = df["Attendance"].astype(int)

# 4. Check: no missing values left
print("\nMissing values after cleaning:")
print(df.isnull().sum())
print(f"Total missing: {df.isnull().sum().sum()}")
print(f"Rows after cleaning: {len(df)}")

# 5. Save without the index
df.to_csv("DATA/students_clean.csv", index=False)
print("\nSaved DATA/students_clean.csv")
````

## Task 11: Grade Bands
Add a column called `Grade_Band` to the DataFrame using these rules:

| Grade | Band |
|-------|------|
| 90 – 100 | `'A'` |
| 80 – 89 | `'B'` |
| 70 – 79 | `'C'` |
| 60 – 69 | `'D'` |
| below 60 | `'F'` |

1. Use a function with `apply()` or `pd.cut()` to create the `Grade_Band` column.
2. Display Name, Grade, and Grade_Band for all students.
3. Print a **count** of students per Grade Band (use `value_counts()`).
4. Which **course** has the most students with a Band A or B grade?

**Hint for `pd.cut()`:**
```python
bins   = [0, 59, 69, 79, 89, 100]
labels = ['F', 'D', 'C', 'B', 'A']
df['Grade_Band'] = pd.cut(df['Grade'], bins=bins, labels=labels)
```

````python
import pandas as pd

df = pd.read_csv("DATA/students.csv")


def grade_band(grade):
    """Return the letter band (A-F) for a grade."""
    if grade >= 90:
        return "A"
    elif grade >= 80:
        return "B"
    elif grade >= 70:
        return "C"
    elif grade >= 60:
        return "D"
    else:
        return "F"


# 1. New column with apply() + our function (no brackets after grade_band!)
df["Grade_Band"] = df["Grade"].apply(grade_band)

# 2. All students
print(df[["Name", "Grade", "Grade_Band"]])

# 3. Students per band, in A-F order
print("\nStudents per Grade Band:")
print(df["Grade_Band"].value_counts().sort_index())

# 4. Which course has the most A or B students?
top_bands = df[df["Grade_Band"].isin(["A", "B"])]
ab_per_course = top_bands["Course"].value_counts()
print("\nA or B students per course:")
print(ab_per_course)
print(f"Most A/B students: {ab_per_course.index[0]} ({ab_per_course.iloc[0]})")
````

> 💡 **Explanation**
>
> **Version 1: a normal function + `apply`**
>
> **Line by line**
> - `def grade_band(grade):`: a normal function, as in week 3. It takes a grade and returns a letter. Here `def` is better than `lambda`, because there are 5 cases, and a lambda with 4 nested `if ... else` would be hard to read.
> - **The order of the `if`s, from top to bottom:** first `>= 90`, then `>= 80`, ... Because an `elif` is only checked when the ones before it were `False`, `elif grade >= 80` really means "80 or more **and** less than 90". You do not need to write `80 <= grade < 90`.
> - `df["Grade"].apply(grade_band)`: **the function name without parentheses!** We give **the function itself** to `apply`, so that `apply` can call it for each grade. `apply(grade_band())` gives an error (`TypeError: missing 1 required positional argument`), because you called the function straight away, without a grade.
> - `print(df[["Name", "Grade", "Grade_Band"]])`: all 25 students. pandas shows tables of up to 60 rows in full.
> - `value_counts().sort_index()`: `value_counts` sorts from most to least (B would come first). `sort_index()` sorts by **label**: A, B, C, D, F. This order is more natural.
> - `df[df["Grade_Band"].isin(["A", "B"])]`: only the A or B students.
> - `top_bands["Course"].value_counts()`: count their courses. The first one is the answer.
>
> **Output (important parts):**
> ```
>               Name  Grade Grade_Band
> 0    Alice Johnson     95          A
> 1       Ben Carter     88          B
> 2     Chloe Davies     82          B
> 3     Daniel Evans     79          C
> ...
> 24      Zane Adams     45          F
>
> Students per Grade Band:
> Grade_Band
> A    3
> B    9
> C    6
> D    4
> F    3
> Name: count, dtype: int64
>
> A or B students per course:
> Course
> Mathematics         5
> Computer Science    3
> Data Science        2
> Cyber Security      2
> Name: count, dtype: int64
> Most A/B students: Mathematics (5)
> ```
>
> **Hand check:** 3 + 9 + 6 + 4 + 3 = 25 ✓. The A students: Alice 95, Isla 91, Oliver 92. Mathematics: Isla 91, Jack 89, Kiran 85, Layla 82, Mohammed 80 = 5 ✓ (Nina, with 66, is in band D).
>
> **Answer to question 4:** **Mathematics**, with 5 A/B students (out of 6!). Computer Science has the most students (8), but only 3 of them are A/B.
>
> **⚠ A point about the band table:** The table only covers **whole numbers**: "80 to 89" and "90 to 100". If there was a grade like 89.5, it is not clear where it would go. Our function, with `>=`, has an answer for **any** number (89.5 → B). All the grades in this file are whole numbers, so there is no problem.

````python
import pandas as pd

df = pd.read_csv("DATA/students.csv")

bins = [0, 59, 69, 79, 89, 100]
labels = ["F", "D", "C", "B", "A"]
df["Grade_Band"] = pd.cut(df["Grade"], bins=bins, labels=labels)

print(df[["Name", "Grade", "Grade_Band"]].head(10))

print("\nStudents per Grade Band:")
print(df["Grade_Band"].value_counts().sort_index(ascending=False))

top_bands = df[df["Grade_Band"].isin(["A", "B"])]
print("\nA or B students per course:")
print(top_bands["Course"].value_counts())
````

> 💡 **Explanation**
>
> **Version 2: `pd.cut()`**
>
> `pd.cut` puts numbers into **ranges (bins)**. It does exactly the same job as our function, but without `if`.
>
> **Line by line**
> - `bins = [0, 59, 69, 79, 89, 100]`: the **edges** of the ranges. 6 edges = 5 ranges. Each range is **open on the left** and **closed on the right**:
>   `(0, 59]`, `(59, 69]`, `(69, 79]`, `(79, 89]`, `(89, 100]`
>   So `(59, 69]` = "more than 59, up to and including 69" = 60 to 69 for whole numbers.
> - `labels = ["F", "D", "C", "B", "A"]`: the name of each range, **in the same order** (from low to high). The number of labels must be **one less** than the number of edges (5 = 6 − 1). Otherwise you get a `ValueError`.
> - `pd.cut(df["Grade"], bins=bins, labels=labels)`: for each grade, it finds its range and gives it that label.
> - The result is a column of type **`category`**, not normal text. The categories have an **order** (F < D < C < B < A). That is why `sort_index(ascending=False)` shows them from A to F.
>
> **Output:** the same numbers as version 1 (A 3, B 9, C 6, D 4, F 3, and Mathematics with 5). The two methods give **the same answer** ✓.
>
> **`apply` or `pd.cut`?**
>
> | | function + `apply` | `pd.cut` |
> |---|---|---|
> | Easy to read for a beginner | very clear | you need to know the rules of the ranges |
> | A decimal grade (e.g. 59.5) | correct with `>=`: F | falls into `(59, 69]` → **D** ✗ |
> | A grade of 0 or a negative grade | F | `NaN`! (because `(0, 59]` does not include 0 itself) |
> | Speed on very large data | slower | faster |
>
> So the edges in the task's hint only work correctly for **whole numbers between 1 and 100**. For this file there is no problem, but always think about the edges before you use `pd.cut`. (A safer version: `bins = [-1, 59.99, 69.99, 79.99, 89.99, 100]`, or just use the function.)
>
> **Trap:** If you later group by a `category` column (`groupby("Grade_Band")`), pandas 2 gives a `FutureWarning` about `observed`. That is why I used a filter + `value_counts` for question 4, not `groupby`.

## Task 12: City Performance Report
Produce a **city-level performance summary** DataFrame with these columns:

| Column | What to calculate |
|--------|------------------|
| `Students` | Count of students per city |
| `Avg_Grade` | Average grade per city (1 dp) |
| `Avg_Attendance` | Average attendance per city (1 dp) |
| `Pass_Rate` | Percentage of students with Grade ≥ 60 (1 dp) |
| `Performance` | `'Above Average'` if Avg_Grade > overall mean, else `'Below Average'` |

**Steps:**
1. Calculate the overall average grade first (`df['Grade'].mean()`).
2. Build the summary using `groupby('City').agg()`.
3. Add the `Pass_Rate` column manually (calculate pass rate within each city group).
4. Add the `Performance` column comparing each city's avg grade to the overall mean.
5. Save the final summary to `DATA/city_report.csv` (no index).
6. Print the table sorted by `Avg_Grade` (highest first).

**Hint for Pass_Rate:**
```python
city_pass = df.groupby('City').apply(lambda g: (g['Grade'] >= 60).sum() / len(g) * 100)
```

````python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

df = pd.read_csv("DATA/students.csv")

# 1. Overall average grade (the line every city is compared with)
overall_mean = df["Grade"].mean()
print(f"Overall average grade: {overall_mean:.2f}\n")

# 2. Basic summary per city
city_report = df.groupby("City").agg(
    Students=("Name", "count"),
    Avg_Grade=("Grade", "mean"),
    Avg_Attendance=("Attendance", "mean"),
)

# 3. Pass rate inside each city group (% with Grade >= 60)
city_pass = df.groupby("City")["Grade"].apply(
    lambda grades: (grades >= 60).sum() / len(grades) * 100
)
city_report["Pass_Rate"] = city_pass

# 4. Above / below the overall mean (compare BEFORE rounding)
city_report["Performance"] = city_report["Avg_Grade"].apply(
    lambda avg: "Above Average" if avg > overall_mean else "Below Average"
)

# Round the numbers to 1 decimal place
city_report["Avg_Grade"] = city_report["Avg_Grade"].round(1)
city_report["Avg_Attendance"] = city_report["Avg_Attendance"].round(1)
city_report["Pass_Rate"] = city_report["Pass_Rate"].round(1)

# City is the index -> make it a normal column, so index=False does not lose it
city_report = city_report.reset_index()

# 6. Highest average grade first
city_report = city_report.sort_values("Avg_Grade", ascending=False)

# 5. Save without the index
city_report.to_csv("DATA/city_report.csv", index=False)

print(city_report)
print("\nSaved DATA/city_report.csv")
````

> 💡 **Explanation**
>
> **Line by line**
> - `overall_mean = df["Grade"].mean()`: the mean of **all** 25 students = 76.16. This is the "baseline" that each city is compared with.
> - `df.groupby("City").agg(...)`: the same pattern as Task 8. These are the first three columns of the report.
> - **`Pass_Rate`:**
>   - `df.groupby("City")["Grade"]`: the grades, grouped by city.
>   - `.apply(lambda grades: ...)`: this time `apply` works on **groups**, not on single numbers. So each time, `grades` is **all the grades of one city** (a Series), for example the 11 grades of London.
>   - `(grades >= 60)`: a Series of True/False. `.sum()` counts the `True` values = the number of passes.
>   - `/ len(grades) * 100`: divide by the total number of students in that city, then multiply by 100 = a percentage.
>   - The result is a Series whose index is the cities.
> - `city_report["Pass_Rate"] = city_pass`: because both indexes are **city names**, pandas puts each number next to the right city by itself. This is called **alignment** by index.
> - **`Performance`:** `apply` with a lambda on the `Avg_Grade` column. A lambda can use the outside variable `overall_mean`.
> - **Why did we compare before rounding?** If a city's mean was 76.18, it would become 76.2 after rounding, which is still larger, so no problem. But if it was 76.155, it would also become 76.2 after rounding. That looks larger than 76.16, but the exact number 76.155 is really **smaller** than 76.16. So comparing the rounded number would give the wrong answer ("Above Average" instead of "Below Average"). The general rule: **make decisions with the exact number, and round only for display.**
> - `city_report.reset_index()`: **this line is very important!** After `groupby`, `City` has become the **index** of the table, not a column. If you use `to_csv(index=False)` directly, **the city names are lost**, and your file becomes a list of numbers with no names. `reset_index()` turns the city back into a normal column and makes a new index 0, 1, 2, 3.
> - **The order of steps 5 and 6:** I sorted first and then saved, so the CSV file is sorted too. The printed output is the same sorted table.
>

Investigate whether scholarships are awarded to the right students. Answer each question with a `print()` statement and the calculation:

````python
import pandas as pd

df = pd.read_csv("DATA/students.csv")

# A 1/0 helper column: 1 = has a scholarship, 0 = does not
df["Has_Scholarship"] = df["Scholarship"].apply(lambda s: 1 if s == "Yes" else 0)

# 1. Percentage of all students with a scholarship
scholars = df["Has_Scholarship"].sum()
percent_scholars = scholars / len(df) * 100
print(f"1) {scholars} of {len(df)} students have a scholarship ({percent_scholars:.1f}%)")

# 2. Average grade: scholarship vs non-scholarship
grade_by_scholarship = df.groupby("Scholarship")["Grade"].mean()
grade_yes = grade_by_scholarship["Yes"]
grade_no = grade_by_scholarship["No"]
print(f"\n2) Average Grade - scholarship: {grade_yes:.1f}, no scholarship: {grade_no:.1f}")
print(f"   Difference: {grade_yes - grade_no:.1f} points -> a big, meaningful gap")

# 3. Per course: total, with scholarship, rate (%)
course_table = df.groupby("Course").agg(
    Total_Students=("Name", "count"),
    With_Scholarship=("Has_Scholarship", "sum"),
)
course_table["Scholarship_Rate"] = (
    course_table["With_Scholarship"] / course_table["Total_Students"] * 100
).round(1)
course_table = course_table.sort_values("Scholarship_Rate", ascending=False)
print("\n3) Scholarships per course:")
print(course_table)

# 4. Attendance: scholarship vs non-scholarship
attendance_by_scholarship = df.groupby("Scholarship")["Attendance"].mean()
att_yes = attendance_by_scholarship["Yes"]
att_no = attendance_by_scholarship["No"]
print(f"\n4) Average Attendance - scholarship: {att_yes:.1f}%, no scholarship: {att_no:.1f}%")
print(f"   Difference: {att_yes - att_no:.1f} points")

# 5. SUMMARY
# 10 of the 25 students (40%) have a scholarship. Scholarship students average
# 86.1 against 69.5 for the rest (16.6 points higher) and also attend more
# (90.1% against 75.5%), so the awards do go to strong, committed students.
# Mathematics has the highest scholarship rate (66.7%) and Data Science the lowest
# (16.7%, only Oliver Price), even though Data Science has other strong students
# such as Priya Shah (84) - that is worth a second look.
````

## Task 16: Iris Dataset – Group Analysis
1. Calculate the **mean** of every measurement per species with `groupby('target')`.
2. Which species has the largest average petal length? Which has the smallest average sepal width?
3. Add a column `petal_area` (`petal length × petal width`) and show the 5 flowers with the largest petal area.
4. Calculate the correlation between `petal length (cm)` and `petal width (cm)` and interpret it in a comment.

````python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

df = pd.read_csv("DATA/iris_dataset.csv")

# 1. Mean of every measurement per species
species_means = df.groupby("target").mean()
print("Mean per species:")
print(species_means.round(3))

# 2. Largest average petal length / smallest average sepal width
longest_petals = species_means["petal length (cm)"].idxmax()
narrowest_sepals = species_means["sepal width (cm)"].idxmin()
print(f"\nLargest average petal length: {longest_petals} "
      f"({species_means['petal length (cm)'].max():.3f} cm)")
print(f"Smallest average sepal width: {narrowest_sepals} "
      f"({species_means['sepal width (cm)'].min():.3f} cm)")

# 3. New column: petal area, and the 5 largest
df["petal_area"] = df["petal length (cm)"] * df["petal width (cm)"]
print("\n5 flowers with the largest petal area:")
print(df.sort_values("petal_area", ascending=False).head(5))

# 4. Correlation between petal length and petal width
r = df["petal length (cm)"].corr(df["petal width (cm)"])
print(f"\nCorrelation petal length vs petal width: {r:.3f}")
# 0.963 is very close to 1.0 -> a very strong positive relationship:
# flowers with longer petals almost always have wider petals too.
````

> - `sort_values("petal_area", ascending=False).head(5)`: the 5 largest.
> - `df["petal length (cm)"].corr(df["petal width (cm)"])`: the correlation of the two columns, as one number.

## Task 17: Cars Dataset
Load `DATA/CARS.text` (comma-separated, columns include `model`, `mpg`, `cyl`, `hp`, `wt`, `am`).
1. Show the 5 cars with the best `mpg` and the 5 with the worst.
2. Show cars with 4 cylinders and `hp` above 90.
3. Group by `cyl` and calculate the count, average `mpg` and average `hp`.
4. Add a column `Efficiency` that is `'High'` if `mpg` ≥ 25, `'Medium'` if `mpg` ≥ 18 and `'Low'` otherwise, and count the cars in each category.
5. Save the result to `cars_clean.csv` (no index).

````python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

df = pd.read_csv("DATA/CARS.text")
print(f"Loaded {len(df)} cars, columns: {df.columns.tolist()}")

# 1. Best and worst mpg
columns = ["model", "mpg", "cyl", "hp", "wt", "am"]
print("\n5 cars with the best mpg:")
print(df.sort_values("mpg", ascending=False)[columns].head(5))
print("\n5 cars with the worst mpg:")
print(df.sort_values("mpg", ascending=True)[columns].head(5))

# 2. 4 cylinders AND more than 90 hp
four_cyl_powerful = df[(df["cyl"] == 4) & (df["hp"] > 90)]
print(f"\n4-cylinder cars with hp above 90: {len(four_cyl_powerful)}")
print(four_cyl_powerful[["model", "cyl", "hp", "mpg"]])

# 3. Per number of cylinders: count, average mpg, average hp
cyl_stats = df.groupby("cyl").agg(
    Count=("model", "count"),
    Avg_mpg=("mpg", "mean"),
    Avg_hp=("hp", "mean"),
).round(1)
print("\nBy number of cylinders:")
print(cyl_stats)

# 4. Efficiency category with apply + lambda
df["Efficiency"] = df["mpg"].apply(
    lambda m: "High" if m >= 25 else "Medium" if m >= 18 else "Low"
)
print("\nCars per Efficiency category:")
print(df["Efficiency"].value_counts())

# 5. Save (in the current folder, as the task says)
df.to_csv("cars_clean.csv", index=False)
print("\nSaved cars_clean.csv")
````

## Task 18: Reading Excel and CSV Files
1. Load `DATA/data.xlsx` with `pd.read_excel()` (you may need `pip install openpyxl`) and load `DATA/sample_data.csv` with `pd.read_csv()`.
2. Compare the two DataFrames: do they have the same columns and the same rows?
3. Combine them into one DataFrame with `pd.concat()`, remove duplicate rows and show the result.
4. Filter the people older than 30 and sort them by `Age`.

````python
import pandas as pd

# 1. Load both files
excel_df = pd.read_excel("DATA/data.xlsx")
csv_df = pd.read_csv("DATA/sample_data.csv")
print("From data.xlsx:")
print(excel_df)
print("\nFrom sample_data.csv:")
print(csv_df)

# 2. Compare them
same_columns = excel_df.columns.tolist() == csv_df.columns.tolist()
same_shape = excel_df.shape == csv_df.shape
same_everything = excel_df.equals(csv_df)
print(f"\nSame columns: {same_columns}")
print(f"Same shape:   {same_shape} {excel_df.shape}")
print(f"Same rows (identical data): {same_everything}")

# 3. Stack them, then remove duplicate rows
combined = pd.concat([excel_df, csv_df], ignore_index=True)
print(f"\nAfter concat: {len(combined)} rows")
combined = combined.drop_duplicates(ignore_index=True)
print(f"After drop_duplicates: {len(combined)} rows")
print(combined)

# 4. People older than 30, youngest first
older = combined[combined["Age"] > 30].sort_values("Age")
print("\nOlder than 30, sorted by Age:")
print(older)
````

> 💡 **Explanation**
>
> **Line by line**
> - `pd.read_excel("DATA/data.xlsx")`: like `read_csv`, but for an Excel file. pandas does not open Excel files by itself. It gets help from the **openpyxl** library. If it is not installed, you get the error `ImportError: Missing optional dependency 'openpyxl'`. In the Mac terminal, type: `pip install openpyxl` (or `pip3 install openpyxl`). > - `excel_df.columns.tolist() == csv_df.columns.tolist()`: compares two Python lists. Both the names and their **order** must be the same.
> - `excel_df.shape == csv_df.shape`: compares two tuples.
> - `excel_df.equals(csv_df)`: the **strictest** comparison: all the values, all the columns, the index and even the **data types** must be the same. It returns a single `True` or `False`.
>   - Why not `excel_df == csv_df`? Because `==` on two DataFrames builds a **table** of True/False (cell by cell), not a single answer.
> - `pd.concat([excel_df, csv_df], ignore_index=True)`: sticks the two tables together, **one under the other**. It takes a **list** of DataFrames. `ignore_index=True` means "make a new index from 0 to 7". Without it, the index would be 0, 1, 2, 3, 0, 1, 2, 3 (repeated and confusing).
> - `drop_duplicates(ignore_index=True)`: deletes rows where **all the columns** are the same as an earlier row, and keeps the first copy. `ignore_index=True` makes the index tidy again.
> - `combined[combined["Age"] > 30].sort_values("Age")`: filter, then sort in ascending order (the default).
