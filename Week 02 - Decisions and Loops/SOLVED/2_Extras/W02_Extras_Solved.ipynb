## Python Pizza
Small $15, Medium $20, Large $25. Pepperoni +$2 on small, +$3 on medium/large. Extra cheese +$1 on any size. Ask for the choices, work out the bill and print exactly `Your final bill is: $<amount>.`

```python
print("Welcome to Python Pizza Deliveries!")
size = input("What size pizza do you want? S, M or L: ")
pepperoni = input("Do you want pepperoni on your pizza? Y or N: ")
extra_cheese = input("Do you want extra cheese? Y or N: ")

bill = 0
if size == "S":
    bill += 15
elif size == "M":
    bill += 20
else:
    bill += 25

if pepperoni == "Y":
    if size == "S":
        bill += 2
    else:
        bill += 3

if extra_cheese == "Y":
    bill += 1

print(f"Your final bill is: ${bill}.")
```

> 💡 **Explanation**
>
> **Main idea: build the bill step by step.** Start from `bill = 0`, and each choice adds an amount to it (`+=`).
>
> - **Three separate `if`s, not one `elif` chain:** size, pepperoni and cheese are three **independent** decisions. All three must be checked. If you join them with `elif`, everything after the first `True` is skipped.
> - **Nested `if`:** the price of pepperoni depends on the size, so inside `if pepperoni == "Y":` we check the size again.
> - The `else` for size means "anything other than S or M counts as large". It's simple, but if the user makes a typo (for example `X`), it also counts as large.
>
> **Check:** L + pepperoni + cheese = 25 + 3 + 1 → `Your final bill is: $29.`

## Treasure Island
A choose-your-own-adventure game: ask questions (left or right?) and use nested `if`/`else` to branch towards different endings. Plan the branches on paper first.

```python
print("Welcome to Treasure Island.")
print("Your mission is to find the treasure.")

choice1 = input('You are at a crossroad. Where do you want to go? Type "left" or "right": ').lower()

if choice1 == "left":
    choice2 = input('You reach a lake. Type "wait" to wait for a boat or "swim" to swim across: ').lower()
    if choice2 == "wait":
        choice3 = input("You arrive at an island with 3 doors: red, yellow and blue. Which colour? ").lower()
        if choice3 == "red":
            print("It's a room full of fire. Game Over.")
        elif choice3 == "yellow":
            print("You found the treasure! You Win!")
        elif choice3 == "blue":
            print("You enter a room of beasts. Game Over.")
        else:
            print("That door doesn't exist. Game Over.")
    else:
        print("You got attacked by an angry trout. Game Over.")
else:
    print("You fell into a hole. Game Over.")
```

> 💡 **Explanation**
>
> **Main idea: the next question is only asked if the previous answer was the right one.** That's why question 2 is **inside** `if choice1 == "left":`, and question 3 is inside `if choice2 == "wait":`. The indentation shows which branch each question belongs to.
>
> - `.lower()` at the end of each `input` turns the answer into lowercase, so `Left` and `LEFT` also work.
> - For the door question we use `if/elif/elif/else`, because **one** of several cases is chosen.
> - Each `else` means "any other answer". So strange answers also lead to an ending, and the program doesn't get stuck.
>
> **Exercise:** add a new branch, for example an extra question before the yellow door.

## Higher or Lower
Hard-code two things and their follower counts, ask which is higher, print Correct/Wrong, and loop until the user types `quit`.

```python
name_a = "Instagram"
followers_a = 676       # example numbers, in millions
name_b = "Cristiano Ronaldo"
followers_b = 650
score = 0

while True:
    print(f"A: {name_a}   vs   B: {name_b}")
    guess = input('Who has more followers (millions)? Type "A", "B" or "quit": ').upper()
    if guess == "QUIT":
        break

    if followers_a > followers_b:
        answer = "A"
    else:
        answer = "B"

    if guess == answer:
        score += 1
        print(f"Correct! Score: {score}")
    else:
        print(f"Wrong. Score: {score}")

print(f"Final score: {score}")
```

> 💡 **Explanation**
>
> - `while True:` + `break` is the same pattern as D11 in Week 2. We don't know how many rounds the user will play, so `while` is right, not `for`.
> - `.upper()` turns the answer into capitals, so `a` is also accepted. That's why we compare with `"QUIT"` (in capitals).
> - **The program works out the right answer, not us:** `if followers_a > followers_b: answer = "A"`. Change the numbers and the code still works correctly.
> - `score` starts before the loop and goes up inside the loop (the counter pattern).
>
> The original game picks two random accounts from a list each round. That belongs to Weeks 3 and 4 (`random` and lists).

## Number Guessing Game
Hard-code a secret number. Loop: ask for a guess, say too high / too low, congratulate and stop when correct. Count the guesses.

```python
secret = 37
guesses = 0

print("I'm thinking of a number between 1 and 100.")
while True:
    guess = int(input("Make a guess: "))
    guesses += 1
    if guess > secret:
        print("Too high.")
    elif guess < secret:
        print("Too low.")
    else:
        print(f"You got it! The answer was {secret}. It took you {guesses} guesses.")
        break
```

> 💡 **Explanation**
>
> - `guesses += 1` comes right **after** each guess is read, so the correct guess is counted too.
> - Three cases with `if`/`elif`/`else`: bigger, smaller, and otherwise (that is, equal), which is a win.
> - `break` is only in the winning branch. So the loop keeps going until the user guesses correctly.
>
> **Best strategy:** always guess the middle of the range (50, then 25 or 75, ...). You need at most 7 guesses. This method is called **binary search**. In Week 3 we replace the secret number with `random`.
