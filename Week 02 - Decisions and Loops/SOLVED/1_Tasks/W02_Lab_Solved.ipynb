## Warm-up 1 · `01_indentation_error.py`

```python

print("OVER")
```

```
IndentationError: expected an indented block after 'if' statement on line 7
```

**What goes wrong:** after a line ending in `:` (an `if`, `while`, `for`…), Python expects a **block**: one or more lines pushed to the right. Here `print` is not indented, so the `if` has nothing inside it.

````python
value = 23.7
limit = 20

if value > limit:
    print("OVER")
````

> 💡 **Explanation**
>
> **Indentation is not decoration in Python: it is the grammar.** The indented lines are "inside" the `if`; they only run when the condition is True. The standard is **4 spaces** (VS Code inserts them when you press Tab).

## Warm-up 2 · `02_syntax_error.py`

```python
while attempts <= max_attempts
    print("checking...")
```

```
SyntaxError: expected ':'
```

**What goes wrong:** every `if`, `elif`, `else`, `while` and `for` line must **end with a colon** `:`. The colon means "the block starts on the next line".

````python
attempts = 3
max_attempts = 3

while attempts <= max_attempts:
    print("checking...")
    attempts += 1
````

> 💡 **Explanation**
>
> It prints `checking...` **once**: `attempts` starts at 3, `3 <= 3` is True, it prints, then becomes 4, and `4 <= 3` is False, so the loop stops.
>
> Remember from Week 1: with a `SyntaxError`, also check the line **above** the one it reports (an unclosed bracket can cause it).

# Topic 1 · Comparisons and Boolean logic

A **comparison** asks a yes/no question and the answer is a **Boolean** (`bool`): `True` or `False`.

## D1 · The six comparison operators

**What it asks:** `a = 10`, `b = 20`, print all six comparisons.

````python
a = 10
b = 20

print(a == b)   # equal?
print(a != b)   # not equal?
print(a < b)    # less than?
print(a > b)    # greater than?
print(a <= b)   # less than or equal?
print(a >= b)   # greater than or equal?
````

> 💡 **Explanation**
>
> Output: `False True True False True False`. Read each one as a question: *"is 10 equal to 20?"* → False. `!=` is "not equal". `<=` means "smaller **or the same**", so `10 <= 10` is True but `10 < 10` is False.

## D2 · `=` stores, `==` asks

**What it asks:** store 50 in `limit`, print `limit == 50` and `limit == 60`, with a comment above each saying *stores* or *asks*.

````python
# stores: put 50 in the box called limit
limit = 50

# asks: is limit equal to 50?
print(limit == 50)

# asks: is limit equal to 60?
print(limit == 60)
````

> 💡 **Explanation**
>
> **The most common beginner bug:** `=` (one sign) **stores** a value; `==` (two signs) **asks** whether two things are equal. `if limit = 50:` is a `SyntaxError`, because you can't store inside a question.

## D3 · A comparison is a value too

**What it asks:** store the result of a comparison in a variable and look at its type.

````python
attempts = 4
max_attempts = 3

is_locked = attempts >= max_attempts

print(is_locked)
print(type(is_locked))
````

> 💡 **Explanation**
>
> `attempts >= max_attempts` is calculated first (`4 >= 3` → `True`), then stored. So a comparison is just another kind of value, like a number or text. Type: `<class 'bool'>`.
>
> **Why useful?** A name like `is_locked` makes later code read like English: `if is_locked:`. Names starting with `is_` / `has_` are a good habit for Booleans.

## D4 · `and`, `or`, `not`

**What it asks:** with `percent = 95` print three combined conditions; then predict them for `percent = 40`.

````python
percent = 95

print(percent >= 90 and percent < 100)
print(percent < 50 or percent > 90)
print(not (percent > 90))
````

> 💡 **Explanation**
>
> | Operator | True when | With 95 |

>
> **With `percent = 40` (predict before running!):** `False and True` → **False** · `True or False` → **True** · `not False` → **True**.

## D6 · `if` / `else`

**What it asks:** print `OVER LIMIT` or `OK`.

````python
value = 87
limit = 100

if value > limit:
    print("OVER LIMIT")
else:
    print("OK")
````

## D7 · Three results with `if` / `elif` / `else`

**What it asks:** calculate `percent`, then print `OVER LIMIT` (100+), `WARNING` (90+) or `OK`. Test with 87, 95, 120.

````python
value = 87
limit = 100

percent = (value / limit) * 100

if percent >= 100:
    print("OVER LIMIT")
elif percent >= 90:
    print("WARNING")
else:
    print("OK")
````

> 💡 **Explanation**
>
> **Python checks from the top and runs the FIRST branch that is True, then skips the rest.**
>
> Trace with `value = 120` → percent 120:
> 1. `120 >= 100`? True → print `OVER LIMIT` → **done**, the `elif` is never checked.
>
> **Why the order matters:** if you checked `>= 90` first, 120 would match it and wrongly print `WARNING`. Always put the **most extreme** condition first.
>
> `elif` = "else if". You can have as many `elif`s as you need.

````python
# Your turn: try it yourself here
# try the three test values: 87, 95, 120
value = 95
limit = 100
percent = (value / limit) * 100
if percent >= 100:
    print("OVER LIMIT")
elif percent >= 90:
    print("WARNING")
else:
    print("OK")
````

## D8 · Indentation decides what is inside the `if`

**What it asks:** predict, run, then indent the last line and explain the change.

````python
# Prediction: 50 > 100 is False, so "OVER LIMIT" is skipped,
# but "Check finished" is NOT indented, so it always runs.
value = 50
limit = 100

if value > limit:
    print("OVER LIMIT")
print("Check finished")
````

````python
# Step 3: now "Check finished" is indented, so it is INSIDE the if.
# The condition is False, so nothing at all is printed.
value = 50
limit = 100

if value > limit:
    print("OVER LIMIT")
    print("Check finished")
````

> 💡 **Explanation**
>
> Same code, one indent different, different program. That's why indentation errors are serious in Python: the code may still run but **do the wrong thing** with no error message.

# Topic 3 · `while` loops

A `while` loop repeats its block **as long as the condition is True**. It checks the condition before every round.

## D9 · Count from 1 to 10

````python
count = 1

while count <= 10:
    print(count)
    count += 1
````

> 💡 **Explanation**
>
> **The three parts of every counting loop:**
> 1. **start** before the loop: `count = 1`
> 2. **test** in the `while` line: `count <= 10`
> 3. **change** inside the loop: `count += 1` (short for `count = count + 1`)
>
> Trace the end: count is 10 → `10 <= 10` True → print 10 → count becomes 11 → `11 <= 10` False → stop.

## D10 · Fix the loop that never stops

```python
count = 1
while count <= 5:
    print(count)
```

**Why it's broken:** nothing inside the loop changes `count`, so `1 <= 5` is True forever (an *infinite loop*). **Fix:** add the "change" part.

````python
count = 1
while count <= 5:
    print(count)
    count += 1
````

> 💡 **Explanation**
>
> If a loop never ends: in a notebook press the ■ stop button; in a terminal press **Ctrl+C**. Ask yourself: *what inside the loop makes the condition become False one day?*

## D11 · `while True` + `break`: add numbers until `done`

**What it asks:** keep asking for numbers and add them up; when the user types `done`, stop and print the total. Try `5`, `10`, `2.5`, `done` → `Total: 17.5`.

````python
total = 0

while True:
    answer = input("Enter a number (or done to finish): ")
    if answer == "done":
        break
    total += float(answer)

print("Total:", total)
````

> 💡 **Explanation**
>
> **Line by line**
> - `total = 0`: a **running total** starts at zero, before the loop.
> - `while True:`: the condition is always True, so the loop runs until something inside **breaks** out.
> - `answer = input(...)`: keep the raw text first. We must check for `done` **before** converting, because `float("done")` would be a `ValueError`.
> - `if answer == "done": break`: `break` jumps out of the loop immediately.
> - `total += float(answer)`: convert and add to the total.
> - `print` is **not** indented, so it runs once, after the loop.
>
> **When to use `while True`:** when you don't know in advance how many rounds there will be, and the stop signal comes from inside (here: the user).

## D12 · `continue`: skip 5

**What it asks:** print 1 to 10 but skip 5, using `continue`.

````python
count = 0

while count < 10:
    count += 1
    if count == 5:
        continue
    print(count)
````

> 💡 **Explanation**
>
> `continue` means: *skip the rest of this round and go back to the top of the loop*.
>
> **The trap the hint warns about:** if you add 1 **after** the `continue`, then when count is 5 the `+= 1` is skipped too, count stays 5 forever → infinite loop. My solution adds 1 **first** (so I start at 0 and test `< 10`). Trace: count becomes 5 → `continue` → back to top → becomes 6 → printed.

# Topic 4 · `for` loops and `range()`

A `for` loop goes through a sequence of values, one by one. `range()` makes a sequence of numbers.

## D13 · `range(stop)` and `range(start, stop)`

````python
for i in range(5):
    print(i)

for i in range(1, 6):
    print(i)
````

> 💡 **Explanation**
>
> - `range(5)` → `0, 1, 2, 3, 4`: starts at 0, **stops before 5** (5 numbers).
> - `range(1, 6)` → `1, 2, 3, 4, 5`: to include 5 you must say 6.
>
> **Rule:** the stop value is **never** included. `i` is just a name; it gets the next value on each round. You don't need `i += 1`: the `for` loop does it for you.

## D14 · `range(start, stop, step)`

**What it asks:** even numbers 2 to 20; then count down 10 to 1.

````python
for n in range(2, 21, 2):
    print(n)

for n in range(10, 0, -1):
    print(n)
````

> 💡 **Explanation**
>
> - `range(2, 21, 2)`: start 2, add 2 each time, stop before 21 → 2, 4, …, 20. (Stop 20 would miss 20!)
> - `range(10, 0, -1)`: a **negative step** counts down; stop before 0 → 10, 9, …, 1.

## D15 · `enumerate()`: position and value together

````python
code = "A7X"

for position, character in enumerate(code, start=1):
    print(f"Character {position}: {character}")
````

> 💡 **Explanation**
>
> A `for` loop over a **string** gives one character at a time. `enumerate()` gives **pairs**: `(1, "A")`, `(2, "7")`, `(3, "X")`. Writing `position, character` takes each pair apart into two names. `start=1` makes counting start at 1 instead of 0 (humans count from 1).

## D16 · `for` or `while`?

**What it asks:** two programs, with a comment above each saying which loop and why. (a) Exactly 3 numbers, printed back. (b) Ask for a password until it's `letmein`, then `Welcome`.

````python
# a) for: I know in advance it runs exactly 3 times.
for i in range(3):
    number = float(input("Enter a number: "))
    print("You typed", number)

# b) while: I don't know how many tries the user will need.
password = input("Password: ")
while password != "letmein":
    password = input("Wrong. Password: ")
print("Welcome")
````

> 💡 **Explanation**
>
> **The rule:** `for` when you **know how many times** (a count, or every item of something); `while` when it depends on something that happens **during** the loop.
>
> In (b) I ask once **before** the loop, so the condition has something to check. Inside the loop I ask again, which is the "change" part, so the loop can end. `!=` means "not equal": *keep going while the password is wrong*.

## C1 · Check three records in one run

**What it asks:** a `for` loop that runs exactly 3 times; each time ask for value and limit, print `OVER LIMIT` or `OK`. Test: 87/100 → OK, 120/100 → OVER LIMIT, 45/100 → OK.

````python
for record_number in range(1, 4):
    print(f"Record {record_number}")
    value = float(input("Value: "))
    limit = float(input("Limit: "))

    if value > limit:
        print("OVER LIMIT")
    else:
        print("OK")
````

> 💡 **Explanation**
>
> **This combines three ideas:** a `for` loop (exactly 3 times → `for`, D16's rule), `input()` + `float()` (Week 1), and `if`/`else` (D6). Notice the `if` is indented **inside** the `for`, so it runs once per record. Using `range(1, 4)` means the record numbers are 1, 2, 3 for the human reading it.

## C2 · Three tries to log in

**What it asks:** password `python123`, at most 3 tries; correct → `Access granted` and stop; 3 wrong → `Account locked`.

````python
correct_password = "python123"
max_tries = 3
tries = 0
granted = False

while tries < max_tries:
    attempt = input("Password: ")
    tries += 1
    if attempt == correct_password:
        granted = True
        break
    print(f"Wrong password ({tries} of {max_tries})")

if granted:
    print(f"Access granted (after {tries} tries)")
else:
    print("Account locked")
````

> 💡 **Explanation**
>
> **Line by line**
> - `tries = 0` counts attempts; `granted = False` remembers whether the user got in (a **flag**: a Boolean that records something that happened inside the loop).
> - `while tries < max_tries:` allows at most 3 rounds.
> - `tries += 1` right after asking, so the count is correct.
> - Correct → set the flag and `break` (stop asking).
> - After the loop we need to know **why** it ended: correct password (break) or out of tries? The flag answers that.
>
> **Test 1:** `abc`, `python123` → Wrong (1 of 3), then Access granted (after 2 tries). **Test 2:** `abc`, `123`, `pass` → three wrongs → Account locked.
>
> **Another way you may see:** Python's `while ... else:` (the `else` runs only if the loop ended without `break`). The flag version is easier to read, so I used it.

# Mini-project · Record Check with decisions and a loop

# Mini-project · Record Check, version 2

This week I use the **IT lane** (hostname, GB used, GB total), because "OVER LIMIT / WARNING" makes natural sense for a disk that's filling up. Example: `srv-01`, `87`, `120`.

The template names the variables `label`, `value`, `limit`: I keep those names (they're clear enough and they match the brief), and add `status`.

## Threshold · decide `OVER LIMIT` or `OK` with `if`/`else`

````python
# ===== INPUT
label = input("Hostname: ")
value = float(input("GB used: "))
limit = float(input("GB total: "))

# ===== PROCESS
if value > limit:
    status = "OVER LIMIT"
else:
    status = "OK"

# ===== OUTPUT
print()
print("=" * 34)
print(f"  RECORD CHECK  -  {label}")
print("=" * 34)
print(f"  Used        : {value:.0f}")
print(f"  Total       : {limit:.0f}")
print(f"  Status      : {status}")
print("=" * 34)
````

> 💡 **Explanation**
>
> **The key idea: decide, then print.** The `if` doesn't print anything; it **stores** the result in `status`. The report then prints `status` like any other value. This keeps the decision (PROCESS) separate from the display (OUTPUT), exactly the structure the template asks for.

## Typical · difference, percent, 3-tier status, aligned numbers

````python
# ===== INPUT
label = input("Hostname: ")
value = float(input("GB used: "))
limit = float(input("GB total: "))

# ===== PROCESS
difference = limit - value              # GB still free
percent = value / limit * 100           # how full the disk is

if percent >= 100:
    status = "OVER LIMIT"
elif percent >= 90:
    status = "WARNING"
else:
    status = "OK"

# ===== OUTPUT
print()
print("=" * 34)
print(f"  RECORD CHECK  -  {label}")
print("=" * 34)
print(f"  Used        : {value:>10.2f}")
print(f"  Total       : {limit:>10.2f}")
print(f"  Free        : {difference:>10.2f}")
print(f"  Percent     : {percent:>10.2f} %")
print(f"  Status      : {status:>10}")
print("=" * 34)
````

> 💡 **Explanation**
>
> **What changed**
> - The status now depends on the **percentage**, not on comparing value and limit directly, so we can have a WARNING zone (90–99%). Same order rule as D7: check 100 first.
> - `{status:>10}`: right-aligns the **text** in 10 characters too, so `OK` ends in the same column as the numbers (as in the brief's example).
> - With `srv-01`, 87, 120 → Free 33.00, Percent 72.50 %, Status OK. Try 110 → WARNING (91.67%), 130 → OVER LIMIT.

## Excellent · loop until `quit`, count OVER LIMIT, print the count after the loop

````python
"""
RECORD CHECK  -  my version
Name  : Parsa
Lane  : IT
Run it:   python template.py
"""

over_limit_count = 0                     # counts OVER LIMIT records in this session

while True:
    # ===== INPUT
    label = input("Hostname (or quit to stop): ")
    if label == "quit":
        break
    value = float(input("GB used: "))
    limit = float(input("GB total: "))

    # ===== PROCESS
    difference = limit - value
    percent = value / limit * 100

    if percent >= 100:
        status = "OVER LIMIT"
        over_limit_count += 1
    elif percent >= 90:
        status = "WARNING"
    else:
        status = "OK"

    # ===== OUTPUT
    print()
    print("=" * 34)
    print(f"  RECORD CHECK  -  {label}")
    print("=" * 34)
    print(f"  Used        : {value:>10.2f}")
    print(f"  Total       : {limit:>10.2f}")
    print(f"  Free        : {difference:>10.2f}")
    print(f"  Percent     : {percent:>10.2f} %")
    print(f"  Status      : {status:>10}")
    print("=" * 34)
    print()

print(f"Session finished: {over_limit_count} record(s) OVER LIMIT.")
````

> 💡 **Explanation**
>
> **What changed**
> - The whole INPUT → PROCESS → OUTPUT is now **indented inside `while True:`**, so it repeats.
> - The very first question checks for `quit` **before** asking the numbers, and `break`s out (D11's pattern).
> - `over_limit_count = 0` is created **before** the loop (if it were inside, it would reset to 0 every round!). It goes up by 1 inside the `OVER LIMIT` branch.
> - The final `print` is **not** indented, so it runs **once**, after the loop, as the brief asks.
>
> **Test session:** `srv-01 87 120` (OK) → `srv-02 125 120` (OVER LIMIT) → `srv-03 110 120` (WARNING) → `quit` → *Session finished: 1 record(s) OVER LIMIT.*
>
> **Total of 0:** `value / 0` → `ZeroDivisionError`. The brief says note it, don't fix it yet. (You *could* now, with `if limit == 0:`; that's coming.)
