## The warm-up: reading an error

Every error has the same shape. **Read the last line first.**

```
Traceback (most recent call last):
  File "01_name_error.py", line 6, in <module>
    print(record_ID)
NameError: name 'record_ID' is not defined. Did you mean: 'record_id'?
```

| Part | Tells you |
|---|---|
| `NameError` | the **kind** of problem |
| `name 'record_ID' is not defined` | the **detail** |
| `line 6` | **where** to look |

## Warm-up 1 · `01_name_error.py`

The broken file:

```python
record_id = "R-004"
print(record_ID)
```

**What goes wrong:** the variable was created as `record_id` (small *id*), but line 6 asks for `record_ID` (capital *ID*). For Python these are **two different names**, and `record_ID` was never created → `NameError`.

**The fix:** use exactly the same spelling as when you created it.

````python
record_id = "R-004"

print(record_id)
````

> 💡 **Explanation**
>
> **Rule:** Python is **case-sensitive**: `score`, `Score` and `SCORE` are three different names. Most `NameError`s are a typo or a capital letter. Newer Python even suggests the fix at the end: *Did you mean: 'record_id'?*

## Warm-up 2 · `02_type_error.py`

```python
value = input("Value: ")
print(value + 1)
```

Type `5` when asked. You get:

```
TypeError: can only concatenate str (not "int") to str
```

**What goes wrong:** `input()` **always returns text** (`str`), even if you type a number. So `value` is the text `"5"`, and `"5" + 1` means "join some text with a number". Python refuses, because it can't guess whether you want `"51"` or `6`.

**The fix:** turn the text into a number **before** you do maths with it.

````python
value = float(input("Value: "))

print(value + 1)
````

> 💡 **Explanation**
>
> **Why `float` and not `int`?** `float()` accepts both `5` and `5.5`; `int()` only accepts whole numbers. When the user might type a decimal, use `float()`.
>
> Read the fixed line **inside out**: `input()` runs first and gives `"5"`, then `float("5")` gives `5.0`, then the result is stored in `value`.

## Warm-up 3 · `03_value_error.py`

```python
value = int(input("Value: "))
print(value)
```

Type `23.7` when asked:

```
ValueError: invalid literal for int() with base 10: '23.7'
```

**What goes wrong:** the *type* is fine (text can be converted), but the *value* is impossible: `int()` cannot read the text `"23.7"` because of the decimal point. That is a `ValueError`: right type, impossible value.

**The fix:** use `float()`, which understands decimals.

````python
value = float(input("Value: "))

print(value)
````

> 💡 **Explanation**
>
> **The difference between the three errors**
>
> | Error | Meaning | This week's cause |
> |---|---|---|
> | `NameError` | Python doesn't know this name | spelling / capitals |
> | `TypeError` | the kinds of value don't fit together | text + number |
> | `ValueError` | right kind, impossible value | `int("23.7")`, `float("abc")` |
>
> Careful: `int(23.7)` (a **number**, no quotes) works and gives `23` (it cuts off the decimals). Only `int("23.7")` (**text**) fails.

## D3 · Work something out (Threshold)

**What it asks:** using the D2 variables, **calculate** and print (1) the difference between value and limit, (2) the value as a percentage of the limit. Don't type the answers.

**What you need:** arithmetic operators `-`, `/`, `*`, and the percentage formula **part ÷ total × 100**.

(The cell below remembers the variables from D2, like a notebook.)

````python
difference = value - limit
percent = value / limit * 100

print("Difference:", difference)
print("Percent of limit:", percent)
````

> 💡 **Explanation**
>
> **Line by line**
> - `difference = value - limit` → `41.2 - 50` = `-8.8`. Negative means the value is **below** the limit (8.8 under).
> - `percent = value / limit * 100` → `41.2 / 50` = `0.824`, × 100 = `82.4`. So the value is 82.4% of the limit.
> - `print("Difference:", difference)`: a **comma** inside `print` puts a space between the parts and works with any type.
>
> **You may see** `-8.799999999999997` instead of `-8.8`. That is normal: computers store decimals in binary and can't store 8.8 exactly. D5 fixes the display with formatting.
>
> **Why calculate instead of typing `-8.8`?** If the value changes to 45, typed answers become wrong; calculated ones stay correct. That's the whole point of a program.

## D4 · Ask, don't hard-code (Typical)

**What it asks:** same as D2 + D3, but the three values come from `input()` (the user types them) instead of being written in the code. *"Remember what input() gives you back"* = text!

When you run it here, boxes appear for your answers. Try `R-017`, `41.2`, `50`.

## D5 · Make it tidy (Typical)

**What it asks:** take D4's output and format it so (1) every number has exactly **2 decimal places**, (2) numbers are **right-aligned** so the decimal points line up, (3) the difference always shows its sign (`+3.70` / `-8.80`).

**What you need:** **f-strings** with a *format code* after a colon: `f"{number:>10.2f}"`.

````python
record_id = input("Record ID: ")
value = float(input("Value: "))
limit = float(input("Limit: "))

difference = value - limit
percent = value / limit * 100

print(f"Record ID  : {record_id}")
print(f"Value      : {value:>10.2f}")
print(f"Limit      : {limit:>10.2f}")
print(f"Difference : {difference:>+10.2f}")
print(f"Percent    : {percent:>10.2f} %")
````

> 💡 **Explanation**
>
> **Reading the format code `{value:>10.2f}`**
>
> | Part | Meaning |
> |---|---|
> | `f"..."` | an f-string: anything inside `{ }` is calculated and put into the text |
> | `:` | "formatting instructions start here" |
> | `>` | align to the **right** |
> | `10` | use 10 characters of space |
> | `.2f` | show as a decimal number with **2** digits after the point |
> | `+` | always show the sign, `+` or `-` |
>
> Output with 41.2 and 50:
> ```
> Record ID  : R-017
> Value      :      41.20
> Limit      :      50.00
> Difference :      -8.80
> Percent    :      82.40 %
> ```
> All numbers end in the same column, so the decimal points line up. The labels are padded with spaces to the same width (`"Value      :"`), so the numbers also **start** at the same place.
>
> `.2f` also **rounds** for display: `-8.799999...` shows as `-8.80`. The variable itself is unchanged.

## D6 · Split a number (Typical)

**What it asks:** 4831 seconds → print `4831 seconds = 1 hours, 20 minutes, 31 seconds`, using `//` and `%`, starting from the largest unit.

**What you need:**
- `//` **floor division**: how many **whole** times does it fit? `4831 // 3600` = `1`
- `%` **modulo** (remainder): what is **left over**? `4831 % 3600` = `1231`

````python
total_seconds = 4831

hours = total_seconds // 3600
remaining = total_seconds % 3600

minutes = remaining // 60
seconds = remaining % 60

print(f"{total_seconds} seconds = {hours} hours, {minutes} minutes, {seconds} seconds")
````

> 💡 **Explanation**
>
> **Trace it step by step** (do this on paper once):
>
> | Step | Calculation | Result |
> |---|---|---|
> | how many whole hours? | `4831 // 3600` | `1` |
> | seconds left after the hours | `4831 % 3600` | `1231` |
> | how many whole minutes in those? | `1231 // 60` | `20` |
> | seconds left after the minutes | `1231 % 60` | `31` |
>
> Check: 1×3600 + 20×60 + 31 = 3600 + 1200 + 31 = **4831** ✓
>
> **Why "largest unit first"?** Once the hours are taken out, the rest is guaranteed to be less than an hour, so it splits cleanly into minutes and seconds. The same pattern converts pence → pounds and pence, cm → m and cm, etc.
>
> **Shortcut you may see:** `divmod(4831, 3600)` gives both at once: `(1, 1231)`.
>
> **Try it:** change 4831 to 100000. How many hours?

## D7 · Three fields, one calculation (Excellent)

**What it asks:** **one** program that asks for a part, a total and a label, then prints a line like

```
rows missing   :  38 of 250  =  15.2 %
```

with the label **left-aligned in 15 characters**. Test with the three rows of the table (rows missing 38/250, failed logins 12/400, disk used GB 87/120).

**What you need:** everything from D4 + D5, plus left alignment `:<15`. The idea: *every percentage question is the same calculation*, only the words change.

````python
label = input("Label: ")
part = float(input("Part: "))
total = float(input("Total: "))

percent = part / total * 100

print(f"{label:<15}: {part:>3.0f} of {total:.0f}  =  {percent:.1f} %")
````

> 💡 **Explanation**
>
> **Line by line**
> - Three inputs: the label stays text, part and total become numbers.
> - `percent = part / total * 100`: the same formula as D3.
> - `{label:<15}`: `<` = align **left**, in 15 characters. `"rows missing"` is 12 characters, so 3 spaces are added after it. That's what makes the colons line up across the three runs.
> - `{part:>3.0f}`: right-aligned in 3 characters, **0** decimals (shows `38`, not `38.0`).
> - `{total:.0f}`: no decimals. `{percent:.1f}`: one decimal, like the example `15.2`.
>
> **The three tests:**
> ```
> rows missing   :  38 of 250  =  15.2 %
> failed logins  :  12 of 400  =  3.0 %
> disk used GB   :  87 of 120  =  72.5 %
> ```
>
> **Why this is "Excellent":** you had to choose the format codes yourself to match an example, and you saw that one program serves three different fields (data, cyber security, IT). That is what *reusable* means.

## Typical · calculate what you weren't given, 2 decimals, right-aligned

Add a **PROCESS** part between input and output: the difference (rows missing) and the percentage (how complete the data is).

````python
# ===== INPUT
dataset_name = input("Dataset name: ")
rows_loaded = float(input("Rows loaded: "))
rows_expected = float(input("Rows expected: "))

# ===== PROCESS
rows_missing = rows_expected - rows_loaded
percent_loaded = rows_loaded / rows_expected * 100

# ===== OUTPUT
print()
print("=" * 34)
print(f"  RECORD CHECK  -  {dataset_name}")
print("=" * 34)
print(f"  Loaded      : {rows_loaded:>10.2f}")
print(f"  Expected    : {rows_expected:>10.2f}")
print(f"  Missing     : {rows_missing:>10.2f}")
print(f"  Percent     : {percent_loaded:>10.2f} %")
print("=" * 34)
````

> 💡 **Explanation**
>
> **What changed and why**
> - `rows_missing = rows_expected - rows_loaded` → 1200 − 1187 = **13** rows didn't load. (I subtract in this order so "missing" comes out positive when rows are lost.)
> - `percent_loaded = rows_loaded / rows_expected * 100` → 98.92%: how complete the dataset is.
> - Every number now uses `:>10.2f` (D5): right-aligned in 10 characters, 2 decimals, so the decimal points line up.
>
> **The program now has the classic shape:** INPUT → PROCESS → OUTPUT. Almost every program you will write this year has these three parts (that's the "boxes" diagram from the workshop).

## Excellent · sign always shown + one useful line of your own

The brief asks for: the difference **always shows its sign**, one extra calculated line that's genuinely useful for your lane (with a comment saying **why**), and names that need no comment.

````python
"""
RECORD CHECK  -  my version
Name  : Parsa
Lane  : AI / Data Science
Run it:   python template.py
"""

# ===== INPUT
dataset_name = input("Dataset name: ")
rows_loaded = float(input("Rows loaded: "))
rows_expected = float(input("Rows expected: "))

# ===== PROCESS
row_difference = rows_loaded - rows_expected          # negative = rows were lost
percent_loaded = rows_loaded / rows_expected * 100
# Extra line: the share of the data that is MISSING.
# Why: before training a model you must know how much data is lost;
# if too much is missing, the results could be biased.
percent_missing = 100 - percent_loaded

# ===== OUTPUT
print()
print("=" * 34)
print(f"  RECORD CHECK  -  {dataset_name}")
print("=" * 34)
print(f"  Loaded      : {rows_loaded:>10.2f}")
print(f"  Expected    : {rows_expected:>10.2f}")
print(f"  Difference  : {row_difference:>+10.2f}")
print(f"  Complete    : {percent_loaded:>10.2f} %")
print(f"  Missing     : {percent_missing:>10.2f} %")
print("=" * 34)
````

> 💡 **Explanation**
>
> **What changed and why**
> - `row_difference = rows_loaded - rows_expected`: I flipped the order, so the **sign tells the story**: `-13.00` = 13 rows short; `+5.00` would mean 5 extra rows (duplicates?). `:>+10.2f` forces the `+`/`-` to show.
> - `percent_missing = 100 - percent_loaded`: the extra line. The comment explains **why** it matters for my lane (missing data can bias a model), as the brief asks.
> - The docstring at the top (`""" ... """`) fills in the template's Name/Lane header.
>
> Output for `survey_2026`, `1187`, `1200`:
