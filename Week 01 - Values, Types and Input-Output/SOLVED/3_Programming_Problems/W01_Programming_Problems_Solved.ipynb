## Task 1: Questionnaire
Write a program that asks for name, interest, nationality, course, and programming experience, then prints a formatted summary using escape sequences (`\n`, `\t`).

**Example Output:**
```
User Information:
	Name: Alice
	Interest: Security
	Nationality: UK
	Course: Cyber
	Experience: Beginner
```

````python
name = input("Enter your name: ")
interest = input("Enter your interest: ")
nationality = input("Enter your nationality: ")
course = input("Enter your course: ")
experience = input("Enter your programming experience: ")

# \n = new line, \t = tab (indent)
print("User Information:\n\tName: " + name
      + "\n\tInterest: " + interest
      + "\n\tNationality: " + nationality
      + "\n\tCourse: " + course
      + "\n\tExperience: " + experience)
````

> 💡 **Explanation**
>
> **Line by line**
> - The first five lines: each `input()` shows a question and gives back the user's answer **as text (str)**. Because all the answers are text, no conversion (`int`/`float`) is needed.
> - The `print(...)` at the end: we build the whole summary with **one** `print`, so you can see what the escape sequences do:
>   - `\n` means "go to the next line right here".
>   - `\t` means "put a tab here", which is why `Name:` and the others move in a little.
> - Because the sentence is long, we broke it over several lines inside the brackets. Until the bracket is closed, Python treats the next line as part of the same statement.
> - We joined the pieces of text with `+` (this is called concatenation). You can also use an f-string: `f"\tName: {name}"`.
>
> **Traps**
> - `\n` and `\t` only work **inside a string**, that is, between `" "`.
> - If you write `/n` (a forward slash, the wrong way round), Python just prints `/n` itself. It must be a **backslash** `\`.
> - `+` only joins text to text. Here everything is text, so there's no problem.
>
> **Try it:** `Alice`, `Security`, `UK`, `Cyber`, `Beginner` → exactly the example output.

## Task 2: Rectangle Area & Perimeter
Calculate and print the area and perimeter of a rectangle using fixed values: width = 4.5 and height = 7.9.
Formulas: `area = width * height`, `perimeter = 2 * (width + height)`.

**Example Output:**
```
Area = 35.55
Perimeter = 24.80
```

````python
width = 4.5
height = 7.9

area = width * height
perimeter = 2 * (width + height)

print(f"Area = {area:.2f}")
print(f"Perimeter = {perimeter:.2f}")
````

> 💡 **Explanation**
>
> **Line by line**
> - `width = 4.5` and `height = 7.9`: two variables of type `float` (a decimal number).
> - `area = width * height`: a simple multiplication.
> - `perimeter = 2 * (width + height)`: **the brackets matter.** First add, then multiply by 2. Without the brackets (`2 * width + height`), only the width is doubled and the answer comes out wrong.
> - `{area:.2f}`: means "show it with exactly 2 decimal places".
>
> **Why do we need `:.2f`?**
> 1. The computer can't store some decimals exactly. `4.5 * 7.9` is actually `35.550000000000004`. With `:.2f` it becomes `35.55`.
> 2. The perimeter is `24.8`, but the example wants `24.80`. `:.2f` also adds the final zero.
>
> **Hand check:** 4.5 × 7.9 = 35.55 ✓ and 2 × (4.5 + 7.9) = 2 × 12.4 = 24.80 ✓

## Task 3: Convert Pounds to Kilograms
Ask the user for a value in pounds, convert to kilograms using `1 lb = 0.454 kg`, and print the result.

**Example Output:**
```
55.5 pounds is 25.197 kilograms
```

````python
KG_PER_POUND = 0.454

pounds = float(input("Enter a value in pounds: "))
kilograms = pounds * KG_PER_POUND

print(f"{pounds} pounds is {kilograms:.3f} kilograms")
````

> 💡 **Explanation**
>
> **Line by line**
> - `KG_PER_POUND = 0.454`: we gave the conversion factor a name. A name in capital letters means "this is a **constant**, don't change it". This makes the code easier to read than having `0.454` in the middle of the formula.
> - `float(input(...))`: `input()` always returns **text** (`"55.5"`). You can't multiply text, so we turn it into a decimal number with `float()`. `int()` is wrong here, because `55.5` has a decimal part and `int("55.5")` gives an error (`ValueError`).
> - `kilograms = pounds * KG_PER_POUND`: the conversion itself.
> - `{kilograms:.3f}`: three decimal places, because our factor (`0.454`) has three decimal places and the example also shows three.
>
> **Trap:** if the user types `55`, `float` makes it `55.0` and the output is `55.0 pounds is 24.970 kilograms`. This is correct; it just looks different.
>
> **Hand check:** 55.5 × 0.454 = 25.197 ✓

## Task 5: Total Time to Watch a Series
Compute total viewing time (no breaks) for a series:
- 5 seasons, 10 episodes per season
- Season 1: 45 minutes 35 seconds per episode
- Seasons 2–5: 35 minutes 15 seconds per episode

Display the total as hours, minutes, and seconds.

**Example Output:**
```
Total time: 39h 30m 0s
```

````python
episodes_per_season = 10

# Length of one episode, in seconds
season1_episode = 45 * 60 + 35        # 45 min 35 s
later_episode = 35 * 60 + 15          # 35 min 15 s (seasons 2-5)

season1_total = 1 * episodes_per_season * season1_episode
later_total = 4 * episodes_per_season * later_episode
total_seconds = season1_total + later_total

hours = total_seconds // 3600
remaining = total_seconds % 3600
minutes = remaining // 60
seconds = remaining % 60

print(f"Total time: {hours}h {minutes}m {seconds}s")
````

> 💡 **Explanation**
>
> **Main idea: turn everything into seconds, add it up, then turn it back into hours/minutes/seconds.** With seconds, which are a simple whole number, the maths is much easier.
>
> **Line by line**
> - `season1_episode = 45 * 60 + 35`: 45 minutes × 60 = 2700 seconds, plus 35 → `2735` seconds.
> - `later_episode = 35 * 60 + 15` → `2115` seconds.
> - `season1_total = 1 * 10 * 2735` → `27350`. (I put in the × 1 so you can see it is "1 season".)
> - `later_total = 4 * 10 * 2115` → `84600`. Seasons 2 to 5 means **4** seasons, not 5. This is one of the main traps.
> - `total_seconds` → `111950` seconds.
> - Now, like the Week 1 lab (the 4831 seconds example), we start from the biggest unit:
>
> | Step | Calculation | Answer |
> |---|---|---|
> | How many full hours? | `111950 // 3600` | `31` |
> | What is left? | `111950 % 3600` | `350` |
> | How many full minutes? | `350 // 60` | `5` |
> | How many seconds are left? | `350 % 60` | `50` |
>
> **Correct answer:** `Total time: 31h 5m 50s`
>
> **Why is it different from the example (`39h 30m 0s`)?** Because the example doesn't match these rules. I tried a few other readings too: even if **all** 50 episodes were 45 minutes 35 seconds, it would be `37h 59m 10s`. So no sensible reading of the question gets to 39 hours 30 minutes. The example number probably comes from an older version of the question (with different numbers). **What matters is that your method is correct.** If your lecturer asks, show them this hand calculation.
>
> **Traps**
> - Don't mix up `//` and `%`: `//` tells you "how many whole ones", `%` tells you "how much is left over".
> - If you use `/` instead of `//`, you get `31.09...`, not `31`.

## Task 6: Savings Projection (3 Years)
Compute final savings after 3 years given:
- Save £10 every week for 3 years
- Year 1 bonus: +£20 each week
- Every 2 weeks: −£0.50 (sister)

Assume 52 weeks per year; use arithmetic and augmented assignment; no conditionals.

**Example Output:**
```
Final savings after 3 years: £4689.50
```

````python
weeks_per_year = 52
years = 3
total_weeks = weeks_per_year * years      # 156 weeks

savings = 0.0
savings += 10 * total_weeks               # £10 every week for 3 years
savings += 20 * weeks_per_year            # +£20 each week, year 1 only
savings -= 0.50 * (total_weeks // 2)      # -£0.50 every 2 weeks (sister)

print(f"Final savings after 3 years: £{savings:.2f}")
````

> 💡 **Explanation**
>
> **Main idea:** instead of working it out week by week (which needs a loop), we work out each rule **in one go**: "how many times it happens × how much".
>
> **Line by line**
> - `total_weeks = 52 * 3` → `156` weeks.
> - `savings = 0.0`: we start from zero.
> - `savings += 10 * total_weeks`: means `savings = savings + 1560`. This is the augmented assignment the question asks for.
> - `savings += 20 * weeks_per_year`: the bonus is **only in year 1**, so we multiply by 52 (not 156) → `+1040`.
> - `savings -= 0.50 * (total_weeks // 2)`: "every two weeks" means `156 // 2 = 78` times. `78 × 0.50 = 39` → `-39`. Here `//` is the right choice, because the number of times must be a whole number.
> - `£{savings:.2f}`: the pound sign and two decimal places, like real money.
>
> **Hand check:**
>
> | Rule | Calculation | Amount |
> |---|---|---|
> | Weekly saving | 10 × 156 | +1560.00 |
> | Year 1 bonus | 20 × 52 | +1040.00 |
> | Sister's share | 0.50 × 78 | −39.00 |
> | **Total** | | **2561.00** |
>
> So the program's output is: `Final savings after 3 years: £2561.00`
>
> **Why is it different from the example (`£4689.50`)?** With these rules you can't reach that number. Even if you assume the £20 bonus was for **all 3 years**, you get `30 × 156 − 39 = £4641.00`, which still isn't 4689.50. So the example number is wrong (or comes from a different version of the question). We wrote the code to follow the **rules**, which is what gets the marks.
>
> **Traps**
> - Don't multiply the bonus by 156. The question says "Year 1", which means only 52 weeks.
> - `0.50 * total_weeks` is wrong. The money is taken every **two** weeks, not every week.

## Task 7: Cyber: HTTP Log Composer
Ask for IP, method, path, status, and bytes, then print a single combined log line (Apache-style).

**Example Output:**
```
192.168.1.10 - "POST /auth" 401 512
```

````python
ip_address = input("IP address: ")
method = input("HTTP method: ")
path = input("Path: ")
status = int(input("Status code: "))
size_bytes = int(input("Bytes sent: "))

print(f'{ip_address} - "{method} {path}" {status} {size_bytes}')
````

> 💡 **Explanation**
>
> **Line by line**
> - The first three `input`s are text and stay as they are. An IP like `192.168.1.10` has several dots, so it is **not a number**; it is text.
> - `int(input(...))` for status and bytes: these really are numbers. Converting them also has a security benefit: if someone types something strange instead of a number, the program stops with an error right there, and a broken log line is not made.
> - I named the variable `size_bytes`, not `bytes`, because `bytes` is the name of a built-in Python data type. It's better not to use built-in names for your variables.
> - The last line: **quotes inside a string.** We want `"POST /auth"` to be printed with double quotes. The easy way: open and close the whole f-string with **single** quotes `' '`, and then the `"` inside it prints normally.
>   - The second way, with an escape sequence: `f"{ip_address} - \"{method} {path}\" {status} {size_bytes}"`. `\"` means "this quote is not the end of the string, print it".
>
> **Try it:** `192.168.1.10`, `POST`, `/auth`, `401`, `512` → exactly the example output.
>
> **Good to know:** the code `401` means Unauthorized (wrong password). If in a real log you see lots of `POST /auth ... 401` from one IP one after another, someone is probably trying to guess the password (brute force).

## Task 9: IT: INI-style Config Printer
Ask for three `key=value` pairs (via separate prompts) and print them as an INI block with indentation.

**Example Output:**
```
[config]
	max_connections = 200
	timeout = 30
	region = eu-west-2
```

````python
# Version A: key and value asked separately (pure Week 1)
key1 = input("Key 1: ")
value1 = input("Value 1: ")
key2 = input("Key 2: ")
value2 = input("Value 2: ")
key3 = input("Key 3: ")
value3 = input("Value 3: ")

print("[config]")
print(f"\t{key1} = {value1}")
print(f"\t{key2} = {value2}")
print(f"\t{key3} = {value3}")
````

> 💡 **Explanation**
>
> **Version A, line by line**
> - For each pair we have two `input`s: one for the key (`max_connections`) and one for the value (`200`).
> - We didn't turn the values into `int`, because we only print them; we don't do any maths with them. Also, `eu-west-2` isn't a number at all.
> - `print("[config]")`: the section heading.
> - `f"\t{key1} = {value1}"`: first a tab (`\t`) for the indent, then the key, then ` = ` (with a space on each side of the equals sign, like the example), then the value.
>
> **Try it:** `max_connections`, `200`, `timeout`, `30`, `region`, `eu-west-2` → the example output.

````python
# Version B: each pair typed in one go, like  max_connections=200
pair1 = input("Pair 1 (key=value): ")
pair2 = input("Pair 2 (key=value): ")
pair3 = input("Pair 3 (key=value): ")

print("[config]")
print("\t" + pair1.replace("=", " = "))
print("\t" + pair2.replace("=", " = "))
print("\t" + pair3.replace("=", " = "))
````

> 💡 **Explanation**
>
> **Version B: closer to the wording of the question ("three key=value pairs")**
> - The user types each pair in one go: `max_connections=200`.
> - `pair1.replace("=", " = ")`: the string method `replace` swaps every `=` for ` = ` (an equals sign with a space on each side). So `max_connections=200` becomes `max_connections = 200`.
> - `"\t" + ...`: we stick the tab on at the start of the line.
>
> **Trap:** if the user types the spaces themselves (`timeout = 30`), the output becomes `timeout  =  30` (two spaces each side). For this exercise it doesn't matter. Later you can fix it with `.split("=")` and `.strip()`.
>
> **Which one should I write?** If your lecturer said Week 1 tools only, Version A. Version B is shorter and closer to the wording of the question.

## Task 10: Sum of First n Integers
Ask for a positive integer `n` and compute the sum `n*(n+1)//2`. Print the result.

**Example Output:**
```
Sum of 1..10 = 55
```

````python
n = int(input("Enter a positive integer n: "))

total = n * (n + 1) // 2

print(f"Sum of 1..{n} = {total}")
````

> 💡 **Explanation**
>
> **Line by line**
> - `int(input(...))`: `n` must be a whole number.
> - `total = n * (n + 1) // 2`: the famous Gauss formula. It gives the sum without any loop.
> - `f"Sum of 1..{n} = {total}"`: `1..` is part of the fixed text, and only `{n}` and `{total}` change.
>
> **Why `//` and not `/`?** `n * (n + 1)` is always even (of two numbers next to each other, one is always even), so dividing by 2 always comes out exactly. With `/` the answer would be `55.0` (because `/` always gives a `float`), but the example wants `55`. `//` keeps the answer as an `int`.
>
> **Why does the formula work?** Write 1 + 2 + … + 10 twice, once backwards, and add them up one under the other: each column makes 11, and there are 10 columns → 110. Because we wrote it twice, take half: 55 ✓
>
> **Hand check:** 10 × 11 // 2 = 110 // 2 = 55 ✓
>
> **Try it yourself:** in Week 2, work out the same thing with a `for` loop and check that the answer is the same as the formula.

## Task 11: Population Projection (5 Years)
Assume current population is 312032486 and 1 year = 365 days.
Rates: birth every 7s, death every 13s, immigrant every 45s.
Compute projected population for each of the next 5 years using integer division (`//`).

**Example Output:**
```
Year 1: 315,xxx,xxx
...
Year 5: ...
```
*Optional hint:* Compute yearly net change once; then add each year.

````python
population = 312032486
seconds_per_year = 365 * 24 * 60 * 60     # 31,536,000

births = seconds_per_year // 7
deaths = seconds_per_year // 13
immigrants = seconds_per_year // 45
net_change = births - deaths + immigrants

# No loops in Week 1, so one explicit step per year
population += net_change
print(f"Year 1: {population:,}")
population += net_change
print(f"Year 2: {population:,}")
population += net_change
print(f"Year 3: {population:,}")
population += net_change
print(f"Year 4: {population:,}")
population += net_change
print(f"Year 5: {population:,}")
````

> 💡 **Explanation**
>
> **Line by line**
> - `seconds_per_year = 365 * 24 * 60 * 60`: days × hours × minutes × seconds → `31536000` seconds in a year.
> - `births = seconds_per_year // 7`: "how many full 7 seconds are there in a year?" → `4505142` births. Here `//` is needed, because there's no such thing as half a person!
> - `deaths = ... // 13` → `2425846` deaths.
> - `immigrants = ... // 45` → `700800` immigrants.
> - `net_change = births - deaths + immigrants` → `2780096`. Births and immigrants **add**, deaths **take away**. This is exactly the point of the hint: work it out once.
> - `population += net_change`: means `population = population + net_change`. Each year's population is built from **the year before**.
> - `{population:,}`: the comma `,` after the colon splits the digits into groups of three: `314812582` → `314,812,582`. The example looked like this too.
>
> **Why 5 repeated lines and not a loop?** The question says "each of the next 5 years". With Week 1 tools, the only way is to write "add, print" 5 times. Can you see how repetitive it is? This repetition is the reason **loops** exist. In Week 2, all these 10 lines become:
> ```python
> for year in range(1, 6):
>     population += net_change
>     print(f"Year {year}: {population:,}")
> ```
>
> **Output:**
> ```
> Year 1: 314,812,582
> Year 2: 317,592,678
> Year 3: 320,372,774
> Year 4: 323,152,870
> Year 5: 325,932,966
> ```
>
> **About the example (`315,xxx,xxx`):** the real number for year 1 is `314,812,582`, not 315 million and something. The example was only rough: if you round 314.8 million to the nearest million, you get about 315 million. Our calculation follows the rules of the question exactly.
>
> **Hand check for year 1:** 312,032,486 + 2,780,096 = 314,812,582 ✓
>
> **Trap:** if you use `/`, `births` becomes `4505142.857...` and the population comes out with decimals.

## Task 12: Payroll Statement
Ask for: employee name, hours worked, hourly rate, federal tax rate, state tax rate.
Print a formatted payroll slip with gross pay, each deduction, and net pay (2 d.p.).

**Example Output:**
```
Employee: Smith
Gross pay: £97.50
Federal withholding: £19.50
State withholding: £8.78
Net pay: £69.22
```

````python
name = input("Enter employee's name: ")
hours = float(input("Enter number of hours worked: "))
hourly_rate = float(input("Enter hourly pay rate: "))
federal_rate = float(input("Enter federal tax rate (%): "))
state_rate = float(input("Enter state tax rate (%): "))

gross_pay = hours * hourly_rate
federal_withholding = gross_pay * federal_rate / 100
state_withholding = gross_pay * state_rate / 100
net_pay = gross_pay - federal_withholding - state_withholding

print(f"Employee: {name}")
print(f"Gross pay: £{gross_pay:.2f}")
print(f"Federal withholding: £{federal_withholding:.2f}")
print(f"State withholding: £{state_withholding:.2f}")
print(f"Net pay: £{net_pay:.2f}")
````

> 💡 **Explanation**
>
> **Line by line**
> - The name is text. The rest are converted with `float()`, because the hours (for example `7.5`), the pay rate (`9.75`) and the tax rates might have decimals.
> - `gross_pay = hours * hourly_rate`: gross pay = hours × pay rate → 10 × 9.75 = `97.5`.
> - `federal_withholding = gross_pay * federal_rate / 100`: we took the rate as a **percentage** (`20`), so we divide by 100 to make it `0.20` → `19.5`.
> - `state_withholding = ... * 9 / 100` → `8.775`.
> - `net_pay = gross - federal - state`: net pay = gross pay minus all the deductions.
> - `:.2f` on all of them: money is always shown with two decimal places (`97.5` → `97.50`).
>
> **Hand check:**
>
> | Line | Calculation | Result |
> |---|---|---|
> | Gross | 10 × 9.75 | 97.50 |
> | Federal | 97.50 × 20% | 19.50 |
> | State | 97.50 × 9% = 8.775 | 8.78 |
> | Net | 97.50 − 19.50 − 8.78 | 69.22 ✓ |
>
> **Traps**
> - **Percentage or decimal?** Here we take the rate as a percentage (`20`). If your lecturer's version takes the rate as `0.20`, remove the `/ 100`. If you enter `0.20` and the `/ 100` stays, the tax is worked out 100 times too small!
> - **Rounding money:** `8.775` only becomes `8.78` **when it is displayed** with `:.2f`. The variable itself stays `8.775`. That's why `net_pay` is `69.225`, which is displayed as `69.22`, and the slip adds up (97.50 − 19.50 − 8.78 = 69.22). In real banking programs, money is stored in pence (a whole number) so these small problems don't happen.
> - Put the `£` sign **outside** the `{ }`: `£{gross_pay:.2f}`.
