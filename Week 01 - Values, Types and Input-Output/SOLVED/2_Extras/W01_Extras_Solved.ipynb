## Band Name Generator
1. Print a greeting. 2. Ask for the city you grew up in. 3. Ask for a pet's name. 4. Combine them into a band name and print it. 5. Make the next thing typed appear on a new line (`\n`).

```python
print("Welcome to the Band Name Generator.")

city = input("Which city did you grow up in?\n")
pet = input("What is the name of a pet?\n")

band_name = f"{city} {pet}"
print(f"Your band name could be {band_name}")
```

> 💡 **Explanation**
>
> **Line by line**
> - `print("Welcome ...")`: the greeting.
> - `input("Which city did you grow up in?\n")`: this is the trick from step 5. `\n` means "go to the next line". Because we put it at the end of the question text, the cursor moves to the next line and the user types there.
> - `city` and `pet` are text (str), so they don't need any conversion.
> - `f"{city} {pet}"`: joins the two pieces of text with a space between them. You can also do it with `+`: `city + " " + pet`.
>
> **Try it:** `London` and `Rex` → `Your band name could be London Rex`.

## Tip Calculator
A bill of $150.00 split between 5 people with a 12% tip: each pays `(150.00 / 5) * 1.12 = 33.6`, shown as `33.60`. Ask for the bill, the number of people and the tip percentage (whole number), then print each person's share to exactly 2 decimal places.

```python
print("Welcome to the tip calculator!")

bill = float(input("What was the total bill? $"))
people = int(input("How many people to split the bill? "))
tip_percent = int(input("What percentage tip would you like to give? 10, 12 or 15? "))

total_with_tip = bill * (1 + tip_percent / 100)
each_pays = total_with_tip / people

print(f"Each person should pay: ${each_pays:.2f}")
```

> 💡 **Explanation**
>
> **Line by line**
> - `float(input(...))` for the bill, because it might have decimals (`150.50`).
> - `int(input(...))` for the number of people and the percentage, because they are whole numbers.
> - `1 + tip_percent / 100`: a 12% tip means "1.12 times the bill". `12 / 100` is `0.12`, and adding 1 gives `1.12`. This way the bill and the tip are worked out together.
> - `each_pays = total_with_tip / people`: split it between the people.
> - `:.2f`: exactly two decimal places. So `33.6` is shown as `33.60` (Week 1, Topic 4).
>
> **Hand check:** 150 × 1.12 = 168, and 168 ÷ 5 = 33.6 → `33.60` ✓
