## Task 1: Clean Survey Responses
You received messy survey responses. Clean each response by:
1. Removing leading/trailing whitespace
2. Converting to lowercase
3. Printing the cleaned version

**Methods to use:** `strip()`, `lower()`

```python
responses = [
    "  VERY SATISFIED  ",
    "somewhat happy   ",
    "  NOT Happy  ",
    "   NEUTRAL",
    "  dissatisfied  "
]
```

````python
responses = [
    "  VERY SATISFIED  ",
    "somewhat happy   ",
    "  NOT Happy  ",
    "   NEUTRAL",
    "  dissatisfied  "
]

cleaned_responses = []
for response in responses:
    cleaned = response.strip().lower()
    cleaned_responses.append(cleaned)
    print(f"[{response}] -> [{cleaned}]")

print()
print("Cleaned list:", cleaned_responses)
````

> 💡 **Explanation**
>
> **Why clean at all?** To a computer, `"NEUTRAL"`, `"neutral"` and `"   neutral"` are three **completely different** things. If you want to count how many people said "neutral", you will get the wrong number without cleaning. (Week 7 lab drill D1: `"SRV-01" == "srv-01"` is `False`.)
>
> **Line by line**
> - `cleaned_responses = []`: an empty list where we collect the clean answers, so we can work with them later (count them, compare them…).
> - `for response in responses:` goes through the answers one by one.
> - `cleaned = response.strip().lower()`: two **chained** methods. Python runs them from left to right:
>   - First, `response.strip()` removes the spaces on both sides: `"  NOT Happy  "` → `"NOT Happy"`
>   - Then `.lower()` runs on the **result** of the first step: `"NOT Happy"` → `"not happy"`
> - `.strip()` only cleans **both ends**. The space in the **middle** (between `not` and `happy`) stays as it is, which is exactly what we want.
> - `print(f"[{response}] -> [{cleaned}]")`: the trick from Week 7 lab drill D3. We print the text between `[` and `]` so the spaces become **visible**. Without the brackets, a space at the end of a line is invisible.
> - `print("Cleaned list:", cleaned_responses)`: the whole clean list. When Python prints a list, it puts `'...'` around each string.
>
> **Actual output:**
> ```
> [  VERY SATISFIED  ] -> [very satisfied]
> [somewhat happy   ] -> [somewhat happy]
> [  NOT Happy  ] -> [not happy]
> [   NEUTRAL] -> [neutral]
> [  dissatisfied  ] -> [dissatisfied]
>
> Cleaned list: ['very satisfied', 'somewhat happy', 'not happy', 'neutral', 'dissatisfied']
> ```
>
> **Traps**
> - `response.strip()` on its own (without `cleaned = ...`) **does nothing that lasts**. Strings are immutable: the method returns a new string, and if you don't catch it, it is lost.
> - `.lower` without brackets gives you the method itself, not the result. Printing it shows something like `<built-in method lower of str object ...>`.
> - The order of `strip` and `lower` doesn't matter here, but make it a habit to `strip` first.
>
> **Try it yourself:** use `cleaned_responses.count("neutral")` to count how many people said neutral. Now try the same thing on the **original** list and see why it gives `0`.

## Task 2: Validate Student IDs
Write a function `is_valid_id(student_id)` that returns `True` if the student ID:
- Starts with `"M00"`
- Has exactly 9 characters
- The remaining characters after "M00" are all digits

Test it with the list below.

**Methods to use:** `startswith()`, `len()`, `isdigit()`

```python
student_ids = ["M00123456", "M0012345", "X00123456", "M00ABCDEF", "M00987654"]
```

````python
def is_valid_id(student_id):
    """Return True if student_id is "M00" followed by 6 digits (9 characters in total)."""
    if not student_id.startswith("M00"):
        return False
    if len(student_id) != 9:
        return False
    return student_id[3:].isdigit()


student_ids = ["M00123456", "M0012345", "X00123456", "M00ABCDEF", "M00987654"]

for student_id in student_ids:
    if is_valid_id(student_id):
        print(f"{student_id}: valid")
    else:
        print(f"{student_id}: INVALID")
````

> 💡 **Explanation**
>
> **Main idea: "return early".** We check each condition separately. As soon as **one** of them fails, we `return False` and we are done. Only if every check passes do we reach the last line. This makes the code much easier to read than one giant `if` with three `and`s.
>
> **Line by line**
> - `def is_valid_id(student_id):` the function takes a string and returns a **boolean**, that is `True`/`False`. Functions like this, whose names start with `is_` and which give a yes/no answer, are called **predicates**.
> - `if not student_id.startswith("M00"):` `.startswith()` tells you whether the string starts with this piece or not. `not` flips the answer. So this means "if it does **not** start with M00".
> - `return False`: `return` leaves the function at that moment. The lines after it do not run at all.
> - `if len(student_id) != 9:` the length must be exactly 9.
> - `return student_id[3:].isdigit()`:
>   - `student_id[3:]` is a **slice**: from index 3 to the end. Indexes 0, 1 and 2 are the `M`, `0`, `0`. So `"M00123456"[3:]` → `"123456"`.
>   - `.isdigit()` on that piece asks: "is all of it digits?" Its answer is already `True` or `False`, so we return it **directly**. You don't need to write `if ...: return True else: return False`.
> - In the loop, `if is_valid_id(student_id):` uses the function's answer directly inside the `if`.
>
> **Actual output:**
> ```
> M00123456: valid
> M0012345: INVALID
> X00123456: INVALID
> M00ABCDEF: INVALID
> M00987654: valid
> ```
> **Why did each one fail?**
>
> | ID | Which check | Why |
> |---|---|---|
> | `M0012345` | length | it has 8 characters |
> | `X00123456` | `startswith` | it starts with `X` |
> | `M00ABCDEF` | `isdigit` | `"ABCDEF"` is not digits |
>
> **Traps**
> - `startswith` is case-sensitive: `"m00123456"` fails. If you want to accept it, do `student_id.strip().upper()` first.
> - `.isdigit()` gives `False` for an **empty** string. That can't happen here, because we check the length first.
> - `.isdigit()` also gives `False` for a negative number (`"-5"`) and a decimal (`"3.5"`), because `-` and `.` are not digits. Remember this; you will need it in Tasks 8 and 12.
> - `student_id[3:]` is different from `student_id[3]`: the second one is just **one** character (`"1"`).
>
> **Try it yourself:** change the function so that, instead of `False`, it returns the **reason** for failing as a string (for example `"wrong length"`), and an empty string for a correct ID.

## Task 3: Parse CSV Data
Split each row of CSV data and print a formatted summary.

**Methods to use:** `split()`, f-strings

```python
csv_data = [
    "Alice,25,London,Data Analyst",
    "Bob,30,Paris,Engineer",
    "Charlie,22,Berlin,Student"
]

# For each row, split by comma and print:
# "Alice is a 25 year old Data Analyst based in London"
```

````python
csv_data = [
    "Alice,25,London,Data Analyst",
    "Bob,30,Paris,Engineer",
    "Charlie,22,Berlin,Student"
]

for row in csv_data:
    name, age, city, job = row.split(",")
    print(f"{name} is a {age} year old {job} based in {city}")
````

> 💡 **Explanation**
>
> **Line by line**
> - `for row in csv_data:` each `row` is a complete string, like `"Alice,25,London,Data Analyst"`.
> - `row.split(",")` breaks the string wherever it sees a comma and returns a **list**:
>   `['Alice', '25', 'London', 'Data Analyst']`
>   Notice that `"Data Analyst"` stayed as **one** piece, even though it has a space. Why? Because we said to split only at `","`, not at spaces.
> - `name, age, city, job = ...`: this is called **unpacking**. The list has 4 items and we wrote 4 variables. Python hands them out in order: the first → `name`, the second → `age`, …
> - `print(f"...")`: we put the variables into the f-string **in the order of the sentence** (not the order in the data). In the data, `city` comes before `job`, but in the sentence it is the other way round. That is the point of this task.
>
> **Actual output:**
> ```
> Alice is a 25 year old Data Analyst based in London
> Bob is a 30 year old Engineer based in Paris
> Charlie is a 22 year old Student based in Berlin
> ```
>
> **Traps**
> - If the number of pieces is not the same as the number of variables, you get a `ValueError`. This is the same error as in the Week 7 warm-up:
>   - `"Bob,30,Paris"` (3 pieces) → `ValueError: not enough values to unpack (expected 4, got 3)`
>   - `"Bob,30,Paris,Engineer,UK"` (5 pieces) → `ValueError: too many values to unpack (expected 4)`
> - Here `age` is a **string** (`'25'`), not a number. That's fine for printing, but if you want to do maths with it (for example `age + 1`), you must do `int(age)` first.
> - If the data had extra spaces (`"Alice, 25, London"`), `city` would be `" London"` with a space at the start. In that case, `.strip()` each piece (Tasks 8 and 12).
>
> **Try it yourself:** add up the ages with `int(age)` and print the average age.

## Task 4: Count Keywords in Feedback
Count how many times each keyword appears across all feedback. Join all feedback into one string (lowercase), then count each keyword.

**Methods to use:** `lower()`, `count()`

```python
feedback = [
    "The product quality is great",
    "Fast delivery but poor quality",
    "Great product and great delivery",
    "Quality could be better"
]

keywords = ['great', 'quality', 'delivery']
```

````python
feedback = [
    "The product quality is great",
    "Fast delivery but poor quality",
    "Great product and great delivery",
    "Quality could be better"
]

keywords = ['great', 'quality', 'delivery']

all_feedback = " ".join(feedback).lower()
print(all_feedback)
print()

keyword_counts = {}
for keyword in keywords:
    keyword_counts[keyword] = all_feedback.count(keyword)

for keyword in keyword_counts:
    print(f"'{keyword}' appears {keyword_counts[keyword]} times")
````

> 💡 **Explanation**
>
> **Line by line**
> - `" ".join(feedback)`: `.join()` sticks all the items of the list together and puts the string that `.join` is called on **between** each pair; here that is a space `" "`. The shape looks strange at first: the separator comes **before** the dot, and the list goes **inside** the brackets.
> - `.lower()` on the whole string: now `Great` and `great`, and `Quality` and `quality`, are the same.
> - `print(all_feedback)`: only so you can see what the final string looks like.
> - `keyword_counts = {}`: an empty **dictionary (dict)**: key = word, value = count (Week 4).
> - `all_feedback.count(keyword)`: `.count()` on a **string** counts how many times a piece of text appears in it.
> - The second loop goes over the keys of the dictionary and prints the result.
>
> **Actual output:**
> ```
> the product quality is great fast delivery but poor quality great product and great delivery quality could be better
>
> 'great' appears 3 times
> 'quality' appears 3 times
> 'delivery' appears 2 times
> ```
> **Hand check:** `great`: once in sentence 1, twice in sentence 3 (`Great` and `great`) = 3 ✓ · `quality`: sentences 1, 2, 4 = 3 ✓ · `delivery`: sentences 2 and 3 = 2 ✓
>
> **Traps**
> - **Without `.lower()`**, the answer for `great` would be 2, because `"Great"` is different from `"great"`.
> - **Why `" ".join` and not `"".join`?** Without a space, the end of one sentence sticks to the start of the next: `...is greatfast delivery...`. Here, by luck, nothing breaks, but two pieces could join to make a **fake** keyword.
> - **`.count()` on a string counts pieces of text, not whole words.** If one piece of feedback said `"the greatest"`, `great` would be counted inside it. To count **whole words**, `split()` first and count on the **list**: `all_feedback.split().count("great")`. You will see this in Tasks 9 and 14.
>
> **Try it yourself:** add the keyword `"product"` to the list. Its answer should be 2.

## Task 5: Format Customer Names
Clean and format each name to **Title Case**, then create an email by taking the first name (lowercase) + `.` + surname (lowercase) + `@company.com`.

**Methods to use:** `strip()`, `title()`, `lower()`, `split()`, string concatenation

```python
raw_names = [
    "  JOHN smith  ",
    "mary JOHNSON",
    "  bob WILSON  "
]

# Expected output:
# John Smith → john.smith@company.com
# Mary Johnson → mary.johnson@company.com
# Bob Wilson → bob.wilson@company.com
```

````python
raw_names = [
    "  JOHN smith  ",
    "mary JOHNSON",
    "  bob WILSON  "
]

for raw_name in raw_names:
    clean_name = raw_name.strip().title()
    first_name, surname = clean_name.split()
    email = first_name.lower() + "." + surname.lower() + "@company.com"
    print(f"{clean_name} → {email}")
````

> 💡 **Explanation**
>
> **Line by line**
> - `raw_name.strip().title()`: first remove the spaces on both sides, then Title Case. `"  JOHN smith  "` → `"JOHN smith"` → `"John Smith"`. `.title()` makes every letter that **starts a word** uppercase and **all the others** lowercase, even if they were uppercase before.
> - `clean_name.split()`: with no argument, it splits the string at **any amount of space** (Week 7 lab drill D9): `"John Smith"` → `['John', 'Smith']`.
> - `first_name, surname = ...`: unpacking into two variables, like Task 3.
> - `first_name.lower() + "." + surname.lower() + "@company.com"`: we join four strings with `+`. Between two strings, `+` means "stick together", not "add up".
> - `print(f"{clean_name} → {email}")`: the `→` sign is a Unicode character, and Python 3 has no problem with it. Just copy it straight from the task statement.
>
> **Actual output:**
> ```
> John Smith → john.smith@company.com
> Mary Johnson → mary.johnson@company.com
> Bob Wilson → bob.wilson@company.com
> ```
>
> **Traps**
> - A three-part name like `"Mary Ann Smith"` gives the error `too many values to unpack` with `first_name, surname = ...`. The fix: `parts = clean_name.split()`, then `first_name = parts[0]` and `surname = parts[-1]` (`[-1]` means the **last** item).
> - `.title()` plays tricks with apostrophes: `"o'brien".title()` → `"O'Brien"` (good), but `"it's".title()` → `"It'S"` (bad). For names, this is usually not a problem.
> - `.split(" ")` (with one space) is different from `.split()`: if there are **two** spaces between two words, `.split(" ")` puts an empty string `''` in the middle of the list. For messy text, always prefer the empty `.split()`.
>
> **Try it yourself:** build the same email with an f-string: `f"{first_name.lower()}.{surname.lower()}@company.com"`. Which one is easier to read?

## Task 6: Classify Server Log Entries
Write a function `classify_logs(logs)` that takes a list of log strings and:
1. Counts how many `ERROR`, `INFO`, and `WARNING` entries there are
2. Extracts and prints the **timestamp** from each entry
3. Returns the three counts

**Methods to use:** `startswith()`, `split()`, `strip()`

```python
logs = [
    "[14:32:05] ERROR: Disk full",
    "[14:32:10] INFO: Server started",
    "[14:33:01] WARNING: High memory usage",
    "[14:33:15] INFO: Backup complete",
    "[14:34:00] ERROR: Connection refused",
    "[14:35:22] WARNING: CPU at 90%"
]
```

````python
def classify_logs(logs):
    """Print each entry's timestamp and return the ERROR, INFO and WARNING counts."""
    error_count = 0
    info_count = 0
    warning_count = 0

    for entry in logs:
        # "[14:32:05] ERROR: Disk full" -> "[14:32:05" and " ERROR: Disk full"
        timestamp_part, message = entry.split("]", 1)
        timestamp = timestamp_part.strip("[")
        message = message.strip()

        if message.startswith("ERROR"):
            error_count += 1
        elif message.startswith("INFO"):
            info_count += 1
        elif message.startswith("WARNING"):
            warning_count += 1

        print(f"Timestamp: {timestamp}  |  {message}")

    return error_count, info_count, warning_count


logs = [
    "[14:32:05] ERROR: Disk full",
    "[14:32:10] INFO: Server started",
    "[14:33:01] WARNING: High memory usage",
    "[14:33:15] INFO: Backup complete",
    "[14:34:00] ERROR: Connection refused",
    "[14:35:22] WARNING: CPU at 90%"
]

error_total, info_total, warning_total = classify_logs(logs)
print()
print(f"ERROR: {error_total}   INFO: {info_total}   WARNING: {warning_total}")
````

> 💡 **Explanation**
>
> **Main idea:** each log line has two parts: the **time** inside square brackets, and the **message**. First we separate these two, then we use `startswith` on the message to find its type.
>
> **Line by line**
> - Three counters start at `0`.
> - `entry.split("]", 1)`: splits at the `]` character. The number `1` is **`maxsplit`** (Week 7 lab drill D11): "split **at most once**". The result is always exactly two pieces:
>   `"[14:32:05] ERROR: Disk full"` → `['[14:32:05', ' ERROR: Disk full']`
>   Even if the message itself contains a `]`, it won't be split again, so the unpacking won't break.
> - `timestamp_part.strip("[")`: with an argument, `.strip()` removes **that character** from both ends instead of spaces (Week 7 lab drill D4): `"[14:32:05"` → `"14:32:05"`.
> - `message.strip()`: removes the space at the start of the message: `" ERROR: Disk full"` → `"ERROR: Disk full"`. **This line is very important:** without it, `" ERROR...".startswith("ERROR")` is `False`, because the string starts with a **space**!
> - `if / elif / elif` with `startswith`: finds the message type and adds `+1` to the right counter.
> - `return error_count, info_count, warning_count`: Python lets you return **several** values separated by commas. What actually comes back is a **tuple**: `(2, 2, 2)`.
> - `error_total, info_total, warning_total = classify_logs(logs)`: we unpack that tuple into three variables.
>
> **Actual output:**
> ```
> Timestamp: 14:32:05  |  ERROR: Disk full
> Timestamp: 14:32:10  |  INFO: Server started
> Timestamp: 14:33:01  |  WARNING: High memory usage
> Timestamp: 14:33:15  |  INFO: Backup complete
> Timestamp: 14:34:00  |  ERROR: Connection refused
> Timestamp: 14:35:22  |  WARNING: CPU at 90%
>
> ERROR: 2   INFO: 2   WARNING: 2
> ```
>
> **Traps**
> - **Don't split at `":"`!** You might be tempted to write `entry.split(":")`. But the time itself has two `:` signs (`14:32:05`), so the result is `['[14', '32', '05] ERROR', ' Disk full']`. Before you choose a separator, always look at the data and see **where** that character appears.
> - `"ERROR" in message` also works, but if an INFO message says `"INFO: no ERROR found"`, it will be counted as an ERROR by mistake. `startswith` only looks at the **start** of the string, so it is more accurate.
> - Don't name a variable `warnings`, because Python has a standard module with that name. That's why we used `warning_total`.
>
> **Try it yourself:** add an `else` that counts unknown lines (like `"[15:00:00] DEBUG: test"`) in an `other_count` counter.

## Task 7: Simple Sentiment Analyser
Write a function `analyse_sentiment(reviews)` that:
1. Cleans each review (`strip()`, `lower()`)
2. Classifies it as **positive**, **negative**, or **neutral** using keyword lists
3. Returns a dictionary with counts and a list of which reviews are negative

**Methods/Functions to use:** `strip()`, `lower()`, `any()`, `in`

```python
reviews = [
    "  AMAZING product, Love it!!!  ",
    "terrible experience   ",
    "  It was okay  ",
    "WORST purchase ever",
    "  fantastic quality  ",
    "poor customer service",
    "  Not bad, not great  "
]

positive_words = ['amazing', 'love', 'fantastic', 'great', 'excellent']
negative_words = ['terrible', 'worst', 'poor', 'awful', 'horrible']
```

````python
positive_words = ['amazing', 'love', 'fantastic', 'great', 'excellent']
negative_words = ['terrible', 'worst', 'poor', 'awful', 'horrible']


def has_word(text, words):
    """Return True if any word from words is in text without "not " just before it."""
    return any(word in text and f"not {word}" not in text for word in words)


def analyse_sentiment(reviews):
    """Classify each review and return a dict of counts plus the list of negative reviews."""
    result = {"positive": 0, "negative": 0, "neutral": 0, "negative_reviews": []}

    for review in reviews:
        clean_review = review.strip().lower()
        is_positive = has_word(clean_review, positive_words)
        is_negative = has_word(clean_review, negative_words)

        if is_positive and not is_negative:
            label = "positive"
        elif is_negative and not is_positive:
            label = "negative"
            result["negative_reviews"].append(clean_review)
        else:
            label = "neutral"

        result[label] += 1
        print(f"{label:<9} <- {clean_review}")

    return result


reviews = [
    "  AMAZING product, Love it!!!  ",
    "terrible experience   ",
    "  It was okay  ",
    "WORST purchase ever",
    "  fantastic quality  ",
    "poor customer service",
    "  Not bad, not great  "
]

summary = analyse_sentiment(reviews)
print()
print(f"Positive: {summary['positive']}")
print(f"Negative: {summary['negative']}")
print(f"Neutral:  {summary['neutral']}")
print("Negative reviews:")
for review in summary["negative_reviews"]:
    print(f"  - {review}")
````

> 💡 **Explanation**
>
> **Main idea:** a review is **positive** if it has at least one positive word and no negative word. It is **negative** if it's the other way round. Every other case (neither, or both) is **neutral**.
>
> **The helper function `has_word`**
> - `any(... for word in words)`: `any` means "at least one?". The thing inside the brackets is a **generator expression**. It is like the list comprehension from Week 4, but without `[ ]` (the same shape you saw in Week 7 lab drill D13 with `",".join(str(n) for n in numbers)`). It means: "for each `word` in the list, work out this condition". `any` stops at the first `True` and returns `True`.
> - The condition: `word in text and f"not {word}" not in text`
>   - `word in text`: is the word in the text? (`in` on a string, topic 2 of this week)
>   - `f"not {word}" not in text`: and is **"not + that same word"** **not** in the text?
> - **Why the second part?** Look at the last review: `"not bad, not great"`. The word `great` is in it, so a simple version would call it **positive**, which is wrong! With this check, a `great` that comes straight after `not ` is not counted, and this review becomes **neutral**, which makes more sense.
>
> **The main function, line by line**
> - `result = {..., "negative_reviews": []}`: a dictionary that holds three counters and an empty list. The value of a key in a dictionary can be **anything**, even a list.
> - `clean_review = review.strip().lower()`: step 1. Now `"AMAZING"` and `"amazing"` are the same, because the word lists are lowercase.
> - `is_positive` and `is_negative`: two booleans.
> - `if is_positive and not is_negative:` only positive → positive.
> - `elif is_negative and not is_positive:` only negative → negative, and right here we add the text to `result["negative_reviews"]`.
> - `else:` neither (`"it was okay"`) or both (for example `"great idea, poor quality"`) → neutral.
> - `result[label] += 1`: a neat trick. Because the category name (`label`) is the same as the dictionary key, we add `+1` to that key directly instead of writing three `if`s.
> - `f"{label:<9}"`: `:<9` means "write it **left-aligned** in 9 spaces". This makes the arrows line up under each other.
>
> **Actual output:**
> ```
> positive  <- amazing product, love it!!!
> negative  <- terrible experience
> neutral   <- it was okay
> negative  <- worst purchase ever
> positive  <- fantastic quality
> negative  <- poor customer service
> neutral   <- not bad, not great
>
> Positive: 2
> Negative: 3
> Neutral:  2
> Negative reviews:
>   - terrible experience
>   - worst purchase ever
>   - poor customer service
> ```
>
> **Traps**
> - **Without `.lower()`**, the review `"WORST purchase ever"` would not be found as negative, because `"worst" in "WORST purchase ever"` → `False`.
> - `in` on a string finds **pieces of text**, not whole words: `"love" in "glove"` → `True`! For this small task it's not a problem, but with real data you should `split()` and check the words one by one.
> - Inside an f-string, use **different quotes** for the dictionary key: `f"{summary['positive']}"`. If you use `"` outside and `"` inside too, you get an error in older versions of Python.
> - This "word list" method is very simple (for example, `"not bad"` is actually a **positive** feeling!). Real sentiment models use machine learning. But the basic idea is the same.
>
> **Try it yourself:** add `"good"` and `"bad"` to the lists. What happens to `"not bad, not great"` now? Why?

## Task 8: Data Validation Pipeline
Write a function `validate_record(record)` that takes a CSV string and checks:
1. It has exactly 4 fields (name, age, email, phone)
2. Name contains only letters and spaces
3. Age is a valid number between 18 and 100
4. Email contains `@` and `.`
5. Phone starts with `+` or `0`

Return a list of **error messages** (empty list = valid record).

Use `all()` to check if ALL fields are non-empty first.

**Methods to use:** `split()`, `isdigit()`, `isalpha()`, `startswith()`, `all()`

```python
records = [
    "Alice Smith,25,alice@email.com,+447123456789",
    "Bob,abc,bob@email,07123456789",
    "Charlie123,30,charlie@mail.com,447123456789",
    ",22,missing@mail.com,+440000000000"
]
```

````python
FIELD_NAMES = ["name", "age", "email", "phone"]


def validate_record(record):
    """Return a list of error messages for one CSV record (an empty list means valid)."""
    errors = []
    fields = [field.strip() for field in record.split(",")]

    # 1. Exactly 4 fields
    if len(fields) != 4:
        errors.append(f"Expected 4 fields, found {len(fields)}")
        return errors

    # Every field must be non-empty before we check anything else
    if not all(fields):
        for index in range(len(fields)):
            if fields[index] == "":
                errors.append(f"The {FIELD_NAMES[index]} field is empty")
        return errors

    name, age, email, phone = fields

    # 2. Name: only letters and spaces
    if not name.replace(" ", "").isalpha():
        errors.append(f"Name '{name}' must contain only letters and spaces")

    # 3. Age: a whole number between 18 and 100
    if not age.isdigit():
        errors.append(f"Age '{age}' is not a number")
    elif not 18 <= int(age) <= 100:
        errors.append(f"Age {age} is not between 18 and 100")

    # 4. Email: needs "@" and "."
    if "@" not in email or "." not in email:
        errors.append(f"Email '{email}' must contain '@' and '.'")

    # 5. Phone: starts with "+" or "0"
    if not (phone.startswith("+") or phone.startswith("0")):
        errors.append(f"Phone '{phone}' must start with '+' or '0'")

    return errors


records = [
    "Alice Smith,25,alice@email.com,+447123456789",
    "Bob,abc,bob@email,07123456789",
    "Charlie123,30,charlie@mail.com,447123456789",
    ",22,missing@mail.com,+440000000000"
]

for record in records:
    problems = validate_record(record)
    print(record)
    if len(problems) == 0:
        print("  VALID")
    else:
        for problem in problems:
            print(f"  - {problem}")
````

>
> **Main idea: "collect all the errors".** Unlike Task 2, which returned `False` at the first problem, here we want to tell the user about **all** the problems at once (like sign-up forms that write in red under every wrong field). So we keep an `errors` list, and every check that fails adds a message to it.
>
> **Line by line**
> - `FIELD_NAMES = [...]`: the field names, in order. We wrote it in capital letters because it is a **constant**, meaning it is not supposed to change (this is a convention).
> - `fields = [field.strip() for field in record.split(",")]`: a **list comprehension** (Week 4). It splits the row at the commas and strips **every** piece. This way `"Bob, 30"` also works correctly.
> - **Check 1:** if the number of fields is not 4, the other checks make no sense (we don't know which piece is the age!). So we `return errors` right there.
> - **`all(fields)`**: `all` means "are all of them true?". In Python, an **empty** string `""` counts as `False`, and every other string counts as `True`. So `all(fields)` means "no field is empty". For the last row: `all(['', '22', ...])` → `False`.
> - If a field is empty, we loop over the **indexes** to find **which** one is empty, and get its name from `FIELD_NAMES[index]`. Then we `return`, because checking an empty name (for example with `isalpha`) would only add a repeated error.
> - **Check 2:** `name.replace(" ", "").isalpha()`: the problem is that `"Alice Smith".isalpha()` is `False`, because **a space is not a letter**! So first we remove the spaces with `.replace(" ", "")` (`"AliceSmith"`), then we check.
> - **Check 3:** two steps:
>   - `if not age.isdigit():` first make sure it is a number. `"abc".isdigit()` → `False`.
>   - `elif not 18 <= int(age) <= 100:` **only if** it is a number, convert it and check the range. `18 <= x <= 100` is a **chained comparison**, meaning "x is between 18 and 100". If this were an `if` and not an `elif`, `int("abc")` would crash (`ValueError`).
> - **Check 4:** `"@" not in email or "." not in email`: if **either** of these two is missing, it's an error. (De Morgan's law: the opposite of "both are there" is "at least one is missing".)
> - **Check 5:** `phone.startswith("+") or phone.startswith("0")`, in brackets with `not` in front. A shorter way: `startswith` also accepts a **tuple**: `phone.startswith(("+", "0"))`.
>
> **Actual output:**

>   - Email 'bob@email' must contain '@' and '.'

> Look: Bob and Charlie each have **two** errors, and both were reported.
>
> **Traps**
> - `isalpha()` and `isdigit()` give `False` on an **empty** string. That's why the `all()` check comes **first**.
> - Don't call `int(age)` **before** `isdigit()`, or you will crash on `"abc"`.
> - `"." in email` is very simple: `"bob.smith@email"` is accepted, because the dot comes **before** the `@`. Task 12 checks this more accurately with `find()`.
> - `if not phone.startswith("+") or phone.startswith("0")` (without brackets) is **wrong**, because `not` only works on the first part.

## Task 10: Password Strength Checker
Write a function `check_password(password)` that scores a password (0–5):

| Rule | Points |
|------|--------|
| At least 8 characters | +1 |
| Contains uppercase letter | +1 |
| Contains lowercase letter | +1 |
| Contains a digit | +1 |
| Contains a special character (`@#$%!`) | +1 |

Print the score and a label: **Weak** (0-2), **Medium** (3-4), **Strong** (5).

If the password contains a space, print a message and reject it.

**Methods to use:** `len()`, `isupper()`, `islower()`, `isdigit()`, `any()`

```python
test_passwords = [
    "hello",
    "Hello123",
    "My Pass!",
    "Str0ng@Pass",
    "12345678"
]
```

````python
SPECIAL_CHARACTERS = "@#$%!"


def check_password(password):
    """Score password from 0 to 5, print the score and label, and return the score."""
    if " " in password:
        print(f"{password:<12} REJECTED: a password cannot contain spaces")
        return 0

    score = 0
    if len(password) >= 8:
        score += 1
    if any(char.isupper() for char in password):
        score += 1
    if any(char.islower() for char in password):
        score += 1
    if any(char.isdigit() for char in password):
        score += 1
    if any(char in SPECIAL_CHARACTERS for char in password):
        score += 1

    if score <= 2:
        label = "Weak"
    elif score <= 4:
        label = "Medium"
    else:
        label = "Strong"

    print(f"{password:<12} score {score}/5  ->  {label}")
    return score


test_passwords = [
    "hello",
    "Hello123",
    "My Pass!",
    "Str0ng@Pass",
    "12345678"
]

for password in test_passwords:
    check_password(password)
````

> 💡 **Explanation**
>
> **Main idea:** five **separate** `if`s (not `elif`!), because each rule is independent of the others, and one password can meet all of them at the same time. Each one that is true gives `score += 1`.
>
> **Line by line**
> - `SPECIAL_CHARACTERS = "@#$%!"`: we put the special characters in a **string**. Because `in` works on strings, `"@" in SPECIAL_CHARACTERS` → `True`. No list is needed.
> - `if " " in password:` first of all we check for a space. If there is one, we print a message and `return 0`. The rest of the function does not run.
> - `len(password) >= 8`: rule 1.
> - `any(char.isupper() for char in password)`: rule 2. It goes through every character of the password and asks "is it uppercase?". If **at least one** is `True`, `any` gives `True`.
>   - **Why not `password.isupper()`?** Because `"Hello123".isupper()` is `False`. That asks "are **all** the letters uppercase?", but we want to know "is there **at least one** uppercase letter?". That is the difference `any` makes.
> - Rules 3 and 4 have exactly the same shape, with `islower` and `isdigit`.
> - `any(char in SPECIAL_CHARACTERS for char in password)`: rule 5. For each character of the password it asks "is it in the string of special characters?".
> - `if score <= 2 / elif score <= 4 / else`: the label. Because `elif` is only checked when the condition before it was false, `score <= 4` really means "3 or 4".
> - `f"{password:<12}"`: writes the password left-aligned in 12 spaces, so the columns line up.
>
> **Actual output:**
> ```
> hello        score 1/5  ->  Weak
> Hello123     score 4/5  ->  Medium
> My Pass!     REJECTED: a password cannot contain spaces
> Str0ng@Pass  score 5/5  ->  Strong
> 12345678     score 2/5  ->  Weak
> ```
> **Hand check:**
>
> | Password | 8+ | Upper | Lower | Digit | Special | Total |
> |---|---|---|---|---|---|---|
> | `hello` | ✗ | ✗ | ✓ | ✗ | ✗ | 1 |
> | `Hello123` | ✓ | ✓ | ✓ | ✓ | ✗ | 4 |
> | `Str0ng@Pass` | ✓ | ✓ | ✓ | ✓ (`0`) | ✓ (`@`) | 5 |
> | `12345678` | ✓ | ✗ | ✗ | ✓ | ✗ | 2 |
>
> **Traps**
> - If you use `elif` instead of five `if`s, only the **first** true rule gets a point and the others are not checked. `"Hello123"` would get 1 instead of 4.
> - The `"0"` (digit zero) in `Str0ng` is different from `"O"` (the letter O). One is `isdigit` and the other is `isupper`.
> - Whoever calls this function can't tell whether `0` means "rejected" or "a score of zero". In a real program it would be better to report a rejection separately (for example `return None` or `-1`). Here, because we only print, `0` is enough.

## Task 11: Caesar Cipher Encoder & Decoder
Write **two** functions:
1. `encode(text, key)` — shifts each letter forward by `key` positions
2. `decode(text, key)` — reverses the shift to get the original text

Rules:
- Only shift letters (a-z). Leave spaces, digits, and punctuation unchanged
- Wrap around: `z` shifted by 1 → `a`
- Preserve case: uppercase stays uppercase
- Verify: `decode(encode(text, key), key)` should return the original text

**Methods to use:** `ord()`, `chr()`, `isalpha()`, `isupper()`, `islower()`

```python
original = "Hello World! 123"
key = 5
```

````python
def encode(text, key):
    """Return text with every letter a-z / A-Z shifted forward by key places."""
    result = ""
    for char in text:
        if not char.isalpha():
            result += char              # space, digit, punctuation: unchanged
            continue

        if char.isupper():
            base = ord("A")             # 65
        else:
            base = ord("a")             # 97

        position = ord(char) - base     # 'a' -> 0, 'b' -> 1, ... 'z' -> 25
        new_position = (position + key) % 26
        result += chr(base + new_position)
    return result


def decode(text, key):
    """Return text with every letter shifted back by key places (undoes encode)."""
    return encode(text, -key)


original = "Hello World! 123"
key = 5

secret = encode(original, key)
back = decode(secret, key)

print(f"Original: {original}")
print(f"Encoded:  {secret}")
print(f"Decoded:  {back}")
print(f"decode(encode(text, key), key) == text? {back == original}")

print()
print(encode("xyz XYZ", 3))        # wrap-around check: abc ABC
print(decode("abc ABC", 3))        # and back: xyz XYZ
````

> 💡 **Explanation**
>
> **Two new functions: `ord()` and `chr()`**
> - The computer stores every character as a **number** (its Unicode code).
> - `ord("a")` → `97` means "what is the number code of this character?"
> - `chr(97)` → `"a"` is the opposite: "which character is this number?"
> - The letters come one after another: `a`=97, `b`=98, … `z`=122. Uppercase letters are separate: `A`=65, … `Z`=90.
>
> **Line by line**
> - `result = ""` and the loop over each `char`: like the Extras project, we build the answer one character at a time.
> - `if not char.isalpha(): result += char; continue`: if it's not a letter, add it **as it is** and move on to the next character. `continue` stops the rest of the loop body from running for this character.
> - `if char.isupper(): base = ord("A") else: base = ord("a")`: we choose the **starting point** of the right alphabet. This line is what keeps the letter's case: an uppercase letter stays in the range 65 to 90, a lowercase one in 97 to 122.
> - `position = ord(char) - base`: turns the code into "the letter's place in the alphabet" (0 to 25). For example, `ord("H") - 65 = 72 - 65 = 7`.
> - `new_position = (position + key) % 26`: the shift and the wrap-around. `% 26` always keeps the answer between 0 and 25.
> - `result += chr(base + new_position)`: the way back: new place + starting point = new code → character.
> - `decode` is just `encode` with a **negative** key. In Python, `%` gives a positive answer for negative numbers too: `(0 - 3) % 26` → `23`, so `a` shifted by `-3` becomes `x`. This means wrapping around backwards is automatically correct as well.
> - `back == original`: the check. `==` between two strings gives `True`/`False`.
>
> **Actual output:**
> ```
> Original: Hello World! 123
> Encoded:  Mjqqt Btwqi! 123
> Decoded:  Hello World! 123
> decode(encode(text, key), key) == text? True
>
> abc ABC
> xyz XYZ
> ```
> **Hand check:** `H`: 72−65 = 7 → 7+5 = 12 → 65+12 = 77 → `M` ✓ · `o`: 111−97 = 14 → 19 → 116 → `t` ✓ · `W`: 87−65 = 22 → 27 % 26 = 1 → 66 → `B` ✓ (it wrapped around!)
>
> **Comparison with the Extras project:**
>
> | | Extras project | This task |
> |---|---|---|
> | Finding the letter's place | `alphabet.index(char)` | `ord(char) - base` |
> | Turning it back into a letter | `alphabet[new_position]` | `chr(base + new_position)` |
> | Keeping the case | `.lower()` then `.upper()` | two separate `base`s |
>
> Both are correct. The `ord`/`chr` way is faster and doesn't need an alphabet string.
>
> **Traps**
> - Without `% 26`: `z` + 1 → `chr(123)` → `"{"`. That's a symbol, not `a`!
> - If you always set `base` to `ord("a")`, `H` (72) becomes 72−97 = −25 and the answer is completely messed up.
> - `isalpha()` also gives `True` for non-English letters (like `é`, or Arabic and Persian letters), but our calculation is only correct for a to z. For English text this is not a problem.
>

````python
FIELD_NAMES = ["name", "age", "email", "country"]


def clean_dataset(records):
    """Clean and validate |-separated records; return (valid_records, invalid_records)."""
    valid_records = []
    invalid_records = []

    for record in records:
        fields = [field.strip() for field in record.split("|")]

        if len(fields) != 4:
            invalid_records.append({"record": record.strip(),
                                    "reasons": [f"expected 4 fields, found {len(fields)}"]})
            continue

        # 1. Skip records with any empty field
        if not all(fields):
            reasons = []
            for index in range(4):
                if fields[index] == "":
                    reasons.append(f"{FIELD_NAMES[index]} is empty")
            invalid_records.append({"record": record.strip(), "reasons": reasons})
            continue

        name, age, email, country = fields
        reasons = []

        # 2. Name: Title Case, and only one space between words
        name = " ".join(name.split()).title()

        # 3. Age: a number from 18 to 120
        if not age.isdigit():
            reasons.append(f"age '{age}' is not a number")
        elif not 18 <= int(age) <= 120:
            reasons.append(f"age {age} is not between 18 and 120")

        # 4. Email: an "@" and then a "." somewhere after it
        at_position = email.find("@")
        if at_position == -1 or email.find(".", at_position) == -1:
            reasons.append(f"email '{email}' is not valid")

        # 5. Country in uppercase
        country = country.upper()

        if len(reasons) == 0:
            valid_records.append({"name": name, "age": int(age),
                                  "email": email, "country": country})
        else:
            invalid_records.append({"record": record.strip(), "reasons": reasons})

    return valid_records, invalid_records


raw_data = [
    "  john SMITH  | 25 | john.smith@email.com | united kingdom ",
    "  MARY  johnson | abc | mary@email | france",
    " | 30 | test@test.com | germany",
    "  bob wilson  | 17 | bob@mail.com | spain  ",
    "  alice Brown  | 28 | alice.b@data.org | italy  ",
    "charlie DAVIS | 35 | | usa"
]

valid, invalid = clean_dataset(raw_data)

print(f"VALID RECORDS ({len(valid)})")
for person in valid:
    print(f"  {person['name']:<14} {person['age']:>3}  {person['email']:<22} {person['country']}")

print()
print(f"INVALID RECORDS ({len(invalid)})")
for item in invalid:
    print(f"  {item['record']}")
    print(f"     reason: {'; '.join(item['reasons'])}")
````

> 💡 **Explanation**
>
> **Main idea:** this is Task 8 again, but one step further: as well as **checking** the data, we also **clean** it and split the rows into two groups. A good row is stored as a **clean dictionary** (ready to use), and a bad row is stored together with its reasons. This structure is called a **pipeline** (like a production line): the data goes in at one end, passes through several stages, and comes out clean at the other end.
>
> **Line by line**
> - `fields = [field.strip() for field in record.split("|")]`: we split at `|` and strip each piece. `" | 30 | ..."` → `['', '30', 'test@test.com', 'germany']`. The first field became **empty** after the strip.
> - The check on the number of fields: a guard for completely broken rows, so the unpacking further down doesn't crash.
> - **Step 1:** `if not all(fields):` the same trick as Task 8: an empty string means `False`. We loop over the indexes to find the name of the empty field, add the row to `invalid_records`, and use `continue` to go to the next row.
> - `name, age, email, country = fields`: unpacking.
> - `reasons = []`: the list of reasons for **this** row. Each row starts from zero.
> - **Step 2:** `" ".join(name.split()).title()`: a well-known trick for removing spaces in the **middle**:
>   - `"MARY  johnson".split()` → `['MARY', 'johnson']` (the two spaces in the middle disappear completely, because the empty `split()` treats any amount of space as one separator)
>   - `" ".join(...)` → `"MARY johnson"` (now exactly **one** space)
>   - `.title()` → `"Mary Johnson"`
>   - `.strip()` alone was not enough, because it only cleans **both ends**, not the middle.
> - **Step 3:** as in Task 8: first `isdigit()`, then check the range with `elif`.
> - **Step 4:** the email, with `find()`:
>   - `at_position = email.find("@")`: finds the position of `@`. If it isn't there, `-1` (Week 7 lab drill D6).
>   - `email.find(".", at_position)`: `find` also takes a **second argument**: "search from this index onwards". So it only finds dots that come **after** the `@`.
>   - `"mary@email"`: the `@` is at index 4, but there is no dot from 4 onwards → `-1` → invalid ✓
>   - `"john.smith@email.com"`: the dot **before** the `@` (in `john.smith`) doesn't count, but `.com` comes after it → valid ✓
> - **Step 5:** `country.upper()`: `"united kingdom"` → `"UNITED KINGDOM"`.
> - At the end of the loop: if `reasons` is empty, a clean dictionary is added to `valid_records` (with `int(age)`, so the age is stored as a real **number**). Otherwise, the original row (stripped) goes to `invalid_records` with its reasons.
> - `return valid_records, invalid_records` and `valid, invalid = clean_dataset(raw_data)`: returning two values, as in Task 6.
> - When printing: `{person['age']:>3}` means right-aligned in 3 spaces (this looks nicer for numbers). `'; '.join(item['reasons'])` joins all the reasons on one line with `; `.
>
> **Actual output:**
> ```

>
> **Traps**
> - If you don't strip the fields before `all()`, `" "` (one space) is **not** counted as empty, because it is a one-character string and counts as `True`. Strip first, `all` after.
> - `email.find(".") > email.find("@")` also looks right, but it isn't: `find(".")` only gives the **first** dot. In `"john.smith@email.com"` the first dot is at index 4 and the `@` is at index 10, so `4 > 10` is false and a **good** email gets rejected! The "search from the `@` onwards" method (`find(".", at_position)`) doesn't have this problem.
> - This email check is still not complete: it accepts `"a@.com"`. Exact email validation is usually done with regex, which you will see later.

````python
STOP_WORDS = ['the', 'a', 'is', 'and', 'of', 'to', 'in']
PUNCTUATION = '.,!?;:"()-*'
WIDTH = 46


def text_report(text):
    """Print a formatted analysis report of text."""
    # ---- 1. Basic stats
    total_characters = len(text)
    raw_words = text.split()
    total_sentences = text.count(".") + text.count("!") + text.count("?")

    clean_words = []
    for raw_word in raw_words:
        word = raw_word.strip(PUNCTUATION).lower()
        if word != "":
            clean_words.append(word)

    total_letters = 0
    for word in clean_words:
        total_letters += len(word)
    average_length = total_letters / len(clean_words)

    # ---- 2. Character breakdown
    uppercase = 0
    lowercase = 0
    digits = 0
    spaces = 0
    punctuation = 0
    for char in text:
        if char.isupper():
            uppercase += 1
        elif char.islower():
            lowercase += 1
        elif char.isdigit():
            digits += 1
        elif char.isspace():
            spaces += 1
        else:
            punctuation += 1

    # ---- 3. Top 5 words (without the stop words)
    word_counts = {}
    for word in clean_words:
        if word in STOP_WORDS:
            continue
        word_counts[word] = word_counts.get(word, 0) + 1
    top_words = sorted(word_counts, key=word_counts.get, reverse=True)[:5]

    # ---- 4. Longest and shortest words
    longest = clean_words[0]
    shortest = clean_words[0]
    for word in clean_words:
        if len(word) > len(longest):
            longest = word
        if len(word) < len(shortest):
            shortest = word

    # ---- 5. The report
    print("=" * WIDTH)
    print("TEXT REPORT".center(WIDTH))
    print("=" * WIDTH)

    print("Basic stats".center(WIDTH, "-"))
    print("Characters:".ljust(28) + str(total_characters))
    print("Words:".ljust(28) + str(len(raw_words)))
    print("Sentences:".ljust(28) + str(total_sentences))
    print("Average word length:".ljust(28) + f"{average_length:.2f}")

    print("Character breakdown".center(WIDTH, "-"))
    print("Uppercase letters:".ljust(28) + str(uppercase))
    print("Lowercase letters:".ljust(28) + str(lowercase))
    print("Digits:".ljust(28) + str(digits))
    print("Spaces (incl. newlines):".ljust(28) + str(spaces))
    print("Punctuation/other:".ljust(28) + str(punctuation))

    print("Top 5 words".center(WIDTH, "-"))
    print("In one line: " + ", ".join(top_words))
    for rank in range(len(top_words)):
        word = top_words[rank]
        print(f"{rank + 1}. " + word.ljust(25) + str(word_counts[word]))

    print("Longest and shortest".center(WIDTH, "-"))
    print("Longest word:".ljust(28) + f"{longest} (length {len(longest)})")
    print("Shortest word:".ljust(28) + f"{shortest} (length {len(shortest)})")
    print("=" * WIDTH)


corpus = """Artificial Intelligence is transforming every industry. From healthcare to finance,
AI systems are making decisions that affect millions of people. Machine Learning,
a subset of AI, enables computers to learn from data without being explicitly programmed.
Deep Learning takes this further with neural networks inspired by the human brain.
In 2024, Large Language Models like GPT have shown remarkable abilities in understanding
and generating human language. The future of AI is both exciting and challenging."""

text_report(corpus)
````

## Task 14: Build a Simple Search Engine
Create a function `search(documents, query)` that:
1. Takes a list of document strings and a search query
2. Returns documents **ranked** by how many times the query words appear
3. The search should be **case-insensitive**
4. If query has multiple words, count matches for **each** word
5. Show a relevance score and highlight which query words were found

**Methods to use:** `lower()`, `split()`, `count()`, `any()`, `all()`, `strip()`, `replace()`

```python
documents = [
    "Python is a great programming language for data science and machine learning",
    "Data analysis with Python involves cleaning and processing large datasets",
    "Machine learning models can predict outcomes from historical data",
    "Web development with JavaScript and HTML is different from data work",
    "Python data science libraries include NumPy and Pandas for data processing"
]

# Test searches
search(documents, "data science")
search(documents, "Python machine learning")
```

````python
PUNCTUATION = '.,!?;:"()'


def get_score(result):
    """Return the score of one search result (used for sorting)."""
    return result["score"]


def search(documents, query):
    """Rank documents by how many times the query words appear; print and return the results."""
    query_words = query.lower().strip().split()
    results = []

    for document in documents:
        clean_document = document.lower()
        for mark in PUNCTUATION:
            clean_document = clean_document.replace(mark, " ")
        document_words = clean_document.split()

        # Skip documents that contain none of the query words
        if not any(word in document_words for word in query_words):
            continue

        score = 0
        found = []
        for word in query_words:
            hits = document_words.count(word)
            if hits > 0:
                score += hits
                found.append(f"{word} x{hits}")

        # Highlight: put [ ] around every word of the document that matches the query
        highlighted = []
        for original_word in document.split():
            if original_word.lower().strip(PUNCTUATION) in query_words:
                highlighted.append(f"[{original_word}]")
            else:
                highlighted.append(original_word)

        results.append({
            "document": document,
            "score": score,
            "found": found,
            "all_found": all(word in document_words for word in query_words),
            "highlighted": " ".join(highlighted),
        })

    results.sort(key=get_score, reverse=True)

    print(f'Search: "{query}"  ->  {len(results)} matching document(s)')
    rank = 1
    for result in results:
        if result["all_found"]:
            tag = "  (all words found)"
        else:
            tag = ""
        print(f"{rank}. score {result['score']} | found: {', '.join(result['found'])}{tag}")
        print(f"   {result['highlighted']}")
        rank += 1
    print()
    return results


documents = [
    "Python is a great programming language for data science and machine learning",
    "Data analysis with Python involves cleaning and processing large datasets",
    "Machine learning models can predict outcomes from historical data",
    "Web development with JavaScript and HTML is different from data work",
    "Python data science libraries include NumPy and Pandas for data processing"
]

# Test searches
search(documents, "data science")
search(documents, "Python machine learning")
search(documents, "cooking")
````

> 💡 **Explanation**
>
> **Main idea:** for each document: (1) make it lowercase and split it into words, (2) count how many times each query word is in it, (3) the sum of these = the **score**. At the end, sort the documents by score, from highest to lowest. Each result is a **dictionary** that keeps all its information together.
>
> **The helper function `get_score`**
> - It takes a result dictionary and returns only its score. Later we tell `sort` "compare with this function" (`key=get_score`). It's the same idea as `key=frequency.get` in Task 9, but this time we wrote the function **ourselves**. (Again: `get_score` **without** brackets.)
>
> **The `search` function, line by line**
> - `query.lower().strip().split()`: makes the query lowercase (step 3), removes the spaces on both ends, and splits it into words (step 4): `"Python machine learning"` → `['python', 'machine', 'learning']`.
> - **Preparing the document:**
>   - `document.lower()`: the document is made lowercase too, so it matches the query.
>   - `for mark in PUNCTUATION: clean_document = clean_document.replace(mark, " ")`: we replace the marks with a **space** (not with nothing), so if somewhere it said `data,science`, it becomes two separate words, not `datascience`.
>   - `document_words = clean_document.split()`: the list of words in the document.
> - `if not any(word in document_words for word in query_words): continue`: if **none** of the query words is in the document, don't bother with it at all. `any` means "is at least one there?".
> - **Score:** for each query word, `document_words.count(word)` counts it in the **list** of words. Their sum is `score`. The words that were found go into `found` with their counts (for example `"data x2"`).
> - **Highlight (step 5):** we go through the **original** document (with its own capital letters) word by word. If the lowercase, mark-free version of a word is in the query, we put `[ ]` around it. At the end, `" ".join(highlighted)` builds a sentence again. This way the user sees the original text, not the lowercase version.
> - `"all_found": all(word in document_words for word in query_words)`: `all` means "are **all** the query words in this document?". Documents that have all the words get an `(all words found)` tag.
> - `results.sort(key=get_score, reverse=True)`: step 2. `.sort()` sorts the list **in place**. `reverse=True` means the highest score comes first. Because sorting is stable, documents with the same score stay in the same order as in the original list.
> - The print loop: `', '.join(result['found'])` puts the found words on one line, separated by commas.
> - `return results`: returns the sorted list so it can be used (step 2 says "returns").

> **Traps**
> - **`.count()` on the string instead of the list:** `document.lower().count("data")` would give 2 in the second document (`data` and `datasets`). For a search engine we usually want whole words, so we count on `document_words` (the list).
> - If you don't `.lower()` the `query` but you do lowercase the document, `"Python"` will never be found.
> - `results.sort(...)` returns nothing (`None`). So `results = results.sort(...)` destroys your list. It's the same trap as `random.shuffle` in Week 4.
> - This engine doesn't recognise different forms of a word: it doesn't treat `learn` and `learning` as the same. Real search engines use **stemming** for this (reducing a word to its root).
