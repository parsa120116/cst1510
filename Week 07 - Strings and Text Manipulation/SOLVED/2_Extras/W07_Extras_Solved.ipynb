## Hangman

**Uses:** string indexing/looping, `in`, lists, `random.choice()`

1. Pick a random word from a list of words: `chosen_word = random.choice(word_list)`.
2. Build a `placeholder` string of underscores, one per letter of `chosen_word` (e.g. `"apple"` → `"_ _ _ _ _"`).
3. Loop: ask the user to guess a letter. If it's `in chosen_word`, reveal it in the display at the correct position(s) — a letter can appear more than once. If not, lose a life.
4. Track `lives` (start at 6). Stop the game when either the display has no more blanks (win) or `lives` reaches 0 (lose).
5. Don't deduct a life for guessing a letter already guessed — just remind them.

````python
import random

word_list = ["apple", "banana", "camel", "python", "string", "letter", "puzzle", "london"]

chosen_word = random.choice(word_list)
lives = 6
guessed_letters = []

# Step 2: one underscore per letter, e.g. "apple" -> "_ _ _ _ _"
placeholder = ""
for letter in chosen_word:
    placeholder += "_ "
print("Welcome to Hangman!")
print(placeholder.strip())

# The display is a list, so we can change one position at a time
display = []
for letter in chosen_word:
    display.append("_")

while "_" in display and lives > 0:
    guess = input("Guess a letter: ").strip().lower()

    if len(guess) != 1 or not guess.isalpha():
        print("Please type exactly one letter.")
        continue

    if guess in guessed_letters:
        print(f"You've already guessed '{guess}'. Try another letter.")
        continue
    guessed_letters.append(guess)

    if guess in chosen_word:
        for position in range(len(chosen_word)):
            if chosen_word[position] == guess:
                display[position] = guess
        print(f"Good guess! '{guess}' is in the word.")
    else:
        lives -= 1
        print(f"'{guess}' is not in the word. You lose a life.")

    print(" ".join(display))
    print(f"Lives left: {lives}")

if "_" not in display:
    print(f"You win! The word was '{chosen_word}'.")
else:
    print(f"You lose! The word was '{chosen_word}'.")
````

> 💡 **Explanation**
>
> **Main idea:** we have two things: the **secret word** (`chosen_word`), which never changes, and the **display** (`display`), which starts as all `_` and where each correct guess replaces some of the `_` with the real letter. The game means: "while there are still `_` left and you still have lives, guess".
>
> **Line by line**
>
> *Setting up*
> - `import random`: the random module (Week 3). We write it at the top of the file.
> - `word_list = [...]`: the list of words. They are all **lowercase**. This matters, because later we also make the user's guess lowercase so the two match.
> - `chosen_word = random.choice(word_list)`: takes a random item from the list (step 1).
> - `lives = 6`: six lives (step 4).
> - `guessed_letters = []`: an empty list for the letters that have been guessed so far (for step 5).
>
> *Step 2: building `placeholder`*
> - `placeholder = ""`: we start from an **empty** string.
> - `for letter in chosen_word:` a `for` loop over a **string** moves forward one character at a time. For `"apple"` it goes round five times.
> - `placeholder += "_ "`: each time round it adds a `_` and a space to the **end** of the string. This is called "building a string piece by piece". After 5 rounds: `"_ _ _ _ _ "` (with one extra space at the end).
> - `print(placeholder.strip())`: `.strip()` removes that extra space at the end (Topic 1, Week 7).
>
> *Why is `display` a **list**, not a string?*
> - Strings in Python are **immutable**. This means you can't write `placeholder[2] = "p"`. You get the error `TypeError: 'str' object does not support item assignment`.
> - But a list is mutable: `display[2] = "p"` is completely fine. So for "change one position" we use a list.
> - `display.append("_")` in the loop: for each letter of the word, it adds one `"_"` to the list. For `"apple"` it becomes `['_', '_', '_', '_', '_']`.
>
> *The main game loop*
> - `while "_" in display and lives > 0:` two conditions joined with `and`: "there is still an empty position" **and** "you still have lives". If either one becomes false, the loop ends. Here `in` works on a **list**: "is `"_"` one of the items in the list?"
> - `guess = input("Guess a letter: ").strip().lower()`: three jobs one after another (this is called **method chaining**): it reads the input, removes extra spaces from both ends, and makes it lowercase. So if the user types `" A "`, `guess` becomes `"a"`.
> - `if len(guess) != 1 or not guess.isalpha():` rejects bad input: if its length is not 1 (like `"ab"` or an empty string) **or** it's not a letter (like `"1"` or `"?"`). `.isalpha()` means "are all the characters letters of the alphabet?" and returns `True`/`False`.
> - `continue`: skip the rest of this round and go to the next round of the loop. That means ask again. No life is lost.
> - `if guess in guessed_letters:` step 5. If they already said it, remind them and go back with `continue`. Because this check comes **before** a life is taken away, a repeated guess never costs a life, whether it was right or wrong before.
> - `guessed_letters.append(guess)`: now that the guess is new, record it.
> - `if guess in chosen_word:` here `in` works on a **string**: "is this letter anywhere in the word?"
> - `for position in range(len(chosen_word)):` we loop over the **indexes** (0, 1, 2, …), not over the letters. Why? Because we want to know **where** the letter is, so we can fill that same position in `display`.
> - `if chosen_word[position] == guess: display[position] = guess`: wherever the letter of the word is the same as the guess, fill that position. Because the loop goes **to the end** of the word, if the letter appears twice (two `p`s in `apple`), **both** are filled.
> - `else: lives -= 1`: a wrong guess, one life less.
> - `print(" ".join(display))`: `.join()` joins the items of the list with a space between them and makes a string (Topic 3, Week 7): `['a', 'p', 'p', '_', '_']` → `"a p p _ _"`.
>
> *After the loop*
> - The loop can end for two reasons. With `if "_" not in display` we find out which one: if no `_` is left, they won. Otherwise the lives must have run out.
>
> **A sample game** (word = `apple`):
> ```
> Welcome to Hangman!
> _ _ _ _ _
> Guess a letter: a
> Good guess! 'a' is in the word.
> a _ _ _ _
> Lives left: 6
> Guess a letter: x
> 'x' is not in the word. You lose a life.
> a _ _ _ _
> Lives left: 5
> Guess a letter: A
> You've already guessed 'a'. Try another letter.
> Guess a letter: p
> Good guess! 'p' is in the word.
> a p p _ _
> Lives left: 5
> ```
> Look: the capital `A` was also recognised as a "repeat" (because of `.lower()`), and one guess of `p` filled **two** positions.
>
> **Traps**
> - If you forget `.lower()`, and the user types `A` when the word is `apple`, then `"A" in "apple"` is `False` and they lose a life for no reason.
> - If you use `chosen_word.find(guess)` or `.index(guess)` instead of the loop over indexes, it only finds the **first** place of the letter. So in `apple` only one of the `p`s is revealed. That's why we used a loop.
> - If you put the "repeat" check **after** `lives -= 1`, a repeated wrong guess takes a life again. The order matters.
> - Don't write the `while` condition with `or`. `"_" in display or lives > 0` means keep going even with 0 lives!
>

## Caesar Cipher

**Uses:** string indexing, `.index()`, `%` (modulo), functions

1. Write `encrypt(original_text, shift_amount)`: for each letter in `original_text`, find its position in the alphabet, add `shift_amount`, and use `% 26` to wrap back around from `z` to `a` instead of running off the end of the alphabet.
2. Write `decrypt(original_text, shift_amount)` the same way, but shifting backwards (multiply the shift by `-1`).
3. Combine both into one `caesar(direction, text, shift)` function that calls the right one based on `direction`.
4. Handle characters that aren't letters (spaces, punctuation, digits) — they should pass through unchanged, not crash the program or vanish.
5. Ask the user if they want to go again in a loop, instead of the program just ending after one message.

````python
alphabet = "abcdefghijklmnopqrstuvwxyz"


def encrypt(original_text, shift_amount):
    """Return original_text with every letter shifted forward by shift_amount."""
    cipher_text = ""
    for char in original_text:
        lower_char = char.lower()
        if lower_char in alphabet:
            position = alphabet.index(lower_char)
            new_position = (position + shift_amount) % 26
            new_char = alphabet[new_position]
            if char.isupper():
                new_char = new_char.upper()
            cipher_text += new_char
        else:
            cipher_text += char          # space, digit, punctuation: keep it
    return cipher_text


def decrypt(original_text, shift_amount):
    """Return original_text with every letter shifted backward by shift_amount."""
    return encrypt(original_text, shift_amount * -1)


def caesar(direction, text, shift):
    """Call encrypt or decrypt depending on direction ('encode' or 'decode')."""
    if direction == "encode":
        return encrypt(text, shift)
    else:
        return decrypt(text, shift)


print("Welcome to the Caesar Cipher!")

should_continue = True
while should_continue:
    direction = input("Type 'encode' to encrypt, type 'decode' to decrypt:\n").strip().lower()
    if direction != "encode" and direction != "decode":
        print("Please type 'encode' or 'decode'.")
        continue

    text = input("Type your message:\n")

    try:
        shift = int(input("Type the shift number:\n"))
    except ValueError:
        print("The shift must be a whole number. Let's start again.")
        continue

    result = caesar(direction, text, shift)
    print(f"Here is the {direction}d result: {result}")

    again = input("Type 'yes' to go again. Otherwise, type 'no':\n").strip().lower()
    if again != "yes":
        should_continue = False
        print("Goodbye!")
````

> 💡 **Explanation**
>
> **Main idea:** turn each letter into a **number** (its place in the alphabet: `a`=0, `b`=1, … `z`=25), move the number, and turn it back into a letter. `% 26` stops the number from going above 25 or below 0.
>
> **Line by line**
>
> *The alphabet*
> - `alphabet = "abcdefghijklmnopqrstuvwxyz"`: a 26-letter string. Because a string has indexes, `alphabet[0]` is `"a"` and `alphabet[25]` is `"z"`.
>
> *The `encrypt` function (steps 1 and 4)*
> - `cipher_text = ""`: we start the answer from an empty string and build it letter by letter.
> - `for char in original_text:` over each character of the text.
> - `lower_char = char.lower()`: the lowercase version of the letter. Why? Because `alphabet` only has lowercase letters. `"H"` is not in it, but `"h"` is.
> - `if lower_char in alphabet:` is this character an English letter? For `" "`, `"!"` or `"7"` the answer is `False`, and it goes to `else`.
> - `position = alphabet.index(lower_char)`: `.index()` tells you **where** this letter is in the string. `alphabet.index("h")` → `7`.
> - `new_position = (position + shift_amount) % 26`: the heart of the program. For example, `y` (24) with a shift of 5: `24 + 5 = 29`, and `29 % 26 = 3` → `d`. Without `% 26`, `alphabet[29]` gives the error `IndexError: string index out of range`.
> - `new_char = alphabet[new_position]`: we turn the new number back into a letter.
> - `if char.isupper(): new_char = new_char.upper()`: if the **original** letter was a capital, make the answer a capital too. So `Hello` becomes `Mjqqt`, not `mjqqt`. (This is a small improvement. Angela's original version makes all the text lowercase.)
> - `else: cipher_text += char`: step 4. Anything that isn't a letter is added to the answer **unchanged**. If you leave out this `else`, the spaces and punctuation **disappear** from the text.
> - `return cipher_text`: returns the string we built (not `print`, so that it can be used).
>
> *The `decrypt` function (step 2)*
> - `return encrypt(original_text, shift_amount * -1)`: decrypting means the same shift but **negative**. Instead of writing the whole loop again, we call the same `encrypt` with a negative shift. We didn't write repeated code.
> - **Why does a negative number work?** `d` (3) with a shift of `-5`: `3 - 5 = -2`. In Python, `-2 % 26` is `24`, that is, `y`. So in Python, `%` keeps the answer between 0 and 25 even for a negative number, and goes from the start of the alphabet back round to the end.
>
> *The `caesar` function (step 3)*
> - A simple `if`: `"encode"` → `encrypt`, otherwise → `decrypt`. It `return`s whatever answer the function gave back.
>
> *The "again?" loop (step 5)*
> - `should_continue = True` and `while should_continue:` a **flag**: while it is `True`, the loop keeps going.
> - `input("...\n").strip().lower()`: the `\n` at the end of the question makes the user type on the next line. `.strip().lower()` means `" Encode "` is accepted too.
> - `if direction != "encode" and direction != "decode":` if it's neither of them, show a message and ask again with `continue`.
> - `try: shift = int(input(...))` / `except ValueError:` if the user writes `five` instead of a number, `int()` gives a `ValueError`. Here we catch it so the program doesn't crash (Week 4).
> - `print(f"Here is the {direction}d result: {result}")`: a small f-string trick: `"encode" + "d"` → `encoded`.
> - `if again != "yes": should_continue = False`: any answer other than `yes` sets the flag to `False`, and next time round the `while` condition is false and the loop ends.
>
> **A sample run:**
> ```
> Welcome to the Caesar Cipher!
> Type 'encode' to encrypt, type 'decode' to decrypt:
> encode
> Type your message:
> Hello, World! 2026
> Type the shift number:
> 5
> Here is the encoded result: Mjqqt, Btwqi! 2026
> Type 'yes' to go again. Otherwise, type 'no':
> yes
> Type 'encode' to encrypt, type 'decode' to decrypt:
> decode
> Type your message:
> Mjqqt, Btwqi! 2026
> Type the shift number:
> 5
> Here is the decoded result: Hello, World! 2026
> Type 'yes' to go again. Otherwise, type 'no':
> no
> Goodbye!
> ```
> Look: the comma, the `!`, the space and `2026` stayed unchanged, and the capitals `H` and `W` were kept too.
>
> **Hand check:** `h` = 7 → 7+5 = 12 → `m` ✓ · `o` = 14 → 19 → `t` ✓ · `w` = 22 → 27 % 26 = 1 → `b` ✓
>
> **Traps**
> - **A big shift:** with a shift of `29`, without `% 26` even `a` runs off the end of the alphabet. With `% 26`, a shift of 29 works exactly like a shift of 3 (`xyz` → `abc`).
> - If you don't use `.lower()` when searching in `alphabet`, capital letters are treated like punctuation and are **not encrypted**. This is a silent bug: the program doesn't crash, the answer is just wrong.
> - `alphabet.index()` gives an **error** if it can't find the letter (`ValueError: substring not found`). That's why we checked with `in` **first**. (`.find()` returns `-1` in the same situation and gives no error.)
>
> **Try it yourself:** encrypt a message with a shift of 3, give it to a friend with a copy of this program, and see if they can open it without knowing the shift, by trying 1 to 25. (This is called brute force, and it shows why the Caesar cipher is not secure.)
