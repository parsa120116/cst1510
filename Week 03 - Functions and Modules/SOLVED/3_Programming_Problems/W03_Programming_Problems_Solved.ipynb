## Task 1: Quiz
Suppose you want to develop a program to help a first grader practice addition, subtraction, multiplication, division, exponent or remainder. The user should have the option of choosing which mathematical operation they wish to practice. The program randomly generates two single-digit integers, number1 and number2, and of course the user will provide the answer to the question.

The program should include:
- A menu allowing the student to choose the operation they want to practice (ideally included in function names or main).
- Functions for each of the mathematical operations.
- A score calculating correct answers.
- Error messages for situations such as division by zero.
- Functionality to keep practicing until the user decides otherwise.

````python
import random


def add(number1, number2):
    """Return number1 + number2."""
    return number1 + number2


def subtract(number1, number2):
    """Return number1 - number2."""
    return number1 - number2


def multiply(number1, number2):
    """Return number1 * number2."""
    return number1 * number2


def divide(number1, number2):
    """Return number1 / number2 rounded to 2 d.p., or None when number2 is 0."""
    if number2 == 0:
        return None
    return round(number1 / number2, 2)


def exponent(number1, number2):
    """Return number1 to the power of number2."""
    return number1 ** number2


def remainder(number1, number2):
    """Return the remainder of number1 / number2, or None when number2 is 0."""
    if number2 == 0:
        return None
    return number1 % number2


def show_menu():
    """Print the operations the student can practise."""
    print()
    print("===== Maths Practice =====")
    print("1) Addition        (+)")
    print("2) Subtraction     (-)")
    print("3) Multiplication  (*)")
    print("4) Division        (/)")
    print("5) Exponent        (**)")
    print("6) Remainder       (%)")
    print("Q) Quit")


def is_number(text):
    """Return True if text is a number such as 7, -3 or 2.5."""
    if text.startswith("-"):
        text = text[1:]
    return text.replace(".", "", 1).isdigit()


def read_answer(question):
    """Ask the question until the user types a number; return it as a float."""
    while True:
        text = input(question).strip()
        if is_number(text):
            return float(text)
        print("Please type a number, for example 7 or 2.5")


def ask_question(choice):
    """Ask one random question; return "correct", "wrong" or "skipped"."""
    number1 = random.randint(0, 9)
    number2 = random.randint(0, 9)

    if choice == "1":
        symbol = "+"
        correct = add(number1, number2)
    elif choice == "2":
        symbol = "-"
        correct = subtract(number1, number2)
    elif choice == "3":
        symbol = "*"
        correct = multiply(number1, number2)
    elif choice == "4":
        symbol = "/"
        correct = divide(number1, number2)
    elif choice == "5":
        symbol = "**"
        correct = exponent(number1, number2)
    else:
        symbol = "%"
        correct = remainder(number1, number2)

    if correct is None:
        print(f"Error: {number1} {symbol} {number2} means dividing by zero, which is not allowed.")
        print("This question is skipped. Let's try another one!")
        return "skipped"

    if symbol == "/":
        print("(Give division answers to 2 decimal places.)")
    answer = read_answer(f"What is {number1} {symbol} {number2}? ")

    if round(answer, 2) == correct:
        print("Correct! Well done.")
        return "correct"
    print(f"Not quite. The answer is {correct}.")
    return "wrong"


def main():
    """Run the quiz until the student chooses to quit."""
    score = 0
    questions = 0
    while True:
        show_menu()
        choice = input("Choose an operation (1-6) or Q to quit: ").strip().upper()
        if choice == "Q":
            break
        if choice not in ("1", "2", "3", "4", "5", "6"):
            print("Sorry, that is not on the menu. Please choose 1-6 or Q.")
            continue

        result = ask_question(choice)
        if result == "skipped":
            continue
        questions += 1
        if result == "correct":
            score += 1
        print(f"Score: {score} / {questions}")

    print(f"Final score: {score} out of {questions}. Goodbye!")


if __name__ == "__main__":
    main()
````

> 💡 **Explanation**
>
> **Main idea: one function for each job.** We split the program into pieces: six maths functions, a menu function, a function to read the answer, a function to ask one question, and `main`, which connects them all.
>
> **Line by line**
> - `import random`: for making random numbers. `random.randint(0, 9)` gives a number from 0 to 9, and **both ends are included**. So it is exactly "a single-digit number".
> - `add`, `subtract`, `multiply`, `exponent`: each one takes two parameters and **`return`s** the answer; it doesn't print it. Why? Because `ask_question` needs to **keep** the answer and compare it with the child's answer.
> - `divide` and `remainder`: before calculating, `if number2 == 0: return None`. **`None`** means "nothing / there is no answer". This way, division by zero never runs and the program doesn't crash (we don't get a `ZeroDivisionError`).
> - `round(number1 / number2, 2)`: for example `1 / 3` is `0.3333…`. No child will type all of that, so we round the correct answer to 2 decimal places (`0.33`).
> - `is_number(text)`: checks whether the text is a number. First, if it is negative, it removes the `-`. Then it removes **only one dot** (`replace(".", "", 1)`) and asks "is everything left a digit?" (`isdigit()`). So `7`, `-3` and `2.5` are accepted, but `abc` or `1.2.3` are not.
> - `read_answer`: a `while True` that doesn't give up until it gets a proper number. When it gets one, `return` both gives back the answer **and** leaves the loop (and the function).
> - `ask_question(choice)`: makes two random numbers and uses `if/elif` to call the right function. If the answer is `None` → error message and `return "skipped"`. Otherwise it asks the question and returns `"correct"` or `"wrong"`.
> - `if correct is None`: for `None`, always use `is`, not `==`.
> - `round(answer, 2) == correct`: the child's answer is a float (`12.0`) and the correct answer may be an int (`12`). In Python `12.0 == 12` is `True`, so there is no problem.
> - `main`: two counters (`score` and `questions`), the menu, and quitting with `Q`. `choice not in ("1", …, "6")` rejects wrong choices, and `continue` goes back to the top of the loop. Skipped questions are not counted.
> - `if __name__ == "__main__": main()`: means "only call `main` when this file is run directly". If one day you `import` this file (for example to test the functions), the quiz won't start by itself (Week 3, D15).
>
> **Traps**
> - `"12" in "123456"` is also `True`! That's why we put the choices in a tuple: `("1", "2", …)`. This way only exactly one of these is accepted.
> - If you use `print` instead of `return` in the maths functions, `correct` becomes `None` and every question is skipped (Week 3, D4: `print` is not the same as `return`).
> - `score` and `questions` are **local** to `main`. If you try to increase them from another function, it won't work. That's why `ask_question` only returns the result and `main` itself increases the score.
>
> **Hand check:** `7 % 3` → `1`. `2 ** 3` → `8`. `5 / 0` → error message and the next question.
>
> **Try it yourself:** add a "hard level" option where the numbers have two digits (`random.randint(10, 99)`). Hint: give `ask_question` a parameter with a default value, for example `highest=9`.

## Task 2: Scientific Calculator (Unit Conversion Chart)
Create a Scientific Calculator imitating the Unit Conversion Chart (Fig. 1).

Allow the user to choose between length, temperature, mass, volume, pressure, and energy. Create separate functions for each calculation within the types of unit. Allow the user to choose between units. For example, if the user chooses length, then the programme should allow the user to choose between conversions: m/ft, in/cm, or mi/km. The user should then enter the value to calculate. Finally, the system should do the conversion and print out the result.

![Unit Conversion Chart](Unit-Conversion-Chart.png)

*Fig 1. Unit Conversion Chart*

````python
# All factors below are copied from the Unit Conversion Chart (Fig. 1).
FT_PER_M = 3.3
CM_PER_IN = 2.43        # chart value (the exact value is 2.54)
KM_PER_MI = 1.61
LB_PER_KG = 2.2
G_PER_OZ = 28.35
CHART_M3_PER_FT3 = 0.028   # chart prints "1 ft = 0.028 m" in the Mass box
CC_PER_ML = 1
ML_PER_CUP = 237
CUPS_PER_GAL = 16
FT3_PER_M3 = 35.31
KPA_PER_ATM = 101.3
MMHG_PER_ATM = 760
PA_PER_BAR = 10 ** 5
MBAR_PER_PSI = 68.948
KCAL_PER_J = 2.39e-4
J_PER_EV = 1.602e-19    # chart prints "K"; the unit should be J


# ---------- Length ----------
def m_to_ft(value):
    """Convert metres to feet."""
    return value * FT_PER_M

def ft_to_m(value):
    """Convert feet to metres."""
    return value / FT_PER_M

def in_to_cm(value):
    """Convert inches to centimetres."""
    return value * CM_PER_IN

def cm_to_in(value):
    """Convert centimetres to inches."""
    return value / CM_PER_IN

def mi_to_km(value):
    """Convert miles to kilometres."""
    return value * KM_PER_MI

def km_to_mi(value):
    """Convert kilometres to miles."""
    return value / KM_PER_MI


# ---------- Temperature ----------
def c_to_f(value):
    """Convert Celsius to Fahrenheit: F = 9/5 C + 32."""
    return 9 / 5 * value + 32

def f_to_c(value):
    """Convert Fahrenheit to Celsius: C = 5/9 (F - 32)."""
    return 5 / 9 * (value - 32)

def c_to_k(value):
    """Convert Celsius to Kelvin: K = C + 273.15."""
    return value + 273.15

def k_to_c(value):
    """Convert Kelvin to Celsius: C = K - 273.15."""
    return value - 273.15

def f_to_k(value):
    """Convert Fahrenheit to Kelvin: K = 5/9 (F - 32) + 273."""
    return 5 / 9 * (value - 32) + 273

def k_to_f(value):
    """Convert Kelvin to Fahrenheit (the chart formula turned around)."""
    return 9 / 5 * (value - 273) + 32


# ---------- Mass ----------
def kg_to_lb(value):
    """Convert kilograms to pounds."""
    return value * LB_PER_KG

def lb_to_kg(value):
    """Convert pounds to kilograms."""
    return value / LB_PER_KG

def oz_to_g(value):
    """Convert ounces to grams."""
    return value * G_PER_OZ

def g_to_oz(value):
    """Convert grams to ounces."""
    return value / G_PER_OZ

def cubic_ft_to_cubic_m(value):
    """Convert cubic feet to cubic metres (Mass box of the chart)."""
    return value * CHART_M3_PER_FT3

def cubic_m_to_cubic_ft(value):
    """Convert cubic metres to cubic feet (Mass box of the chart)."""
    return value / CHART_M3_PER_FT3


# ---------- Volume ----------
def ml_to_cc(value):
    """Convert millilitres to cubic centimetres."""
    return value * CC_PER_ML

def cc_to_ml(value):
    """Convert cubic centimetres to millilitres."""
    return value / CC_PER_ML

def cup_to_ml(value):
    """Convert cups to millilitres."""
    return value * ML_PER_CUP

def ml_to_cup(value):
    """Convert millilitres to cups."""
    return value / ML_PER_CUP

def gal_to_cup(value):
    """Convert gallons to cups."""
    return value * CUPS_PER_GAL

def cup_to_gal(value):
    """Convert cups to gallons."""
    return value / CUPS_PER_GAL

def m3_to_ft3(value):
    """Convert cubic metres to cubic feet."""
    return value * FT3_PER_M3

def ft3_to_m3(value):
    """Convert cubic feet to cubic metres."""
    return value / FT3_PER_M3


# ---------- Pressure ----------
def atm_to_kpa(value):
    """Convert atmospheres to kilopascals."""
    return value * KPA_PER_ATM

def kpa_to_atm(value):
    """Convert kilopascals to atmospheres."""
    return value / KPA_PER_ATM

def atm_to_mmhg(value):
    """Convert atmospheres to millimetres of mercury."""
    return value * MMHG_PER_ATM

def mmhg_to_atm(value):
    """Convert millimetres of mercury to atmospheres."""
    return value / MMHG_PER_ATM

def bar_to_pa(value):
    """Convert bar to pascals."""
    return value * PA_PER_BAR

def pa_to_bar(value):
    """Convert pascals to bar."""
    return value / PA_PER_BAR

def psi_to_mbar(value):
    """Convert psi to millibar."""
    return value * MBAR_PER_PSI

def mbar_to_psi(value):
    """Convert millibar to psi."""
    return value / MBAR_PER_PSI


# ---------- Energy ----------
def j_to_kcal(value):
    """Convert joules to kilocalories."""
    return value * KCAL_PER_J

def kcal_to_j(value):
    """Convert kilocalories to joules."""
    return value / KCAL_PER_J

def ev_to_j(value):
    """Convert electronvolts to joules."""
    return value * J_PER_EV

def j_to_ev(value):
    """Convert joules to electronvolts."""
    return value / J_PER_EV


# ---------- Helpers for the menus ----------
def is_number(text):
    """Return True if text is a number such as 7, -3 or 2.5."""
    if text.startswith("-"):
        text = text[1:]
    return text.replace(".", "", 1).isdigit()


def read_value():
    """Ask until the user types a number; return it as a float."""
    while True:
        text = input("Enter the value to convert: ").strip()
        if is_number(text):
            return float(text)
        print("That is not a number, please try again.")


def ask_choice(highest, lowest=1):
    """Ask until the user picks a whole number from lowest to highest; return it."""
    while True:
        text = input(f"Choose {lowest}-{highest}: ").strip()
        if text.isdigit() and lowest <= int(text) <= highest:
            return int(text)
        print("Not on the menu, please try again.")


def show_result(value, from_unit, result, to_unit):
    """Print one conversion, e.g. 10 m = 33 ft."""
    print(f"{value:g} {from_unit} = {result:.6g} {to_unit}")


# ---------- One menu per unit type ----------
def length_menu():
    """Let the user pick a length conversion and print the result."""
    print("1) m -> ft     2) ft -> m")
    print("3) in -> cm    4) cm -> in")
    print("5) mi -> km    6) km -> mi")
    choice = ask_choice(6)
    value = read_value()
    if choice == 1:
        show_result(value, "m", m_to_ft(value), "ft")
    elif choice == 2:
        show_result(value, "ft", ft_to_m(value), "m")
    elif choice == 3:
        show_result(value, "in", in_to_cm(value), "cm")
    elif choice == 4:
        show_result(value, "cm", cm_to_in(value), "in")
    elif choice == 5:
        show_result(value, "mi", mi_to_km(value), "km")
    else:
        show_result(value, "km", km_to_mi(value), "mi")


def temperature_menu():
    """Let the user pick a temperature conversion and print the result."""
    print("1) °C -> °F    2) °F -> °C")
    print("3) °C -> K     4) K -> °C")
    print("5) °F -> K     6) K -> °F")
    choice = ask_choice(6)
    value = read_value()
    if choice == 1:
        show_result(value, "°C", c_to_f(value), "°F")
    elif choice == 2:
        show_result(value, "°F", f_to_c(value), "°C")
    elif choice == 3:
        show_result(value, "°C", c_to_k(value), "K")
    elif choice == 4:
        show_result(value, "K", k_to_c(value), "°C")
    elif choice == 5:
        show_result(value, "°F", f_to_k(value), "K")
    else:
        show_result(value, "K", k_to_f(value), "°F")


def mass_menu():
    """Let the user pick a mass conversion and print the result."""
    print("1) kg -> lb    2) lb -> kg")
    print("3) oz -> g     4) g -> oz")
    print("5) ft³ -> m³   6) m³ -> ft³   (the chart's '1 ft = 0.028 m')")
    choice = ask_choice(6)
    value = read_value()
    if choice == 1:
        show_result(value, "kg", kg_to_lb(value), "lb")
    elif choice == 2:
        show_result(value, "lb", lb_to_kg(value), "kg")
    elif choice == 3:
        show_result(value, "oz", oz_to_g(value), "g")
    elif choice == 4:
        show_result(value, "g", g_to_oz(value), "oz")
    elif choice == 5:
        show_result(value, "ft³", cubic_ft_to_cubic_m(value), "m³")
    else:
        show_result(value, "m³", cubic_m_to_cubic_ft(value), "ft³")


def volume_menu():
    """Let the user pick a volume conversion and print the result."""
    print("1) ml -> cc     2) cc -> ml")
    print("3) cup -> ml    4) ml -> cup")
    print("5) gal -> cup   6) cup -> gal")
    print("7) m³ -> ft³    8) ft³ -> m³")
    choice = ask_choice(8)
    value = read_value()
    if choice == 1:
        show_result(value, "ml", ml_to_cc(value), "cc")
    elif choice == 2:
        show_result(value, "cc", cc_to_ml(value), "ml")
    elif choice == 3:
        show_result(value, "cup", cup_to_ml(value), "ml")
    elif choice == 4:
        show_result(value, "ml", ml_to_cup(value), "cup")
    elif choice == 5:
        show_result(value, "gal", gal_to_cup(value), "cup")
    elif choice == 6:
        show_result(value, "cup", cup_to_gal(value), "gal")
    elif choice == 7:
        show_result(value, "m³", m3_to_ft3(value), "ft³")
    else:
        show_result(value, "ft³", ft3_to_m3(value), "m³")


def pressure_menu():
    """Let the user pick a pressure conversion and print the result."""
    print("1) atm -> kPa     2) kPa -> atm")
    print("3) atm -> mm Hg   4) mm Hg -> atm")
    print("5) bar -> Pa      6) Pa -> bar")
    print("7) psi -> mbar    8) mbar -> psi")
    choice = ask_choice(8)
    value = read_value()
    if choice == 1:
        show_result(value, "atm", atm_to_kpa(value), "kPa")
    elif choice == 2:
        show_result(value, "kPa", kpa_to_atm(value), "atm")
    elif choice == 3:
        show_result(value, "atm", atm_to_mmhg(value), "mm Hg")
    elif choice == 4:
        show_result(value, "mm Hg", mmhg_to_atm(value), "atm")
    elif choice == 5:
        show_result(value, "bar", bar_to_pa(value), "Pa")
    elif choice == 6:
        show_result(value, "Pa", pa_to_bar(value), "bar")
    elif choice == 7:
        show_result(value, "psi", psi_to_mbar(value), "mbar")
    else:
        show_result(value, "mbar", mbar_to_psi(value), "psi")


def energy_menu():
    """Let the user pick an energy conversion and print the result."""
    print("1) J -> kcal    2) kcal -> J")
    print("3) eV -> J      4) J -> eV")
    choice = ask_choice(4)
    value = read_value()
    if choice == 1:
        show_result(value, "J", j_to_kcal(value), "kcal")
    elif choice == 2:
        show_result(value, "kcal", kcal_to_j(value), "J")
    elif choice == 3:
        show_result(value, "eV", ev_to_j(value), "J")
    else:
        show_result(value, "J", j_to_ev(value), "eV")


def main():
    """Show the main menu until the user chooses 0 to quit."""
    while True:
        print()
        print("===== Unit Conversion Calculator =====")
        print("1) Length    2) Temperature   3) Mass")
        print("4) Volume    5) Pressure      6) Energy")
        print("0) Quit")
        unit_type = ask_choice(6, 0)
        if unit_type == 0:
            break
        elif unit_type == 1:
            length_menu()
        elif unit_type == 2:
            temperature_menu()
        elif unit_type == 3:
            mass_menu()
        elif unit_type == 4:
            volume_menu()
        elif unit_type == 5:
            pressure_menu()
        else:
            energy_menu()
    print("Goodbye!")


if __name__ == "__main__":
    main()
````

> 💡 **Explanation**
>
> **The conversion numbers (factors) are taken exactly from the chart in Fig. 1.** At the top of the code we wrote them all once as **constants**: names in capital letters, such as `FT_PER_M = 3.3` (meaning "how many feet in each metre").
>
> **Structure of the program, three layers:**
> 1. **38 small conversion functions**: for each pair of units, both directions (for example `m_to_ft` and `ft_to_m`). Each one is a single line: it takes one parameter `value` and `return`s the answer.
> 2. **Helper functions**: `read_value` (get a number), `ask_choice` (get a menu option), `show_result` (print it nicely).
> 3. **Menus**: one menu for each category (`length_menu`, `temperature_menu`, …) and `main`, which shows the main menu.
>
> **Line by line (the important points)**
> - **Multiply or divide?** The chart says `1 m = 3.3 ft`. So metres → feet means **multiply** by 3.3, and feet → metres means **divide** by 3.3. The same rule for every pair: one direction `*`, the opposite direction `/`.
> - **Temperature is different**: it's not just multiplying; there is adding and subtracting too. `c_to_f` is the chart's formula: `9 / 5 * value + 32`. `k_to_f` is not on the chart; we turned the formula `K = 5/9(F − 32) + 273` around: `F = 9/5 (K − 273) + 32`.
> - **The constants are global**: `FT_PER_M` is defined outside all the functions, so the functions can **read** it (scope). It would only cause trouble if you wanted to **change** it inside a function, and we don't need to here.
> - `ask_choice(highest, lowest=1)`: here a **default value** is used. The sub-menus call `ask_choice(6)`, meaning 1 to 6. The main menu calls `ask_choice(6, 0)`, meaning 0 to 6 (because 0 = quit).
> - `text.isdigit() and lowest <= int(text) <= highest`: first it checks that the text is digits, **then** it turns it into an int. The order matters: if `int("abc")` ran first, you would get a `ValueError`. If the first half of an `and` is `False`, the second half doesn't run at all.
> - `show_result`: `{value:g}` prints the number without extra zeros (`10`, not `10.000000`). `{result:.6g}` shows at most 6 significant digits, and for very small numbers it switches to scientific notation by itself: `1 eV = 1.602e-19 J`. (`e-19` means "times 10 to the power of minus 19".)
> - In Python, `2.39e-4` means `2.39 × 10⁻⁴` = `0.000239`. `10 ** 5` means `100000`.
>
> **⚠️ Some mistakes on the chart itself (good to know):**
> | On the chart | In reality | In our code |
> |---|---|---|
> | `1 in = 2.43 cm` | `1 in = 2.54 cm` | the chart's number (2.43), with a comment |
> | `1 m = 3.3 ft` | exactly `3.28084 ft` | the chart's number (rounded, that's fine) |
> | `1 ft = 0.028 m` in the Mass box | it means `1 ft³ = 0.028 m³` (volume!) | labelled `ft³ → m³` in the Mass menu |
> | `1 eV = 1.602 × 10⁻¹⁹ K` | the correct unit is **J** (joules) | `eV → J` |
> | `K = 5/9(°F − 32) + 273` | exactly `+ 273.15` | the chart's number (273) |
>
> Because all the numbers are in the constants at the top of the file, if you want to use the correct number you only change **one line** (for example `CM_PER_IN = 2.54`) and all the functions are fixed automatically. This is one of the reasons for using constants.
>
> **Hand check (with the chart's numbers):**
> - `10 m` → `10 × 3.3 = 33 ft` ✓
> - `100 °C` → `9/5 × 100 + 32 = 212 °F` ✓
> - `212 °F` → `5/9 × 180 + 273 = 373 K` ✓
> - `2 gal` → `2 × 16 = 32 cup` ✓
> - `1 atm` → `760 mm Hg` ✓
>
> **Trap:** in temperature, the order of operations matters: `5 / 9 * value - 32` is wrong; you need brackets: `5 / 9 * (value - 32)`.

## Task 3: Palindrome Checker
Write a function to check if a given string is a palindrome (reads the same backward as forward).

Some palindrome words are: civic, radar, level, rotor, kayak, madam, refer, rotator, deified, racecar and reviver.

The function should take one string, and return `True` if it is a palindrome, otherwise `False`.

````python
def is_palindrome(text):
    """Return True if text reads the same backward as forward."""
    cleaned = text.lower()
    return cleaned == cleaned[::-1]


words = ["civic", "radar", "level", "rotor", "kayak", "madam", "refer",
         "rotator", "deified", "racecar", "reviver", "Level", "python"]
for word in words:
    print(f"{word}: {is_palindrome(word)}")

user_word = input("Type a word to check: ")
print(is_palindrome(user_word))
````

> 💡 **Explanation**
>
> **Line by line**
> - `cleaned = text.lower()`: we make everything lowercase. Otherwise `Level` would not count as a palindrome, because `L` is different from `l`.
> - `cleaned[::-1]`: this is **slicing** with a step of `-1`, meaning "from the end to the start". `"radar"[::-1]` is `"radar"` and `"python"[::-1]` is `"nohtyp"`.
> - `return cleaned == cleaned[::-1]`: the comparison itself is a `True` or `False` value (Week 2, D3). So we `return` it directly. You don't need to write `if ...: return True else: return False`.
> - The `for` loop below just tests the function on the words from the task. At the end it also gets a word from the user.
>
> **Output:** all the task's words and `Level` → `True`, and `python` → `False`.
>
> **Trap:** if you give it a sentence (`"never odd or even"`), the spaces break it. For a sentence, remove the spaces first: `text.lower().replace(" ", "")`.
>
> **Second version (to understand it better):** if you want to see what happens behind `[::-1]`, you can also write it with a loop: compare the first letter with the last, the second with the second-to-last, and so on up to the middle.

````python
def is_palindrome(text):
    """Return True if text reads the same backward as forward (loop version)."""
    cleaned = text.lower()
    left = 0
    right = len(cleaned) - 1
    while left < right:
        if cleaned[left] != cleaned[right]:
            return False
        left += 1
        right -= 1
    return True


print(is_palindrome("racecar"))   # True
print(is_palindrome("Kayak"))     # True
print(is_palindrome("python"))    # False
````

> 💡 **Explanation**
>
> - `left` starts at the beginning of the word (index 0), `right` at the end (index `len - 1`).
> - Each round: if the two letters are different, `return False` **right there**. `return` ends the function immediately, so the rest of the loop doesn't run.
> - Then `left` moves one step forward and `right` one step back. When they meet (`left < right` becomes false), it means all the pairs were equal → `return True`.
>
> **Hand check with `kayak`:** k=k ✓, a=a ✓, then left=2 and right=2 → the loop ends → `True`.

## Task 4: Text Compression
Write a Python function called `compress_string` that takes a string as input and returns a compressed version of the string. The compression should replace repeated characters with the character followed by the count of repetitions. If the compressed string is longer than the original string, return the original string.

```python
compressed = compress_string("aaabbbbcccddeee")
# The expected output is "a3b4c3d2e3" because 'a' is repeated 3 times, 'b' 4 times, 'c' 3 times, etc.
compressed = compress_string("Hello")
# The expected output is "Hello", because the compressed string in this case H1e1l2o1 is longer than the original string "Hello"
```

````python
def compress_string(text):
    """Return text with each run of a repeated character written as char + count."""
    if text == "":
        return text

    compressed = ""
    current_char = text[0]
    count = 1
    for char in text[1:]:
        if char == current_char:
            count += 1
        else:
            compressed += current_char + str(count)
            current_char = char
            count = 1
    compressed += current_char + str(count)   # add the last run

    if len(compressed) > len(text):
        return text
    return compressed


print(compress_string("aaabbbbcccddeee"))   # a3b4c3d2e3
print(compress_string("Hello"))             # Hello
````

> 💡 **Explanation**
>
> **Main idea: count the "runs".** We keep one letter (`current_char`) and count how many times it has come in a row (`count`). Whenever the letter changes, we add the result of the previous run to the answer and start counting again.
>
> **Line by line**
> - `if text == "": return text`: we return empty text straight away; otherwise `text[0]` gives an `IndexError`.
> - `current_char = text[0]` and `count = 1`: we have counted the first letter.
> - `for char in text[1:]`: from the **second** letter onwards (`[1:]` means from index 1 to the end).
> - If the letter is the same as before → `count += 1`.
> - If it changed → add `current_char + str(count)` to `compressed`. `str(count)` is needed because you can't add text and a number (`"a" + 3` → `TypeError`). Then keep the new letter and set `count = 1`.
> - **The line after the loop** (`# add the last run`): the last run never sees a "change", so we have to add it by hand. **This is the most common bug in this task!** Without this line the answer is `a3b4c3d2` (no `e3`).
> - `if len(compressed) > len(text): return text`: the task's own rule. `"Hello"` (5 letters) → `"H1e1l2o1"` (8 letters) is longer, so `"Hello"` is returned.
>
> **Hand check with `aaabbbbcccddeee`:** a×3, b×4, c×3, d×2, e×3 → `a3b4c3d2e3` (10 letters, shorter than 15) ✓
>
> **A small point:** the task says "if it is **longer**". So if the lengths are **equal** (for example `"aabb"` → `"a2b2"`), the compressed version is returned. If you also want to return the original when they are equal, change `>` to `>=`.

## Task 5: Lambda Rewrite
Rewrite the quiz or the scientific converter including mainly lambda functions.

````python
import random

# Each maths operation is now a one-line lambda, stored in a dictionary
# under its symbol. Division and remainder give None when b is 0.
OPERATIONS = {
    "+": lambda a, b: a + b,
    "-": lambda a, b: a - b,
    "*": lambda a, b: a * b,
    "/": lambda a, b: round(a / b, 2) if b != 0 else None,
    "**": lambda a, b: a ** b,
    "%": lambda a, b: a % b if b != 0 else None,
}

# Small helpers written as lambdas too
random_digit = lambda: random.randint(0, 9)
is_number = lambda text: (text[1:] if text.startswith("-") else text).replace(".", "", 1).isdigit()
is_correct = lambda answer, correct: round(answer, 2) == correct


def main():
    """Run the lambda version of the quiz until the user types q."""
    score = 0
    questions = 0
    while True:
        print()
        print("Operations:  +   -   *   /   **   %")
        symbol = input("Type the operation to practise (or q to quit): ").strip().lower()
        if symbol == "q":
            break
        if symbol not in OPERATIONS:
            print("Sorry, that operation is not on the list.")
            continue

        number1 = random_digit()
        number2 = random_digit()
        correct = OPERATIONS[symbol](number1, number2)
        if correct is None:
            print(f"Error: {number1} {symbol} {number2} divides by zero. Let's skip this one.")
            continue

        if symbol == "/":
            print("(Give division answers to 2 decimal places.)")
        text = input(f"What is {number1} {symbol} {number2}? ").strip()
        while not is_number(text):
            text = input("Please type a number: ").strip()

        questions += 1
        if is_correct(float(text), correct):
            score += 1
            print("Correct! Well done.")
        else:
            print(f"Not quite. The answer is {correct}.")
        print(f"Score: {score} / {questions}")

    print(f"Final score: {score} out of {questions}. Goodbye!")


if __name__ == "__main__":
    main()
````

> 💡 **Explanation**
>
> **What changed?** The six `def` functions from Task 1 became six lambdas in a **dictionary**. The key for each one is its symbol (`"+"`, `"**"`, …). Dictionaries are Week 4 material; for now, just know it is a "key → value" table.
>
> **Line by line**
> - `"+": lambda a, b: a + b`: the key is `"+"`, and the value is a small function that adds two numbers.
> - `lambda a, b: round(a / b, 2) if b != 0 else None`: this is the **one-line if (conditional expression)** from Week 3. It means "if `b` is not zero, divide; otherwise `None`". This way we catch the division-by-zero error inside the lambda itself.
> - `OPERATIONS[symbol]`: uses the key to get **the function itself**. `OPERATIONS[symbol](number1, number2)`: **calls** that function with the two numbers. This one line replaces the whole `if/elif` chain from Task 1!
> - `random_digit = lambda: random.randint(0, 9)`: a lambda with no parameters is possible too. We call it with `random_digit()`.
> - `is_number`: the same function as in Task 1, squeezed into one line. `(text[1:] if text.startswith("-") else text)` means "if it starts with `-`, take it without the `-`".
> - `is_correct`: rounds the user's answer, compares it with the correct answer, and returns `True`/`False`.
> - `main` is still a normal `def`. Why? Because it has a `while` loop and `input`, and these **don't fit** in a lambda. So it's "mostly lambda", not "all lambda".
> - The menu is simpler this time: the user types the symbol itself (`+`, `**`, …) and `if symbol not in OPERATIONS` checks whether it is in the dictionary.
>
> **When lambda, when def?**
> | lambda | def |
> |---|---|
> | one short expression | several lines, loops, a full `if` |
> | usually no name, inside a dictionary or as an argument | has a name and a docstring |
> | `lambda a, b: a + b` | `def add(a, b): return a + b` |
>
> **A professional tip:** putting a lambda in a variable (`random_digit = lambda: …`) works, but Python's style guide (PEP 8) says that if you want a name, `def` is better. The main place for a lambda is when you put the function **directly** somewhere, like this dictionary, or `sorted(names, key=lambda name: len(name))`, which you will see later. Here we did it because the task itself asked for "mostly lambda".
>
> **Hand check:** `OPERATIONS["**"](2, 3)` → `8`. `OPERATIONS["%"](7, 0)` → `None` → error message.

## Task 7: Pure Python EDA: Mini Statistics Report
Ask for five numeric values (space-separated). Print the values, total, mean, median, and sample standard deviation.

**Example Output:**
```
Values: 80, 90, 70, 85, 75
Total = 400.00
Mean = 80.00
Median = 80.00
Std Dev = 7.91
```
*Optional hint:* `import statistics as stats`

````python
import statistics as stats


def read_five_numbers():
    """Ask until the user types exactly five numbers; return the typed pieces."""
    while True:
        parts = input("Enter five numbers separated by spaces: ").split()
        if len(parts) == 5:
            return parts
        print("Please enter exactly five numbers.")


parts = read_five_numbers()
values = []
for part in parts:
    values.append(float(part))

shown = ", ".join(parts)
print(f"Values: {shown}")
print(f"Total = {sum(values):.2f}")
print(f"Mean = {stats.mean(values):.2f}")
print(f"Median = {stats.median(values):.2f}")
print(f"Std Dev = {stats.stdev(values):.2f}")
````

> 💡 **Explanation**
>
> **⚠️ Run this version in VS Code on your Mac.** The online book's browser Python doesn't have the `statistics` module. The second version below (calculated by hand) works in the browser too.
>
> **Line by line**
> - `import statistics as stats`: imports the module with a shorter name (`stats`). `as` is just a nickname.
> - `input(...).split()`: cuts the text at the spaces: `"80 90 70 85 75"` → `["80", "90", "70", "85", "75"]`. This is a **list** (Week 4); for now, just know it is a container that holds several values.
> - `if len(parts) == 5`: if there aren't exactly 5, ask again.
> - The `for` loop + `values.append(float(part))`: turns each text piece into a number and adds it to the `values` list.
> - `", ".join(parts)`: joins the pieces with `, ` → `80, 90, 70, 85, 75`. We used `parts` (the original text) so the numbers are shown exactly as they were typed (`80`, not `80.0`).
> - `sum(values)`: the total. `stats.mean`: the mean. `stats.median`: the median. `stats.stdev`: the **sample standard deviation**.
> - `:.2f`: two decimal places, exactly like the sample output.
>
> **Trap:** `stats.pstdev` is the **population** standard deviation (divide by n). The task asks for the **sample** one, which is `stats.stdev` (divide by n − 1). With the sample numbers, `pstdev` gives `7.07`, which is wrong.

````python
import math


def mean(values):
    """Return the average of the values."""
    return sum(values) / len(values)


def median(values):
    """Return the middle value (or the average of the two middle values)."""
    ordered = sorted(values)
    middle = len(ordered) // 2
    if len(ordered) % 2 == 1:
        return ordered[middle]
    return (ordered[middle - 1] + ordered[middle]) / 2


def sample_std_dev(values):
    """Return the sample standard deviation (divide by n - 1)."""
    average = mean(values)
    total_squares = 0
    for value in values:
        total_squares += (value - average) ** 2
    return math.sqrt(total_squares / (len(values) - 1))


parts = input("Enter five numbers separated by spaces: ").split()
while len(parts) != 5:
    print("Please enter exactly five numbers.")
    parts = input("Enter five numbers separated by spaces: ").split()

values = []
for part in parts:
    values.append(float(part))

shown = ", ".join(parts)
print(f"Values: {shown}")
print(f"Total = {sum(values):.2f}")
print(f"Mean = {mean(values):.2f}")
print(f"Median = {median(values):.2f}")
print(f"Std Dev = {sample_std_dev(values):.2f}")
````

> 💡 **Explanation**
>
> **Second version: we calculate everything ourselves (without `statistics`).** This runs in the browser too, and more importantly, you understand what happens behind `stdev`.
>
> **The functions**
> - `mean`: the total divided by the count.
> - `median`: first we **sort** (`sorted`). If the count is odd, the middle number is the answer. If it is even, the average of the two middle numbers. `//` is whole-number division: `5 // 2` → `2`, which is the index of the middle.
> - `sample_std_dev`: its formula is:
>
> ```
> std = sqrt( sum((x - mean)²) / (n - 1) )
> ```
> That means: work out how far each number is from the mean, square it, add them all up, divide by **n − 1**, and finally take the square root (`math.sqrt`).
>
> **Hand check with `80 90 70 85 75`** (we also checked that the task's sample output is correct ✓):
>
> | x | x − 80 | (x − 80)² |
> |---|---|---|
> | 80 | 0 | 0 |
> | 90 | 10 | 100 |
> | 70 | −10 | 100 |
> | 85 | 5 | 25 |
> | 75 | −5 | 25 |
> | **Total** | | **250** |
>
> - Total = 400 → `400.00` ✓, Mean = 400 / 5 = 80 → `80.00` ✓
> - Median: sorted `70, 75, 80, 85, 90` → the middle one is `80` → `80.00` ✓
> - Std Dev = √(250 / 4) = √62.5 ≈ 7.9057 → `7.91` ✓
>
> **Why n − 1?** When you only have a **sample** of the data (not the whole population), dividing by n − 1 gives a better estimate. If you divide by n you get √50 ≈ `7.07`, which is the population standard deviation.

## Task 8: IT: Path Decomposition Report
Ask for a file or directory path and print its absolute path, parent, name, stem, and suffix.

Absolute path: /home/user/project/service.log
Parent: /home/user/project

*Optional hint:* `from pathlib import Path`

````python
from pathlib import Path


def print_path_report(path_text):
    """Print the absolute path, parent, name, stem and suffix of a path."""
    path = Path(path_text).expanduser().resolve()
    print(f"Absolute path: {path}")
    print(f"Parent: {path.parent}")
    print(f"Name: {path.name}")
    print(f"Stem: {path.stem}")
    print(f"Suffix: {path.suffix}")


path_text = input("Enter a file or directory path: ").strip()
print_path_report(path_text)
````

> 💡 **Explanation**

> **Line by line**
> - `from pathlib import Path`: `Path` turns a path into an **object** that knows its own parts. You no longer need to cut the text up with `split("/")`.
> - `.expanduser()`: if the path starts with `~` (meaning your home folder, for example `~/Desktop`), it turns it into the real path.
> - `.resolve()`: makes the path **absolute** (complete). If you only write `service.log`, it joins it to the folder the program is running in right now.
> - `path.parent`: the folder the file is in. `path.name`: the full name. `path.stem`: the name without the extension. `path.suffix`: the extension, with its dot.
>
> | `/home/user/project/service.log` | |
> |---|---|
> | `parent` | `/home/user/project` |
> | `name` | `service.log` |
> | `stem` | `service` |
> | `suffix` | `.log` |
>
> **Traps**
> - The file doesn't need to really exist. `Path` only analyses the **text of the path**.
> - If you give it a folder (for example `project`), `suffix` prints as empty, because a folder has no extension.
> - `archive.tar.gz` → `stem` is `archive.tar` and `suffix` is `.gz` (only the last extension).

## Task 10: Pentagon Area
Ask for the side length `s` and compute the area: `area = 5*s**2 / (4*tan(pi/5))`. Print to 2 d.p.

Area = 172.05

````python
from math import tan, pi


def pentagon_area(side):
    """Return the area of a regular pentagon with the given side length."""
    return 5 * side ** 2 / (4 * tan(pi / 5))


side = float(input("Enter the side length s: "))
print(f"Area = {pentagon_area(side):.2f}")
````

> 💡 **Explanation**

> - `tan(pi / 5)`: **important trap:** the `math` functions take angles in **radians**, not degrees. `pi / 5` radians = 36 degrees. If you write `tan(36)`, the answer is completely wrong.
> - The brackets around `(4 * tan(pi / 5))` are needed; otherwise Python divides by 4 first and then multiplies by tan.
> - `float(input(...))`: because the side length may have decimals (`2.5`).
