## Rock, Paper, Scissors
`get_computer_choice()` returns rock/paper/scissors from `random.randint(0, 2)`; `decide_winner(user_choice, computer_choice)` returns `"draw"`, `"user"` or `"computer"`; ask the user, print the computer's choice and the winner.

```python
import random


def get_computer_choice():
    """Return rock, paper or scissors at random."""
    number = random.randint(0, 2)
    if number == 0:
        return "rock"
    elif number == 1:
        return "paper"
    else:
        return "scissors"


def decide_winner(user_choice, computer_choice):
    """Return "draw", "user" or "computer"."""
    if user_choice == computer_choice:
        return "draw"
    elif (user_choice == "rock" and computer_choice == "scissors") or \
         (user_choice == "paper" and computer_choice == "rock") or \
         (user_choice == "scissors" and computer_choice == "paper"):
        return "user"
    else:
        return "computer"


user_choice = input("Choose rock, paper or scissors: ").lower()
computer_choice = get_computer_choice()
print(f"Computer chose: {computer_choice}")

winner = decide_winner(user_choice, computer_choice)
print(f"Winner: {winner}")
```

> 💡 **Explanation**
>
> **Line by line**
> - `import random` goes at the top of the file, because the first function uses it.
> - `get_computer_choice` turns the number 0, 1 or 2 into a word and **returns** it. It doesn't print it. This way the rest of the program can use it.
> - `decide_winner`: first it checks for a draw. Then it joins the three cases where the user wins with `or`. Each case has two conditions joined with `and`. Anything else means the computer won.
> - `\` at the end of a line means "this statement continues on the next line". The brackets around each case also make it easier to read.
> - The main part (outside the functions) only calls the functions, stores the results and prints them.
>
> **Check:** rock against scissors → `user`. paper against scissors → `computer`.

## Calculator
Four functions (`add`, `subtract`, `multiply`, `divide`), a `while keep_going:` loop that asks for an operator and a number, calls the matching function, prints the result, and either continues with the result (`y`) or starts over. Extension (after Week 4): replace the `if`/`elif` with a dictionary of functions.

```python
def add(n1, n2):
    """Return n1 + n2."""
    return n1 + n2

def subtract(n1, n2):
    """Return n1 - n2."""
    return n1 - n2

def multiply(n1, n2):
    """Return n1 * n2."""
    return n1 * n2

def divide(n1, n2):
    """Return n1 / n2."""
    return n1 / n2

operations = {"+": add, "-": subtract, "*": multiply, "/": divide}

n1 = float(input("What's the first number? "))
keep_going = True

while keep_going:
    operator = input("Pick an operation (+ - * /): ")
    if operator not in operations:
        print("Unknown operation, try again.")
        continue
    n2 = float(input("What's the next number? "))

    result = operations[operator](n1, n2)
    print(f"{n1} {operator} {n2} = {result}")

    choice = input("Type 'y' to continue with the result, 'n' to start over, or 'q' to quit: ")
    if choice == "y":
        n1 = result
    elif choice == "n":
        n1 = float(input("What's the first number? "))
    else:
        keep_going = False

print("Goodbye!")
```

> 💡 **Explanation**
>
> **An interesting idea: a dictionary of functions.** In Python, a function is also a value, and you can put it in a dictionary. **Careful: no brackets!** `add` means "the function itself"; `add(...)` means "run it".
>
> - `operations[operator]` with the key `"+"` gives back the function `add`.
> - `operations[operator](n1, n2)` calls that function with the two numbers.
> - This one line replaces the whole `if/elif` chain. If you want to add a new operator (for example `**`), you only add one function and one key.
> - `if operator not in operations: continue`: if the operator is not valid, it goes back to the start of the loop and the program doesn't crash.
> - `keep_going = False` ends the loop. This is another way to leave a `while` loop instead of `break`.
>
> **Risk:** dividing by zero → `ZeroDivisionError`. You can fix this with `try/except` from Week 4.

## Number Guessing Game, upgraded
Open your Week 2 game, add `import random`, and replace the fixed secret with `random.randint(1, 100)`.

```python
import random

secret = random.randint(1, 100)
guesses = 0

print("I'm thinking of a number between 1 and 100.")
while True:
    guess = int(input("Make a guess: "))
    guesses += 1
    if guess > secret:
        print("Too high.")
    elif guess < secret:
        print("Too low.")
    else:
        print(f"You got it! The answer was {secret}. It took you {guesses} guesses.")
        break
```

> 💡 **Explanation**
>
> Only **two lines** have changed: `import random` and `secret = random.randint(1, 100)`. The rest of the game is exactly the same. This is why we write clean code: you change one part and the rest keeps working untouched.
>
> `randint(1, 100)` includes **both ends** (1 and 100 can both come up). This is different from `range`, where the last number is not included.
