# Week 3 · Solved lab, explained

This week you package code into **functions**: named, reusable blocks. A function is like a recipe: you write it once (`def`), then *use* it (call it) as many times as you want.

## Warm-up 1 · `01_missing_argument.py`

```python
def status_of(percent, warning_at):
    ...
print(status_of(95))
```

```
TypeError: status_of() missing 1 required positional argument: 'warning_at'
```

**What goes wrong:** the function says it needs **two** inputs (`percent` and `warning_at`), but the call only gives **one** (`95`). Python doesn't know what `warning_at` should be.

**Two possible fixes:** pass the second argument, *or* give the parameter a **default value** so it can be skipped. The default is the better fix here, because 90 is the normal warning level.

````python
def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

print(status_of(95))        # uses the default 90  -> WARNING
print(status_of(95, 96))    # gives its own value  -> OK
````

> 💡 **Explanation**
>
> `warning_at=90` in the `def` line means: *if the caller doesn't give a value, use 90*. Parameters with defaults must come **after** the ones without (`def f(a, b=2)` is fine, `def f(a=1, b)` is a `SyntaxError`).

## Warm-up 2 · `02_scope_error.py`

```python
def check(value, limit):
    status = "OVER LIMIT" if value > limit else "OK"

check(87, 100)
print(status)
```

```
NameError: name 'status' is not defined
```

**What goes wrong:** `status` is created **inside** the function, so it is **local**: it exists only while the function runs, and disappears when the function ends. Outside, Python has never heard of it.

**The fix:** `return` the value from the function, and store it when you call it.

````python
def check(value, limit):
    status = "OVER LIMIT" if value > limit else "OK"
    return status

status = check(87, 100)
print(status)
````

> 💡 **Explanation**
>
> **`return` is the only door out of a function.** Values inside a function stay inside, unless you `return` them. The caller then catches the returned value in a variable: `status = check(87, 100)`.
>
> The line `"OVER LIMIT" if value > limit else "OK"` is a one-line `if`/`else` (a *conditional expression*): *give "OVER LIMIT" if the condition is True, otherwise "OK"*.

## D3 · `return` a value + docstring

**What it asks:** `percent_of(part, total)` **returns** the percentage (no print inside), with a one-line docstring; store `percent_of(87, 100)` and print it; run `help(percent_of)`.

````python
def percent_of(part, total):
    """Return part as a percentage of total."""
    return (part / total) * 100

result = percent_of(87, 100)
print(result)
help(percent_of)
````

> 💡 **Explanation**
>
> - The **docstring** is the text in triple quotes on the first line of the body. It's documentation that Python stores *with* the function, so `help()` can show it. CW1's code-quality marks expect one on every function.
> - `return` **sends the answer back** to whoever called. `result = percent_of(87, 100)` catches it: `result` is `87.0`.
> - Why not `print` inside? Because a returned value can be **used**: stored, compared, passed to another function (C1 does exactly that). A printed value is only shown on screen and then it's gone.

## D4 · `print` is not the same as `return`

**What it asks:** predict, run, explain why the second line is `None`.

````python
# Prediction: it prints 87.0 (from inside the function), then None.
def show_percent(part, total):
    print((part / total) * 100)

result = show_percent(87, 100)
print(result)
# The second line is None because show_percent has no return statement.
# A function without return gives back None ("nothing"),
# so result stores None even though 87.0 was shown on the screen.
````

# Topic 2 · Parameters and arguments

## D5 · Arguments are matched by position

````python
def within_limit(value, limit):
    return value <= limit

print(within_limit(87, 100))
print(within_limit(100, 87))
# Different because arguments are matched by POSITION:
# first call: value=87, limit=100 -> 87 <= 100 -> True
# second call: value=100, limit=87 -> 100 <= 87 -> False
````

> 💡 **Explanation**
>
> Python fills parameters **left to right**: the first argument goes into the first parameter, and so on. It doesn't look at what the numbers *mean*. Swapping the arguments silently gives a different answer, with no error.

## D6 · A default value

**What it asks:** `status_of(percent, warning_at=90)` returning the three statuses; print `status_of(95)`, `status_of(85)`, `status_of(85, 80)`.

````python
def status_of(percent, warning_at=90):
    """Return OVER LIMIT, WARNING or OK for a percentage."""
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

print(status_of(95))
print(status_of(85))
print(status_of(85, 80))
````

> 💡 **Explanation**
>
> - `status_of(95)`: warning_at = 90 (default) → 95 ≥ 90 → **WARNING**
> - `status_of(85)`: 85 < 90 → **OK**
> - `status_of(85, 80)`: warning_at = 80 → 85 ≥ 80 → **WARNING**
>
> Notice: as soon as a `return` runs, the function **ends**. That's why `elif`/`else` aren't strictly needed after a `return`, but they make the three cases easy to read.

## D7 · Keyword arguments

````python
print(status_of(percent=95, warning_at=96))
print(status_of(warning_at=96, percent=95))
print(status_of(95, warning_at=96))
````

> 💡 **Explanation**
>
> All three print **OK** (95 is below 96). **Keyword arguments** use the parameter name (`percent=95`), so the **order doesn't matter** and the call explains itself.
>
> The lab then asks you to try `status_of(percent=95, 96)` and read the error:
>
> ```
> SyntaxError: positional argument follows keyword argument
> ```
>

## D8 · `*args` and `**kwargs`

**What it asks:** (a) `total(*numbers)` returns the sum of any number of numbers; (b) `configure(**settings)` prints each setting as `key = value`.

````python
def total(*numbers):
    """Return the sum of all the numbers passed in."""
    result = 0
    for n in numbers:
        result += n
    return result

print(total(1, 2, 3))
print(total(10, 20, 30, 40))
print(total())


def configure(**settings):
    """Print each setting as key = value."""
    for key, value in settings.items():
        print(f"{key} = {value}")

configure(host="10.0.0.1", port=8080)
````

> 💡 **Explanation**
>
> - `*numbers`: the `*` collects **any number of positional arguments** into a tuple. `total(1, 2, 3)` → `numbers = (1, 2, 3)`. `total()` → `()`, so the loop never runs and 0 is returned.
> - `**settings`: the `**` collects **any number of keyword arguments** into a dictionary: `{"host": "10.0.0.1", "port": 8080}`. `.items()` gives each key-value pair (dictionaries are Week 4).
> - I added the numbers with a loop to show the idea; `return sum(numbers)` does the same in one line.

# Topic 3 · Scope

**Scope** = where a name can be seen. A name created inside a function is **local** to it.

## D9 · Local means local

````python
# Prediction: "inside: OK" is printed, then a NameError for result.
def make_status():
    result = "OK"
    print("inside:", result)

make_status()
print("outside:", result)
# result was created inside make_status, so it only exists while the
# function runs. Outside the function the name does not exist -> NameError.
````

> 💡 **Explanation**
>
> Running this cell **should** end with `NameError`. That is the expected output. Think of a function as a room: things made inside the room stay in the room. `return` is how you hand something out through the door (Warm-up 2).

## D10 · Parameters are local too

````python
# Prediction: 42, then 21.
def double(n):
    n = n * 2
    return n

number = 21
print(double(number))
print(number)
# number is still 21: the function received a COPY of the value (21) in its
# own local name n. Changing n inside does not change number outside.
````

> 💡 **Explanation**
>
> The call `double(number)` passes the **value** 21 into the new local name `n`. `n` becomes 42 and is returned, but `number` outside is a different name and is never touched. (This is true for numbers and text; Week 4's lists behave differently, you'll see why then.)

## D12 · Returning more than one value

````python
def check_record(value, limit):
    """Return the difference and the percentage of value against limit."""
    difference = value - limit
    percent = (value / limit) * 100
    return difference, percent

difference, percent = check_record(87.0, 100.0)
print("Difference:", difference)
print("Percent:", percent)
````

> 💡 **Explanation**
>
> `return difference, percent` returns **two values packed together** (a tuple: `(-13.0, 87.0)`). On the calling side, `difference, percent = ...` **unpacks** them into two names, in the same order. You'll use exactly this in the mini-project's `check()` function.

## D15 · `if __name__ == "__main__":`

````python
import math

print(__name__)

def area_of_circle(radius):
    """Return the area of a circle with the given radius."""
    return math.pi * radius ** 2

if __name__ == "__main__":
    print(round(area_of_circle(5), 2))
    # This would NOT run if another file imported this one
    # (import my_file): then __name__ is "my_file", not "__main__".
````

> 💡 **Explanation**
>
> Every Python file has a hidden variable `__name__`:
> - when you **run** the file directly, `__name__` is `"__main__"`
> - when another file **imports** it, `__name__` is the file's name
>
> (The hub may show `78.54000000000001`: a display quirk of the in-browser Python. Real Python on your Mac prints `78.54`.)
>
> So `if __name__ == "__main__":` means *"only do this when I am the program being run, not when I'm being used as a module"*. It lets one file be both a **toolbox** (its functions can be imported) and a **program** (it runs its own test when started). Area: π × 5² = 78.54.

## C1 · Wire two functions together

**What it asks:** use `percent_of` (D3) and `status_of` (D6) **together** to print the status for 87/100, 120/100, 45/100, without rewriting their logic.

````python
def percent_of(part, total):
    """Return part as a percentage of total."""
    return (part / total) * 100

def status_of(percent, warning_at=90):
    """Return OVER LIMIT, WARNING or OK for a percentage."""
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

print(status_of(percent_of(87, 100)))
print(status_of(percent_of(120, 100)))
print(status_of(percent_of(45, 100)))
````

> 💡 **Explanation**
>
> **Read `status_of(percent_of(87, 100))` from the inside out:** first `percent_of(87, 100)` runs and returns `87.0`; that value becomes the argument of `status_of(87.0)`, which returns `"OK"`; `print` shows it.
>
> This only works because `percent_of` **returns** (D4's lesson). I repeated the two definitions so this cell works on its own; in the lab notebook you'd just reuse them from D3 and D6. *"Do not write the logic a second time"* is the key principle: each calculation lives in **one** place.

## C2 · How many are over?

**What it asks:** `count_over(*values, limit=100)` with a docstring, returning how many values are over the limit.

````python
def count_over(*values, limit=100):
    """Return how many of the values are greater than limit."""
    count = 0
    for value in values:
        if value > limit:
            count += 1
    return count

print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101, limit=50))
print(count_over())
````

> 💡 **Explanation**
>
> - `*values` collects any number of values (D8); `limit=100` is a default (D6). Because it comes **after** `*values`, it can **only** be given by keyword (`limit=50`); a plain number would be swallowed into `values`.
> - Inside: the **counter pattern** from Week 2 (start at 0, add 1 when the condition is True).
> - Results: 120 and 101 are over 100 → **2**; with limit 50: 87, 120, 101 → **3**; no values → the loop doesn't run → **0**.

# Mini-project · Record Check, version 3: functions

Same program as Week 2 (IT lane: hostname, GB used, GB total), but now **organised into functions**. The output looks the same; the *code* becomes cleaner and reusable. The brief's rules: every function does **one job**, nothing is calculated twice, every function has a docstring.

## Threshold · `status_of(percent)`, the rest inline

````python
def status_of(percent):
    """Return OVER LIMIT, WARNING or OK for a percentage."""
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= 90:
        return "WARNING"
    else:
        return "OK"


# INPUT
label = input("Hostname: ")
value = float(input("GB used: "))
limit = float(input("GB total: "))

# PROCESS
difference = limit - value
percent = value / limit * 100
status = status_of(percent)

# OUTPUT
print()
print("=" * 34)
print(f"  RECORD CHECK  -  {label}")
print("=" * 34)
print(f"  Used        : {value:.0f}")
print(f"  Total       : {limit:.0f}")
print(f"  Status      : {status}")
print("=" * 34)
````

> 💡 **Explanation**
>
> Compare with Week 2: the `if/elif/else` moved **into a function** that `return`s the status, and the main code just calls `status = status_of(percent)`. Functions are defined at the **top** of the file, because Python must read the `def` before you can call it.

## Typical · add `check(value, limit)`; numbers 2 decimals, right-aligned

````python
def status_of(percent):
    """Return OVER LIMIT, WARNING or OK for a percentage."""
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= 90:
        return "WARNING"
    else:
        return "OK"


def check(value, limit):
    """Return the space left and the percentage used."""
    difference = limit - value
    percent = value / limit * 100
    return difference, percent


label = input("Hostname: ")
value = float(input("GB used: "))
limit = float(input("GB total: "))

difference, percent = check(value, limit)
status = status_of(percent)

print()
print("=" * 34)
print(f"  RECORD CHECK  -  {label}")
print("=" * 34)
print(f"  Used        : {value:>10.2f}")
print(f"  Total       : {limit:>10.2f}")
print(f"  Free        : {difference:>10.2f}")
print(f"  Percent     : {percent:>10.2f} %")
print(f"  Status      : {status:>10}")
print("=" * 34)
````

> 💡 **Explanation**
>
> `check()` **only calculates and returns** (no print): that's the template's instruction and the "one job" rule. It returns two values, unpacked with `difference, percent = check(value, limit)` (D12).

## Excellent · `print_report(...)` does all printing + loop until `quit` + count OVER LIMIT

````python
"""
RECORD CHECK  -  my version
Name  : Parsa
Lane  : IT
Run it:   python template.py
"""

# ================================================================ FUNCTIONS
def status_of(percent):
    """Return OVER LIMIT, WARNING or OK for a percentage."""
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= 90:
        return "WARNING"
    else:
        return "OK"


def check(value, limit):
    """Return the space left and the percentage used."""
    difference = limit - value
    percent = value / limit * 100
    return difference, percent


def print_report(label, value, limit, difference, percent, status):
    """Print the bordered report for one record."""
    print()
    print("=" * 34)
    print(f"  RECORD CHECK  -  {label}")
    print("=" * 34)
    print(f"  Used        : {value:>10.2f}")
    print(f"  Total       : {limit:>10.2f}")
    print(f"  Free        : {difference:>10.2f}")
    print(f"  Percent     : {percent:>10.2f} %")
    print(f"  Status      : {status:>10}")
    print("=" * 34)


# ===================================================================== MAIN
over_limit_count = 0

while True:
    # INPUT
    label = input("Hostname (or quit to stop): ")
    if label == "quit":
        break
    value = float(input("GB used: "))
    limit = float(input("GB total: "))

    # PROCESS
    difference, percent = check(value, limit)
    status = status_of(percent)
    if status == "OVER LIMIT":
        over_limit_count += 1

    # OUTPUT
    print_report(label, value, limit, difference, percent, status)

print(f"Session finished: {over_limit_count} record(s) OVER LIMIT.")
````

> 💡 **Explanation**
>
> **How the program is now organised**
>
> | Part | Job | Prints? |
> |---|---|---|
> | `status_of(percent)` | decide the status | no |
> | `check(value, limit)` | calculate | no |
> | `print_report(...)` | display one record | **yes, all of it** |
> | main loop | ask, call the functions, count | only the final summary |
>
> - Every report `print()` lives inside `print_report`, as the brief requires. (The final summary line is outside the loop by design: it's the session total, not the report.)
> - `if status == "OVER LIMIT": over_limit_count += 1`: I count by comparing the **returned status**, so the 100% rule is written **only once** (inside `status_of`). Writing `if percent >= 100` here would duplicate the logic: if the rule ever changes, you'd have to change two places.
> - The loop and `quit` are Week 2's pattern; only the inside got shorter.
>
> **Why bother with functions if the output is the same?** Each piece can be tested alone (`print(status_of(95))`), reused (C1), and changed in one place. That's the "code quality" part of the CW1 marks.

# Your `calculator.py`

You already started the *Calculator* extra project in your Week 3 LAB folder. Here is a complete version of the instructions in that file, explained. (We'll do all the extra projects properly later, as you planned; this one is here because it's already in your folder.)

````python
def add(n1, n2):
    """Return n1 + n2."""
    return n1 + n2

def subtract(n1, n2):
    """Return n1 - n2."""
    return n1 - n2

def multiply(n1, n2):
    """Return n1 * n2."""
    return n1 * n2

def divide(n1, n2):
    """Return n1 / n2."""
    return n1 / n2


n1 = float(input("First number: "))
keep_going = True

while keep_going:
    operator = input("Operator (+ - * /): ")
    n2 = float(input("Next number: "))

    if operator == "+":
        result = add(n1, n2)
    elif operator == "-":
        result = subtract(n1, n2)
    elif operator == "*":
        result = multiply(n1, n2)
    elif operator == "/":
        result = divide(n1, n2)
    else:
        print("Unknown operator")
        continue

    print(f"{n1} {operator} {n2} = {result}")

    again = input("Type 'y' to continue with the result, or 'n' to start over: ")
    if again == "y":
        n1 = result
    else:
        n1 = float(input("First number: "))
````

> 💡 **Explanation**
>
> - Four tiny functions, one job each (step 1–4 of the instructions).
> - `keep_going = True` + `while keep_going:` (steps 6–7). Note: this loop never sets `keep_going` to False, exactly as the instructions say, so you stop it with ■ / Ctrl+C. **Improvement for you:** add a choice `q` that sets `keep_going = False`.
> - The `if/elif` chain picks which function to call (step 10). I added an `else` with `continue` so a wrong operator doesn't crash on `print(result)`.
> - `n1 = result` chains calculations: 2 + 3 = 5, then 5 * 4 = 20 …
