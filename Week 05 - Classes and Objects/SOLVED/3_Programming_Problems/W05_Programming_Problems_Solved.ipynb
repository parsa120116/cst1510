## Task 1: Student Class

Create a class `Student` with attributes `name` and `age` set in `__init__`. Create two student objects and print their attributes using dot notation.

````python
class Student:
    """A student with a name and an age."""

    def __init__(self, name, age):
        """Store the name and age on the new object."""
        self.name = name
        self.age = age


student1 = Student("Alice", 19)
student2 = Student("Parsa", 20)

print(student1.name, student1.age)
print(student2.name, student2.age)

print(f"{student1.name} is {student1.age} years old.")
print(f"{student2.name} is {student2.age} years old.")
````

> 💡 **Explanation**
>
> **Line by line**
> - `class Student:` starts the template. A class name starts with a **capital** letter (this is a convention, not a rule).
> - `def __init__(self, name, age):` is a special method that runs **automatically** when an object is created. "init" means initialise (get ready).
> - `self.name = name`: the right side is the **parameter** (the value you sent). The left side, `self.name`, stores it **on the object itself**. This is called an **instance attribute**: each student has their own `name`.
> - `student1 = Student("Alice", 19)`: Python makes an empty object, calls `__init__` with `self` = that object, `name = "Alice"`, `age = 19`, and hands the object back to you.
> - `student1.name`: the dot means "the `name` attribute of **this** object".
>
> **Output:**
> ```
> Alice 19
> Parsa 20
> Alice is 19 years old.
> Parsa is 20 years old.
> ```
>
> **Traps**
> - If you write `name = name` (without `self.`), you only make a local variable that disappears at the end of `__init__`. Later, `student1.name` gives an `AttributeError` (Week 5, D4).
> - Don't forget `self` in the brackets of `def __init__`. You **don't send it** when you create an object; Python sends it itself: `Student("Alice", 19)` has only two arguments.
> - `__init__` has two underscores **on each side**. `_init_` with one underscore never runs at all.
>

## Task 2: Dog Class

Create a class `Dog` with `name` and `age`, and two methods:
- `bark()` prints "<name> says Woof!"
- `birthday()` increases the dog's age by 1

Create a dog, call `bark()`, call `birthday()` twice and print the new age.

````python
class Dog:
    """A dog with a name and an age."""

    def __init__(self, name, age):
        """Store the dog's name and age."""
        self.name = name
        self.age = age

    def bark(self):
        """Print this dog's bark."""
        print(f"{self.name} says Woof!")

    def birthday(self):
        """Make this dog one year older."""
        self.age += 1


my_dog = Dog("Rex", 3)
my_dog.bark()

print(f"{my_dog.name} is {my_dog.age} years old.")
my_dog.birthday()
my_dog.birthday()
print(f"After two birthdays, {my_dog.name} is {my_dog.age} years old.")
````

> 💡 **Explanation**
>
> **Line by line**
> - `def bark(self):` is a **method**: a function defined **inside** a class. Its first parameter is always `self`, meaning "the dog this method was called on".
> - `print(f"{self.name} says Woof!")`: uses `self.name`, so each dog says its own name.
> - `def birthday(self): self.age += 1`: this method **changes the object itself** (Week 5, D8). Because the number is stored on `self`, the change stays after the method ends. That's why it doesn't need a `return`.
> - `my_dog.bark()`: behind the scenes, Python runs this: `Dog.bark(my_dog)`. So `self` becomes `my_dog`.
>
> **Output:**
> ```
> Rex says Woof!
> Rex is 3 years old.
> After two birthdays, Rex is 5 years old.
> ```
> **Hand check:** 3 → 4 → 5 ✓
>
> **Traps**
> - `my_dog.bark` without brackets **does not run**. It only points to the method itself. Always add `()`.
> - If you write `age += 1` in `birthday` (without `self.`), you get an `UnboundLocalError`, because Python looks for a local variable called `age`.
> - `bark()` prints and returns nothing. So `print(my_dog.bark())` also prints an extra `None`.
>
> **Try it yourself:** make a second dog and show that calling `birthday()` on the first dog does **not** change the second dog's age.

## Task 3: Rectangle Class

Create a class `Rectangle` with `width` and `height` and methods `area()` and `perimeter()`. Create a list of 4 rectangles and print the area and perimeter of each, then print which rectangle has the largest area.

````python
class Rectangle:
    """A rectangle with a width and a height."""

    def __init__(self, width, height):
        """Store the width and height."""
        self.width = width
        self.height = height

    def area(self):
        """Return the area of this rectangle."""
        return self.width * self.height

    def perimeter(self):
        """Return the perimeter of this rectangle."""
        return 2 * (self.width + self.height)


rectangles = [
    Rectangle(3, 4),
    Rectangle(5, 2),
    Rectangle(6, 6),
    Rectangle(2, 8),
]

for number, rectangle in enumerate(rectangles, start=1):
    print(f"Rectangle {number}: {rectangle.width} x {rectangle.height} "
          f"-> area {rectangle.area()}, perimeter {rectangle.perimeter()}")

# Find the largest area (the "best so far" pattern)
largest = rectangles[0]
largest_number = 1
for number, rectangle in enumerate(rectangles, start=1):
    if rectangle.area() > largest.area():
        largest = rectangle
        largest_number = number

print(f"Largest: Rectangle {largest_number} ({largest.width} x {largest.height}) "
      f"with area {largest.area()}")
````

> 💡 **Explanation**
>
> **Line by line**
> - `area` and `perimeter` **don't print** anything; they **return** a value. Why? Because later we want to **compare** the areas with each other. With `print` we couldn't (the Week 3 lesson: return, don't print).
> - `rectangles = [Rectangle(3, 4), ...]`: a **list of objects** (Week 5, D9). Exactly like the list of dicts in Week 4, but now each item can also **do** something: `rectangle.area()`.
> - `for number, rectangle in enumerate(rectangles, start=1):` the same `enumerate` as Week 2 (D15): it gives both a number and the object itself. `start=1` means count from 1.
> - Two f-strings one after the other inside brackets: Python joins them together by itself. This is only so the line doesn't get too long.
> - **The "best so far" pattern:** we assume `largest` is the first rectangle, then any rectangle with a bigger area takes its place. This is the same max-finding pattern from Weeks 2 and 4, but this time we keep **the whole object**, not just the number.
>
> **Output:**
> ```
> Rectangle 1: 3 x 4 -> area 12, perimeter 14
> Rectangle 2: 5 x 2 -> area 10, perimeter 14
> Rectangle 3: 6 x 6 -> area 36, perimeter 24
> Rectangle 4: 2 x 8 -> area 16, perimeter 20
> Largest: Rectangle 3 (6 x 6) with area 36
> ```
> **Hand check:** 6×6 = 36 and its perimeter is 2×(6+6) = 24 ✓
>
> **Traps**
> - `rectangle.area` without `()` is not a number; it is the method itself. `rectangle.area > largest.area` gives a `TypeError`.
> - If you start with `largest = 0`, later you can't write `largest.width`, because `0` is a number, not a rectangle. That's why we started with `rectangles[0]`.
> - `>` means that if two rectangles have the same area, the **first** one stays the winner. With `>=` the last one would win.
>
> **Try it yourself:** write a method `is_square()` that returns `True` if `width == height`, and print the squares in the list.

## Task 4: Bank Account

Create a class `BankAccount` with `owner` and `balance` (default 0) and methods:
- `deposit(amount)`
- `withdraw(amount)` – refuse the withdrawal and print a message if there is not enough money
- `show_balance()`

Test it with several deposits and withdrawals, including one that should fail.

````python
class BankAccount:
    """A simple bank account with an owner and a balance."""

    def __init__(self, owner, balance=0):
        """Store the owner and the starting balance (0 if not given)."""
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        """Add money to the account."""
        if amount <= 0:
            print("Deposit refused: the amount must be more than 0.")
            return
        self.balance += amount
        print(f"Deposited {amount:.2f}. New balance: {self.balance:.2f}")

    def withdraw(self, amount):
        """Take money out, but only if there is enough."""
        if amount > self.balance:
            print(f"Withdrawal of {amount:.2f} refused: not enough money "
                  f"(balance is {self.balance:.2f}).")
            return
        self.balance -= amount
        print(f"Withdrew {amount:.2f}. New balance: {self.balance:.2f}")

    def show_balance(self):
        """Print the owner and the current balance."""
        print(f"{self.owner}'s balance: {self.balance:.2f}")


account = BankAccount("Parsa")
account.show_balance()

account.deposit(100)
account.deposit(50.5)
account.withdraw(30)
account.withdraw(500)          # should fail: not enough money
account.deposit(-20)           # should fail: not a real deposit
account.withdraw(120.5)
account.show_balance()

savings = BankAccount("Sara", 1000)
savings.show_balance()
````

> 💡 **Explanation**
>
> **Line by line**
> - `def __init__(self, owner, balance=0):` a **default value**, the same thing you saw for functions in Week 3. `BankAccount("Parsa")` means a balance of 0; `BankAccount("Sara", 1000)` means a balance of 1000.
> - `deposit` and `withdraw` **change the object itself**: `self.balance += amount` and `self.balance -= amount` (Week 5, D8). The account "remembers" how much money it has, even after the method ends.
> - `if amount > self.balance:` is the guard at the start of the method: we check the bad case first and **leave** the method with an empty `return`, so the line `self.balance -= amount` never runs.
> - `{amount:.2f}`: always shows money with two decimal places (Week 1).
> - `show_balance` only **reads**; it doesn't change anything.
>
> **Output:**
> ```
> Parsa's balance: 0.00
> Deposited 100.00. New balance: 100.00
> Deposited 50.50. New balance: 150.50
> Withdrew 30.00. New balance: 120.50
> Withdrawal of 500.00 refused: not enough money (balance is 120.50).
> Deposit refused: the amount must be more than 0.
> Withdrew 120.50. New balance: 0.00
> Parsa's balance: 0.00
> Sara's balance: 1000.00
> ```
> **Hand check:** 0 + 100 + 50.5 − 30 = 120.5. The 500 withdrawal is refused. Then 120.5 − 120.5 = 0 ✓ (Taking out **all** the money is allowed, because the condition is `>`, not `>=`.)
>
> **Traps**
> - If you forget the `return` after the error message, the program prints the "refused" message **and** still takes the money out, so the balance goes negative!
> - `account.balance = -999` from outside also works; Python doesn't stop you. The convention is to change the balance **only through the methods**, because that's where the rules (the balance check) are.
> - The task didn't ask for the `amount <= 0` check in `deposit`, but without it `deposit(-20)` would actually take money **out**. Be careful: `withdraw` has no such check, so in this code `withdraw(-50)` still **adds** 50 to the balance (try it!). Put the same `amount <= 0` check at the start of `withdraw` too. Think like a real bank.
>

## Task 5: Printing Objects with `__str__`

Add a `__str__` method to the `Student` class from Task 1 (e.g. "Alice (age 19)"). Create a list of 5 students and print them with a loop using `print(student)`.

````python
class Student:
    """A student with a name and an age."""

    def __init__(self, name, age):
        """Store the name and age on the new object."""
        self.name = name
        self.age = age

    def __str__(self):
        """Return how this student looks when printed."""
        return f"{self.name} (age {self.age})"


students = [
    Student("Alice", 19),
    Student("Parsa", 20),
    Student("Sara", 22),
    Student("Reza", 18),
    Student("Mina", 21),
]

for student in students:
    print(student)

# __str__ is also used by str() and inside f-strings
first = students[0]
print("First on the list: " + str(first))
print(f"Last on the list: {students[-1]}")
````

> 💡 **Explanation**
>
> **Line by line**
> - `def __str__(self):` another special method, with two underscores on each side, like `__init__`. You never call it directly yourself. Python calls it **by itself** whenever it needs to turn the object into text: in `print(student)`, in `str(student)` and in an f-string such as `{students[-1]}`.
> - `return f"{self.name} (age {self.age})"`: it must **return** a **string**. Not print it!
> - `for student in students: print(student)`: a list of objects (D9). Each time round, `print` calls `student.__str__()` behind the scenes.
>
> **Output:**
> ```
> Alice (age 19)
> Parsa (age 20)
> Sara (age 22)
> Reza (age 18)
> Mina (age 21)
> First on the list: Alice (age 19)
> Last on the list: Mina (age 21)
> ```
>
> **`__str__` compared with printing the attributes**
>
> | Without `__str__` | With `__str__` |
> |---|---|
> | `print(student)` prints something meaningless: `<__main__.Student object at 0x10a3f2d90>` (the object's address in memory) | `print(student)` → `Alice (age 19)` |
> | Everywhere you want to print, you must write `print(student.name, student.age)` | The text format is defined in **one place**, inside the class |
> | If you change the format, you must change every `print` | You only change `__str__` |
>
> Task 1 read the attributes one by one **from outside**. Here, **the object itself** knows how to show itself. This is the big idea of a class: data and its behaviour live together.
>
> **Traps**
> - If you write `print(...)` instead of `return` inside `__str__`, the method returns `None` and you get the error `TypeError: __str__ returned non-string (type NoneType)`.
> - `"First: " + first` doesn't work (`TypeError`), because `first` is an object, not a string. You must write `str(first)`, or use an f-string.
> - The name is exactly `__str__`. Python doesn't recognise `__string__` or `_str_` and quietly ignores them.
>
> **Try it yourself:** with a loop and an `if`, `print(student)` only for the students older than 20.

## Task 6: Library Books

Create a class `Book` with `title`, `author` and `is_available` (default `True`) and methods `borrow()` and `return_book()`. Store 4 books in a list, then:
1. Borrow one book and try to borrow it a second time (print a message).
2. Print all books that are currently available.
3. Ask the user for an author and print all of that author's books.

````python
class Book:
    """A library book that can be borrowed and returned."""

    def __init__(self, title, author, is_available=True):
        """Store the book details; a new book is available by default."""
        self.title = title
        self.author = author
        self.is_available = is_available

    def borrow(self):
        """Borrow the book if it is on the shelf."""
        if self.is_available:
            self.is_available = False
            print(f"You borrowed '{self.title}'.")
        else:
            print(f"Sorry, '{self.title}' is already borrowed.")

    def return_book(self):
        """Put the book back on the shelf."""
        if self.is_available:
            print(f"'{self.title}' was not borrowed.")
        else:
            self.is_available = True
            print(f"Thank you for returning '{self.title}'.")

    def __str__(self):
        """Return the book as 'title by author'."""
        return f"{self.title} by {self.author}"


books = [
    Book("1984", "George Orwell"),
    Book("Animal Farm", "George Orwell"),
    Book("The Little Prince", "Antoine de Saint-Exupery"),
    Book("The Blind Owl", "Sadegh Hedayat"),
]

# 1. Borrow a book, then try again
print("--- Borrowing ---")
books[0].borrow()
books[0].borrow()

# 2. Show the books that are on the shelf
print("--- Available books ---")
for book in books:
    if book.is_available:
        print(book)

# 3. Search by author (not case-sensitive)
print("--- Search by author ---")
wanted = input("Enter an author: ").strip()
found = 0
for book in books:
    if book.author.lower() == wanted.lower():
        if book.is_available:
            status = "available"
        else:
            status = "borrowed"
        print(f"{book} ({status})")
        found += 1

if found == 0:
    print(f"No books by {wanted} in the library.")
````

> 💡 **Explanation**
>
> **Line by line**
> - `is_available=True` in `__init__`: a default, like `balance=0` in Task 4. So `Book("1984", "George Orwell")` is automatically available.
> - `borrow()` **changes the object itself**: `self.is_available = False` (D8). The second time, `if self.is_available:` is now `False`, so it goes into the `else` and prints the "already borrowed" message. In other words, the object "remembers" that it has been lent out.
> - `return_book()` is the opposite. Its name is `return_book`, not `return`, because `return` is a reserved Python word (keyword) and can't be a method name.
> - The task didn't ask for `__str__`, but I added it so `print(book)` gives clean text (the same lesson as Task 5).
> - `books[0].borrow()`: the first item in the list is a `Book` object, so you can call its method directly.
> - **Part 2:** a loop over the **list of objects** + `if book.is_available:`. The same filtering pattern as Week 4, but this time we read the attribute with a dot.
> - **Part 3:** `input(...).strip()` removes extra spaces at the start and end. `.lower()` on **both sides** means `george orwell` is found too.
> - `found` is a counter (the Week 2 pattern), so that if no book is found we give a proper message, instead of printing nothing.
>
> **Output** (if the user types `george orwell`):
> ```
> --- Borrowing ---
> You borrowed '1984'.
> Sorry, '1984' is already borrowed.
> --- Available books ---
> Animal Farm by George Orwell
> The Little Prince by Antoine de Saint-Exupery
> The Blind Owl by Sadegh Hedayat
> --- Search by author ---
> Enter an author: george orwell
> 1984 by George Orwell (borrowed)
> Animal Farm by George Orwell (available)
> ```
> If they type `Rumi` → `No books by Rumi in the library.`
>
> **Traps**
> - `if book.is_available == "True":` is wrong. `is_available` is a **boolean (bool)**, not a string. Just write `if book.is_available:`.
> - If you put the "No books" `print(...)` **inside** the loop, it prints once for every book that doesn't match. The check on `found` must come **after** the loop.
> - Without `.lower()`, `george orwell` is not equal to `George Orwell` and nothing is found.
>
> **Try it yourself:** at the end of the program, call `books[0].return_book()` and print the list of available books again. Now `1984` should be back.

## Task 7: Shop Inventory

Create a class `Product` with `name`, `price` and `quantity`, and methods `total_value()` and `apply_discount(percent)` (changes the product's own price). Store 5 products in a list and:
1. Print the total value of the whole inventory.
2. Apply a 10% discount to every product and print the new prices.

````python
class Product:
    """A product in a shop, with a price and a quantity in stock."""

    def __init__(self, name, price, quantity):
        """Store the product details."""
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_value(self):
        """Return price x quantity for this product."""
        return self.price * self.quantity

    def apply_discount(self, percent):
        """Lower this product's own price by the given percent."""
        self.price = self.price * (1 - percent / 100)


inventory = [
    Product("Laptop", 800.00, 3),
    Product("Mouse", 20.00, 25),
    Product("Keyboard", 45.00, 10),
    Product("Monitor", 150.00, 4),
    Product("USB cable", 5.00, 40),
]

# 1. Total value of the whole inventory
total = 0
for product in inventory:
    print(f"{product.name}: {product.quantity} x {product.price:.2f} = {product.total_value():.2f}")
    total += product.total_value()
print(f"Total inventory value: {total:.2f}")

# 2. 10% off everything
print()
print("After a 10% discount:")
for product in inventory:
    old_price = product.price
    product.apply_discount(10)
    print(f"{product.name}: {old_price:.2f} -> {product.price:.2f}")

new_total = 0
for product in inventory:
    new_total += product.total_value()
print(f"New total inventory value: {new_total:.2f}")
````

> 💡 **Explanation**
>
> **Line by line**
> - `total_value()` **returns** a value, because outside the class we want to add them together.
> - `apply_discount(percent)` **changes the object itself** (D8): `self.price = ...`. It returns nothing; the new price stays on the product.
> - `1 - percent / 100`: a 10% discount means "0.9 times". `10 / 100 = 0.1` and `1 - 0.1 = 0.9`. The same trick as the Week 1 tip calculator, only the other way round.
> - `total = 0` and `total += product.total_value()`: the Week 2 **accumulator** pattern, on a list of objects.
> - We keep `old_price = product.price` **before** the discount, because after it the old price is gone.
> - The empty `print()` just adds a blank line so the output is easy to read.
>
> **Output:**
> ```
> Laptop: 3 x 800.00 = 2400.00
> Mouse: 25 x 20.00 = 500.00
> Keyboard: 10 x 45.00 = 450.00
> Monitor: 4 x 150.00 = 600.00
> USB cable: 40 x 5.00 = 200.00
> Total inventory value: 4150.00
>
> After a 10% discount:
> Laptop: 800.00 -> 720.00
> Mouse: 20.00 -> 18.00
> Keyboard: 45.00 -> 40.50
> Monitor: 150.00 -> 135.00
> USB cable: 5.00 -> 4.50
> New total inventory value: 3735.00
> ```
> **Hand check:** 2400 + 500 + 450 + 600 + 200 = 4150. After the discount: 4150 × 0.9 = 3735 ✓
>
> **Traps**
> - `self.price * percent / 100` is only the **discount amount** (80), not the new price (720). You must either multiply by `1 - percent/100`, or **subtract** the discount amount.
> - If you run the discount loop twice, the discount is applied **twice** (800 → 720 → 648), because the method really changes the object. This is how changing methods behave, so be careful.
> - Decimal numbers (floats) sometimes come out as something like `40.49999999`. That's why you should always print with `:.2f`.
>
> **Try it yourself:** write a method `is_low_stock()` that returns `True` if `quantity < 5`, and print the products that need to be ordered again.

## Task 8: Cyber: Login Attempt Tracker

Create a class `Account` with `username`, `password` and `failed_attempts`. Add a method `login(password)` that returns `True` for the correct password and resets the failed attempts, otherwise increases `failed_attempts`. After 3 failed attempts the account becomes locked and every later login is refused. Test the class with correct and incorrect passwords.

````python
class Account:
    """A user account that locks after 3 failed logins."""

    MAX_ATTEMPTS = 3

    def __init__(self, username, password):
        """Store the login details and start with 0 failed attempts."""
        self.username = username
        self.password = password
        self.failed_attempts = 0

    def is_locked(self):
        """Return True if the account has too many failed attempts."""
        return self.failed_attempts >= Account.MAX_ATTEMPTS

    def login(self, password):
        """Try to log in; return True on success, False otherwise."""
        if self.is_locked():
            print(f"  {self.username}: account LOCKED, login refused.")
            return False
        if password == self.password:
            self.failed_attempts = 0
            print(f"  {self.username}: login OK.")
            return True
        self.failed_attempts += 1
        print(f"  {self.username}: wrong password ({self.failed_attempts}/{Account.MAX_ATTEMPTS}).")
        if self.is_locked():
            print(f"  {self.username}: too many failed attempts, account is now locked.")
        return False


user = Account("parsa", "S3cure!pass")

print("Test 1: correct password")
print(user.login("S3cure!pass"))

print("Test 2: one wrong, then correct (counter resets)")
print(user.login("hello123"))
print(user.login("S3cure!pass"))
print("Failed attempts now:", user.failed_attempts)

print("Test 3: three wrong passwords")
for guess in ["password", "123456", "qwerty"]:
    print(user.login(guess))

print("Test 4: correct password after the lock")
print(user.login("S3cure!pass"))
print("Locked?", user.is_locked())

print("Test 5: another user is not affected")
other = Account("sara", "tulip99")
print(other.login("tulip99"))
````

> 💡 **Explanation**
>
> **Line by line**
> - `MAX_ATTEMPTS = 3`: a **class attribute**, like `Counted.count` in Week 5, D10. One number shared by **all** accounts. Instead of writing the number `3` in three places in the code, we define it in one place.
> - `self.failed_attempts = 0`: an **instance attribute**. Each account has **its own** counter (Test 5 shows this).
> - `is_locked()` only **reads**, and returns a comparison, which is itself `True` or `False`. There's no need for `if ... return True else return False` (like C2 in Week 5).
> - **The order of the checks in `login` is very important:**
>   1. **First** we check whether it is locked. If it is locked, even the correct password is refused.
>   2. Then, correct password → counter back to zero, `return True`.
>   3. Otherwise, counter `+= 1` (a method that changes the object, D8) and `return False`.
> - `login` uses `self.is_locked()`: one method calls another method of the same object with `self.` (D7).
>
> **Output:**
> ```
> Test 1: correct password
>   parsa: login OK.
> True
> Test 2: one wrong, then correct (counter resets)
>   parsa: wrong password (1/3).
> False
>   parsa: login OK.
> True
> Failed attempts now: 0
> Test 3: three wrong passwords
>   parsa: wrong password (1/3).
> False
>   parsa: wrong password (2/3).
> False
>   parsa: wrong password (3/3).
>   parsa: too many failed attempts, account is now locked.
> False
> Test 4: correct password after the lock
>   parsa: account LOCKED, login refused.
> False
> Locked? True
> Test 5: another user is not affected
>   sara: login OK.
> True
> ```
>
> **Traps**
> - If you put the lock check **after** the password check, someone whose account is locked can open it with the correct password, and the lock becomes useless.
> - `self.failed_attempts = 0` should only be in the **success** branch. If you put it anywhere else, the counter never reaches 3.
> - The parameter `password` (the password the user typed) is different from `self.password` (the stored password). If you write `password == password` by mistake, it is always `True`!
>
> **Security note:** this is exactly the real **account lockout** that stops people guessing passwords (brute force). In a real system, passwords are never stored as plain text; a **hash** of the password is kept. Here we used plain text to keep it simple.

## Task 9: Gradebook

Create a class `Course` that stores a course name and a list of `Student` objects (each with a name and a list of marks). Add methods `add_student()`, `average_mark(student_name)` and `top_student()`. Create a course with at least 4 students and print each student's average and the top student.

````python
class Student:
    """A student with a name and a list of marks."""

    def __init__(self, name, marks):
        """Store the name and the list of marks."""
        self.name = name
        self.marks = marks

    def average(self):
        """Return the average of this student's marks (0 if there are none)."""
        if len(self.marks) == 0:
            return 0
        return sum(self.marks) / len(self.marks)

    def __str__(self):
        """Return the student's name and average."""
        return f"{self.name}: {self.average():.1f}"


class Course:
    """A course that holds a list of Student objects."""

    def __init__(self, name):
        """Store the course name and start with no students."""
        self.name = name
        self.students = []

    def add_student(self, student):
        """Add one Student object to this course."""
        self.students.append(student)

    def average_mark(self, student_name):
        """Return the average for the named student, or None if not found."""
        for student in self.students:
            if student.name == student_name:
                return student.average()
        return None

    def top_student(self):
        """Return the Student with the highest average (None if empty)."""
        if len(self.students) == 0:
            return None
        best = self.students[0]
        for student in self.students:
            if student.average() > best.average():
                best = student
        return best


course = Course("CST1510 Programming")
course.add_student(Student("Ali", [65, 70, 80]))
course.add_student(Student("Sara", [90, 85, 92]))
course.add_student(Student("Reza", [55, 60, 58]))
course.add_student(Student("Mina", [78, 88, 84]))

print(f"Course: {course.name}")
for student in course.students:
    print(f"  {student.name}: marks {student.marks}, average {course.average_mark(student.name):.1f}")

top = course.top_student()
print(f"Top student: {top.name} with an average of {top.average():.1f}")

# Looking up a name that is not on the course
result = course.average_mark("Bob")
if result is None:
    print("Bob is not on this course.")
````

> 💡 **Explanation**
>
> **Main idea: an object inside an object.** Each `Course` has an attribute `self.students`, which is a **list**, and each item in that list is a `Student` object, which itself has a list of `marks`. This is called **composition**: one object holds other objects (the same as in the Blackjack stretch task).
>
> ```
> course ──► name = "CST1510 Programming"
>        └─► students = [ Student(Ali, [65,70,80]), Student(Sara, [...]), ... ]
> ```
>
> **Line by line**
> - `Student.average()`: `sum(...) / len(...)` from Week 4. The `len(self.marks) == 0` check prevents a `ZeroDivisionError` for a student who has no marks yet.
> - `self.students = []` in the course's `__init__`: each course starts with **its own empty** list (instance attribute).
> - `add_student(self, student)`: takes an **object** (not just a name) and `append`s it. So this method also **changes the object itself** (D8); this time it makes the list inside it bigger.
> - `average_mark(student_name)`: the **search** pattern: loop over the list; if the name matches, `return` **right there**. If the loop ends and nothing was found, `return None`, meaning "there's nothing".
> - `top_student()`: the same "best so far" pattern as Task 3. It returns **the whole** `Student` object, not just the name, so outside we can both read `top.name` and call `top.average()`.
> - `if result is None:` is the correct way to check for `None`.
>
> **Output:**
> ```
> Course: CST1510 Programming
>   Ali: marks [65, 70, 80], average 71.7
>   Sara: marks [90, 85, 92], average 89.0
>   Reza: marks [55, 60, 58], average 57.7
>   Mina: marks [78, 88, 84], average 83.3
> Top student: Sara with an average of 89.0
> Bob is not on this course.
> ```
> **Hand check:** Sara: 90 + 85 + 92 = 267, divided by 3 = 89.0 ✓ Ali: 215 ÷ 3 = 71.66… → `71.7` ✓
>
> **Traps**
> - `return None` must come **after** the loop (lined up with `for`), not inside it in an `else`. If you put it in an `else`, then when the **first** student's name doesn't match, the method returns `None` immediately and never searches the rest of the list.
> - You **must** put `self.students = []` in `__init__`. If you write it outside the methods (as a class attribute), **all** courses share one list, and a student from one course also appears in another course! (The same difference between instance and class attributes as in D10.)
> - `course.add_student("Ali")` (just a string) later gives `AttributeError: 'str' object has no attribute 'name'` at `student.name`. You must send a `Student(...)` object.
>
> **Compared with Week 4:** with a dictionary you could write `{"Ali": [65, 70, 80], ...}`. But with a class, the **jobs** (the average, the best student) live next to the data too, and `course.top_student()` is much easier to read than a loop in the middle of the program.
