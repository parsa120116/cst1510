# Warm-up · fixing 3 broken files

# Week 5 · Solved lab, explained

This week you make your **own types** with `class`. Until now, a record was a dict: data only. A class bundles the **data** (attributes) *and* the **actions** on that data (methods) together.

**The picture to keep in mind:** a class is a **cookie cutter** (the shape); each object is a **cookie** made with it. `Record` is the cutter; `Record("R-001", 87.0, 100.0)` makes one cookie with its own data.

## Warm-up 1 · `01_missing_self.py`

```python
class Record:
    def status(percent):
        return percent

r = Record()
print(r.status(95))
```
```
TypeError: Record.status() takes 1 positional argument but 2 were given
```

**What goes wrong:** when you call `r.status(95)`, Python secretly passes the **object itself** as the first argument, so it really calls `Record.status(r, 95)`: **two** arguments. The method only has room for one.

**The fix:** every method's first parameter is `self` (the object), and your own parameters come after.

````python
class Record:
    def status(self, percent):
        return percent

r = Record()
print(r.status(95))
````

> 💡 **Explanation**
>
> Remember this sentence: **`r.status(95)` becomes `Record.status(r, 95)`**. `self` catches `r`, `percent` catches `95`. Forgetting `self` is the #1 beginner error with classes, and the error message ("takes 1 … but 2 were given") is the clue: the extra one is the object.

## Warm-up 2 · `02_attribute_error.py`

```python
self.amount = value    # stored as "amount", not "value"
...
print(r.value)
```
```
AttributeError: 'Record' object has no attribute 'value'
```

**What goes wrong:** the value was stored on the object under the name `amount`, but the last line asks for `value`. An **attribute** is a variable that belongs to an object, and it only exists under the name you gave it with `self.`

````python
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

r = Record("R-001", 87.0, 100.0)
print(r.value)
````

> 💡 **Explanation**
>
> `AttributeError` = "this object has no attribute with that name". Look at `__init__`: the names after `self.` are exactly the names you can use later with `r.`

## Warm-up 3 · `03_name_error.py`

```python
print(status())
```
```
NameError: name 'status' is not defined
```

**What goes wrong:** `status` is a method **inside** the class, not a normal function. It only exists *on objects*. You must call it on a specific record: `r.status()`.

````python
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def status(self):
        return "OVER LIMIT" if self.value > self.limit else "OK"

r = Record("R-001", 120.0, 100.0)
print(r.status())
````

> 💡 **Explanation**
>
> Without `r.`, Python can't know **which** record's status you want: there could be thousands of `Record` objects. The object before the dot is what becomes `self`.

## D1 · Define a class and create an object

````python
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

r1 = Record("R-001", 87.0, 100.0)
print(r1.label, r1.value, r1.limit)
````

> 💡 **Explanation**
>
> **Line by line**
> - `class Record:` starts the blueprint. Class names start with a **capital** letter (a convention).
> - `def __init__(self, label, value, limit):` is the special method that runs **automatically** when you create an object. "init" = initialise (set up). The double underscores mark it as special to Python.
> - `self.label = label`: the right side is the **parameter** (the value you passed in); the left side `self.label` stores it **on the object**, so it stays after `__init__` ends.
> - `r1 = Record("R-001", 87.0, 100.0)`: Python makes a new empty object, calls `__init__(that_object, "R-001", 87.0, 100.0)`, and gives you the object.
> - `r1.label`: the **dot** reads an attribute of the object.

> 💡 **Explanation**

````python
record_dict = {"label": "R-001", "value": 87.0}
record_obj = Record("R-001", 87.0, 100.0)

print(record_dict["label"])   # dict: square brackets and the key as a string
print(record_obj.label)       # object: a dot and the attribute name (no quotes)
# A dict can have any keys (or be missing some); every Record object is
# guaranteed to have label, value and limit, because __init__ sets them.
# And an object can also have methods (behaviour), which a dict cannot.
````

> 💡 **Explanation**
>
> | | dict | object |
> |---|---|---|
> | read | `d["label"]` | `obj.label` |
> | shape | any keys, any time | fixed by `__init__` |
> | actions | none | methods (`obj.status()`) |

## D4 · Fix the class that stores nothing

```python
def __init__(self, label, value, limit):
    label = label
    value = value
    limit = limit
```
**Why it fails:** `label = label` just assigns the local parameter to itself: a **local variable** (Week 3 scope!) that disappears when `__init__` ends. Nothing is stored on the object → `AttributeError` on `r.label`.

````python
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

r = Record("R-001", 87.0, 100.0)
print(r.label)
````

> 💡 **Explanation**
>
> The fix is three words: `self.` in front of each name on the left. That's also one of the mini-project rules: *attributes are set through `self` in `__init__`, nothing stored as a plain local variable*.

# Drills D5–D8 · methods

# Topic 2 · Methods

A **method** is a function defined **inside** a class. Its first parameter is `self`, so it can read (and change) the object's own attributes.

## D5 · Add a method

````python
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Return a short text summary of this record."""
        return f"{self.label}: {self.value}/{self.limit}"

r = Record("R-001", 87.0, 100.0)
print(r.summary())
````

> 💡 **Explanation**
>
> The method is **indented inside** the class (same level as `__init__`). It **returns** the string (Week 3: return, don't print, so the caller decides what to do with it). Inside the method, `self.label` is "the label of whichever record I was called on".

## D6 · `self` is whichever object made the call

````python
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def summary(self):
        """Return a short text summary of this record."""
        return f"{self.label}: {self.value}/{self.limit}"

    def status(self):
        """Return OVER LIMIT, WARNING or OK for this record."""
        percent = (self.value / self.limit) * 100
        if percent >= 100:
            return "OVER LIMIT"
        elif percent >= 90:
            return "WARNING"
        else:
            return "OK"

r1 = Record("R-001", 87.0, 100.0)
r2 = Record("R-002", 120.0, 100.0)
r3 = Record("R-003", 95.0, 100.0)
print(r1.status())
print(r2.status())
print(r3.status())
````

> 💡 **Explanation**
>
> The code of `status` exists **once**, in the class. When you call `r2.status()`, `self` is `r2`, so it uses 120.0 and 100.0 → `OVER LIMIT`. When you call `r3.status()`, `self` is `r3` → 95% → `WARNING`. Same method, different data each time.

## D7 · One method calls another

````python
class Record:
    """One record: a label, a value and its limit."""

    def __init__(self, label, value, limit):
        """Store the three fields on the new object."""
        self.label = label
        self.value = value
        self.limit = limit

    def percent(self):
        """Return the value as a percentage of the limit."""
        return (self.value / self.limit) * 100

    def status(self):
        """Return OVER LIMIT, WARNING or OK for this record."""
        p = self.percent()
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else:
            return "OK"

r = Record("R-001", 87.0, 100.0)
print(r.percent())
print(r.status())
````

> 💡 **Explanation**
>
> `status` now calls `self.percent()` instead of repeating the formula. Two reasons: (1) **no duplication**: the percentage is calculated in exactly one place; (2) `percent()` is useful on its own too.
>
> Inside a class, call another method of the same object through **`self.`**: writing just `percent()` would be a `NameError` (Warm-up 3's lesson again).

## D8 · A method can change its own object

````python
class Checker:
    def __init__(self):
        self.over_limit_count = 0

    def check(self, value, limit):
        """Add 1 to the count if value is over limit."""
        if value > limit:
            self.over_limit_count += 1

checker = Checker()
checker.check(87, 100)
checker.check(120, 100)
checker.check(130, 100)
print(checker.over_limit_count)
````

> 💡 **Explanation**
>
> - `__init__(self)` takes no other data: every new `Checker` just starts with `over_limit_count = 0`.
> - `self.over_limit_count += 1` **changes the object**, and the change **stays** after the method ends, because it's stored on the object (not a local variable). No `return` needed.
> - Calls: 87 (no), 120 (yes), 130 (yes) → **2**.
>
> This is how objects **remember** things over time: a bank account's balance, a game player's score, a login's failed attempts (C2).

## D9 · A list of objects

````python
class Record:
    """One record: a label, a value and its limit."""

    def __init__(self, label, value, limit):
        """Store the three fields on the new object."""
        self.label = label
        self.value = value
        self.limit = limit

    def percent(self):
        """Return the value as a percentage of the limit."""
        return (self.value / self.limit) * 100

    def status(self):
        """Return OVER LIMIT, WARNING or OK for this record."""
        p = self.percent()
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else:
            return "OK"


records = [
    Record("R-001", 87.0, 100.0),
    Record("R-002", 120.0, 100.0),
    Record("R-003", 45.0, 100.0),
]

for record in records:
    print(record.label, record.status())
````

> 💡 **Explanation**
>
> Exactly Week 4's list of dicts, but each item is now an **object**: `record.label` (dot) instead of `record["label"]`, and each item can **do** something: `record.status()`.

## D10 · A class-level counter

````python
class Counted:
    count = 0                    # class attribute: ONE value shared by the class

    def __init__(self):
        Counted.count += 1       # change the shared value through the class name

a = Counted()
b = Counted()
c = Counted()
print(Counted.count)
# The count is shared because it belongs to the CLASS, not to each object.
# There is only one Counted.count; every new object adds 1 to that same number.
````

> 💡 **Explanation**
>
> | | instance attribute | class attribute |
> |---|---|---|
> | created | `self.x = ...` in `__init__` | `x = 0` directly under `class` |
> | copies | one **per object** | **one**, shared |
> | change it with | `self.x = ...` | `Counted.x = ...` |
>
> Like a classroom: each student has their own name (instance), but the class has one shared headcount (class attribute). If you wrote `self.count += 1`, Python would create a separate `count` on each object, and the shared number would stay 0.

## D11 · Load objects from a file

````python
def load_records(path):
    """Read label,value,limit lines from a file and return a list of Record objects."""
    records = []
    with open(path) as f:
        for line in f:
            label, value, limit = line.strip().split(",")
            records.append(Record(label, float(value), float(limit)))
    return records

records = load_records("records_sample.txt")
for record in records:
    print(record.label, record.status())
````

> 💡 **Explanation**
>
> Week 4's `load_records`, with **one line changed**: instead of appending a dict, we append `Record(label, float(value), float(limit))`. The values must go in the same order as `__init__`'s parameters: label, value, limit.

## C1 · Count the records over the limit

````python
class Record:
    """One record: a label, a value and its limit."""

    def __init__(self, label, value, limit):
        """Store the three fields on the new object."""
        self.label = label
        self.value = value
        self.limit = limit

    def is_over(self):
        """Return True if the value is over the limit."""
        return self.value > self.limit


def load_records(path):
    """Read label,value,limit lines from a file and return a list of Record objects."""
    records = []
    with open(path) as f:
        for line in f:
            label, value, limit = line.strip().split(",")
            records.append(Record(label, float(value), float(limit)))
    return records


records = load_records("records_sample.txt")
over = 0
for record in records:
    if record.is_over():
        over += 1
print(f"{over} of {len(records)} records are OVER LIMIT")
````

> 💡 **Explanation**
>
> `is_over` returns a **comparison**, which is already `True` or `False`, so it's one line. No `if ... return True else return False` needed. Then the Week 2 counter pattern, and `len(records)` for the total → `1 of 3 records are OVER LIMIT`.

## C2 · An account that locks

````python
class Account:
    """A user account that locks after 3 failed logins."""

    def __init__(self, username):
        """Store the username and start with 0 failed logins."""
        self.username = username
        self.failed_logins = 0

    def fail(self):
        """Record one failed login."""
        self.failed_logins += 1

    def locked(self):
        """Return True if there have been 3 or more failed logins."""
        return self.failed_logins >= 3


anna = Account("anna")
anna.fail()
anna.fail()
anna.fail()
print(anna.locked())

ben = Account("ben")
ben.fail()
print(ben.locked())
````

> 💡 **Explanation**
>
> - The **state** (what the object remembers): `username` and `failed_logins` (starts at 0).
> - `fail()` **changes** the state (like D8).
> - `locked()` only **reads** the state and answers a question → `True`/`False`.
> - `anna` and `ben` are independent: anna's 3 fails don't affect ben (D2's lesson). → `True`, `False`.
>
> This is a real cyber-security pattern (account lockout against password guessing).

## Typical · add `status(self)` with a docstring, print it in each report

````python
class Record:
    """One server record: hostname, GB used and GB total."""

    def __init__(self, label, value, limit):
        """Store the three fields as attributes."""
        self.label = label
        self.value = value
        self.limit = limit

    def status(self):
        """Return OVER LIMIT (100%+), WARNING (90%+) or OK."""
        percent = (self.value / self.limit) * 100
        if percent >= 100:
            return "OVER LIMIT"
        elif percent >= 90:
            return "WARNING"
        else:
            return "OK"


records = [
    Record("srv-01", 87.0, 120.0),
    Record("srv-02", 125.0, 120.0),
    Record("srv-03", 110.0, 120.0),
]

for record in records:
    status = record.status()
    print()
    print("=" * 34)
    print(f"  RECORD CHECK  -  {record.label}")
    print("=" * 34)
    print(f"  Used        : {record.value:>10.2f}")
    print(f"  Total       : {record.limit:>10.2f}")
    print(f"  Status      : {status:>10}")
    print("=" * 34)
````

> 💡 **Explanation**
>
> Rule from the brief: *call `status()` once per record, store it, then use the variable*. That's `status = record.status()`. If you called `record.status()` twice (once to print, once to count), it would do the same work twice.

## Excellent · `load_records(path)` returns `Record` objects + OVER LIMIT count

On your Mac use the `servers.txt` from Week 4's SOLVED folder (copy it next to `template.py`). In the hub, this cell makes sure it exists:

````python
"""
RECORD CHECK  -  my version
Name  : Parsa
Lane  : IT
Run it:   python template.py
Data  :   servers.txt  (one record per line: hostname,GB used,GB total)
"""

# ===================================================================== CLASS
class Record:
    """One server record: hostname, GB used and GB total."""

    def __init__(self, label, value, limit):
        """Store the three fields as attributes."""
        self.label = label
        self.value = value
        self.limit = limit

    def percent(self):
        """Return the space used as a percentage of the total."""
        return (self.value / self.limit) * 100

    def status(self):
        """Return OVER LIMIT (100%+), WARNING (90%+) or OK."""
        p = self.percent()
        if p >= 100:
            return "OVER LIMIT"
        elif p >= 90:
            return "WARNING"
        else:
            return "OK"


# =================================================================== RECORDS
def load_records(path):
    """Read label,value,limit lines from a file and return a list of Record objects."""
    records = []
    with open(path) as f:
        for line in f:
            line = line.strip()
            if line == "":
                continue
            label, value, limit = line.split(",")
            records.append(Record(label, float(value), float(limit)))
    return records


records = load_records("servers.txt")

# ==================================================================== OUTPUT
over_limit_count = 0

for record in records:
    status = record.status()                 # call once, keep the result
    if status == "OVER LIMIT":
        over_limit_count += 1

    print()
    print("=" * 34)
    print(f"  RECORD CHECK  -  {record.label}")
    print("=" * 34)
    print(f"  Used        : {record.value:>10.2f}")
    print(f"  Total       : {record.limit:>10.2f}")
    print(f"  Percent     : {record.percent():>10.2f} %")
    print(f"  Status      : {status:>10}")
    print("=" * 34)

print(f"\n{over_limit_count} of {len(records)} record(s) OVER LIMIT.")
````

> 💡 **Explanation**

> - The loop: get the status **once**, count it if `OVER LIMIT`, print the report. The count is a plain variable outside the loop, which the template says is enough.
